## 모델 함수

<img align="left" src="https://raw.githubusercontent.com/greyhatguy007/Machine-Learning-Specialization-Coursera/d5c95ffb5b1524367e1a9147707ec4725298d715/C1%20-%20Supervised%20Machine%20Learning%20-%20Regression%20and%20Classification/week1/Optional%20Labs/images/C1_W1_L3_S1_model.png"     style=" width:380px; padding: 10px; " > 강의에서 설명한 것처럼 linear regression(`x`에서 `y`로 매핑하는 함수)에 대한 모델 함수는 다음과 같이 표현됩니다.

$$ f_{w,b}(x^{(i)}) = wx^{(i)} + b \tag{1}$$

위 공식은 직선을 표현하는 방법입니다. $w$와 $b$의 값에 따라 플롯에 다른 직선이 표시됩니다. <br/> <br/> <br/> <br/> <br/>

참고 표기: `m`는 예시 개수입니다. `i`는 예시를 선택합니다. `j`는 feature를 선택합니다. 예측값은 모델 출력을 의미합니다. 기울기는 매개변수가 변경될 때 비용이 어떻게 변경되는지를 의미합니다.

1. **예측값을 계산합니다**

하나의 feature에 대해 $\hat y = wx+b$; 여러 feature의 경우 $\hat y = \sum_j x_j w_j+b$. `i`로 인덱싱된 현재 예제를 사용하세요.

**함수 참조:** [내적](https://numpy.org/doc/stable/reference/generated/numpy.dot.html), [열 평균 및 축](https://numpy.org/doc/stable/reference/generated/numpy.mean.html), [지수](https://numpy.org/doc/stable/reference/generated/numpy.exp.html). 아래 원본 실습에는 전체 수학적 맥락이 포함되어 있습니다.

출처: [C1_W1_Lab03_Model_Representation_Soln.ipynb](https://github.com/greyhatguy007/Machine-Learning-Specialization-Coursera/blob/d5c95ffb5b1524367e1a9147707ec4725298d715/C1%20-%20Supervised%20Machine%20Learning%20-%20Regression%20and%20Classification/week1/Optional%20Labs/C1_W1_Lab03_Model_Representation_Soln.ipynb)

```python
def compute_model_output(x, w, b):
    """
    Computes the prediction of a linear model
    Args:
      x (ndarray (m,)): Data, m examples
      w,b (scalar)    : model parameters
    Returns
      y (ndarray (m,)): target values
    """
    m = x.shape[0]
    f_wb = np.zeros(m)
    for i in range(m):
        f_wb[i] = b + x[i] * w

    return f_wb
```

## Cost 계산
이 과제에서 '비용'이라는 용어는 데이터가 주택 비용이므로 약간 혼란스러울 수 있습니다. 여기서 비용은 모델이 집의 target 가격을 얼마나 잘 예측하는지 측정하는 것입니다. 주택 데이터에는 '가격'이라는 용어가 사용됩니다.

변수가 하나인 비용 방정식은 다음과 같습니다.
  $$J(w,b) = \frac{1}{2m} \sum\limits_{i = 0}^{m-1} (f_{w,b}(x^{(i)}) - y^{(i)})^2 \tag{1}$$

어디서
  $$f_{w,b}(x^{(i)}) = wx^{(i)} + b \tag{2}$$
  
- $f_{w,b}(x^{(i)})$는 예측값입니다(예: $w,b$ 매개변수를 사용하는 $i$).  
- $(f_{w,b}(x^{(i)}) -y^{(i)})^2$는 target 값과 예측값 사이의 제곱 차이입니다.   
- 이러한 차이는 모든 $m$ 예에 대해 합산되고 `2m`로 나누어 비용 $J(w,b)$를 생성합니다.  
>강의에서 합산하는 인덱스 범위는 일반적으로 1에서 m까지이고 코드는 0에서 m-1까지입니다.

아래 코드는 각 예를 반복하여 비용을 계산합니다. 각 루프에서:
- `f_wb`, 예측값가 계산됩니다.
- target과 예측값의 차이를 계산하여 제곱합니다.
- 총 비용에 추가됩니다.

### 참조 솔루션: 한 번에 한 줄씩

아래의 각 연습은 다음 코드 셀의 한 줄에 하나의 표현식으로 해결됩니다. 번호가 매겨진 참조는 해당 표현식을 코드 순서대로 설명합니다. 학생용 버전과 비교할 때 주변 루프와 반환 문을 유지하세요.

참고 표기: `m`는 예시 개수입니다. `i`는 예시를 선택합니다. `j`는 feature를 선택합니다. 예측값은 모델 출력을 의미합니다. 기울기는 매개변수가 변경될 때 비용이 어떻게 변경되는지를 의미합니다.

1. **예측값을 계산합니다**

하나의 feature에 대해 $\hat y = wx+b$; 여러 feature의 경우 $\hat y = \sum_j x_j w_j+b$. `i`로 인덱싱된 현재 예제를 사용하세요.

2. **제곱 오차 계산**

$(\hat y-y)^2$. `f_wb`를 `y[i]`와 비교하세요. 제곱에는 `** 2`를 사용하세요.

3. **전체에 이 오류를 추가하세요**

새 합계 = 이전 `cost_sum` + 이 예의 `cost`.

4. **평균 비용 계산**

$J = \frac{1}{2m}\sum_i(\hat y_i-y_i)^2$. 누적된 오류를 `m`의 두 배로 나눕니다.

**함수 참조:** [내적](https://numpy.org/doc/stable/reference/generated/numpy.dot.html), [열 평균 및 축](https://numpy.org/doc/stable/reference/generated/numpy.mean.html), [지수](https://numpy.org/doc/stable/reference/generated/numpy.exp.html). 아래 원본 실습에는 전체 수학적 맥락이 포함되어 있습니다.

출처: [C1_W1_Lab04_Cost_function_Soln.ipynb](https://github.com/greyhatguy007/Machine-Learning-Specialization-Coursera/blob/d5c95ffb5b1524367e1a9147707ec4725298d715/C1%20-%20Supervised%20Machine%20Learning%20-%20Regression%20and%20Classification/week1/Optional%20Labs/C1_W1_Lab04_Cost_function_Soln.ipynb)

```python
def compute_cost(x, y, w, b):
    """
    Computes the cost function for linear regression.

    Args:
      x (ndarray (m,)): Data, m examples
      y (ndarray (m,)): target values
      w,b (scalar)    : model parameters

    Returns
        total_cost (float): The cost of using w,b as the parameters for linear regression
               to fit the data points in x and y
    """
    # number of training examples
    m = x.shape[0]

    cost_sum = 0
    for i in range(m):
        f_wb = b + x[i] * w
        cost = np.square(y[i] - f_wb)
        cost_sum = cost + cost_sum
    total_cost = cost_sum / (2.0 * m)

    return total_cost
```

```python
# Verify the reference implementation
np.testing.assert_allclose(compute_cost(np.array([1., 2.]), np.array([3., 5.]), 1., 0.), 3.25)
print("Check passed.")
```

### 참조 솔루션: 한 번에 한 줄씩

아래의 각 연습은 다음 코드 셀의 한 줄에 하나의 표현식으로 해결됩니다. 번호가 매겨진 참조는 해당 표현식을 코드 순서대로 설명합니다. 학생용 버전과 비교할 때 주변 루프와 반환 문을 유지하세요.

참고 표기: `m`는 예시 개수입니다. `i`는 예시를 선택합니다. `j`는 feature를 선택합니다. 예측값은 모델 출력을 의미합니다. 기울기는 매개변수가 변경될 때 비용이 어떻게 변경되는지를 의미합니다.

1. **예측값을 계산합니다**

하나의 feature에 대해 $\hat y = wx+b$; 여러 feature의 경우 $\hat y = \sum_j x_j w_j+b$. `i`로 인덱싱된 현재 예제를 사용하세요.

2. **합계에 제곱오차를 더합니다**

새로운 합계 = 기존 `cost` + $(\hat y_i-y_i)^2$. 이전 줄에서 예측값 변수를 사용합니다.

3. **평균 비용 계산**

$J = \frac{1}{2m}\sum_i(\hat y_i-y_i)^2$. 누적된 오류를 `m`의 두 배로 나눕니다.

**함수 참조:** [내적](https://numpy.org/doc/stable/reference/generated/numpy.dot.html), [열 평균 및 축](https://numpy.org/doc/stable/reference/generated/numpy.mean.html), [지수](https://numpy.org/doc/stable/reference/generated/numpy.exp.html). 아래 원본 실습에는 전체 수학적 맥락이 포함되어 있습니다.

출처: [C1_W1_Lab05_Gradient_Descent_Soln.ipynb](https://github.com/greyhatguy007/Machine-Learning-Specialization-Coursera/blob/d5c95ffb5b1524367e1a9147707ec4725298d715/C1%20-%20Supervised%20Machine%20Learning%20-%20Regression%20and%20Classification/week1/Optional%20Labs/C1_W1_Lab05_Gradient_Descent_Soln.ipynb)

```python
#Function to calculate the cost
def compute_cost(x, y, w, b):

    m = x.shape[0]
    cost = 0

    for i in range(m):
        f_wb = b + x[i] * w
        cost = cost + np.square(y[i] - f_wb)
    total_cost = cost / (2.0 * m)

    return total_cost
```

```python
# Verify the reference implementation
np.testing.assert_allclose(compute_cost(np.array([1., 2.]), np.array([3., 5.]), 1., 0.), 3.25)
print("Check passed.")
```

## Gradient descent 요약
지금까지 이 과정에서는 $f_{w,b}(x^{(i)})$를 예측하는 선형 모델을 개발했습니다.
$$f_{w,b}(x^{(i)}) = wx^{(i)} + b \tag{1}$$
linear regression에서는 예측값 $f_{w,b}(x^{(i)})$와 실제 데이터 $y^{(i)}$ 사이의 오류 측정을 최소화하여 매개변수 $w$,$b$에 맞게 입력 학습 데이터를 활용합니다. 측정값은 $cost$, $J(w,b)$라고 합니다. 교육에서는 모든 교육 샘플 $x^{(i)},y^{(i)}$에 대한 비용을 측정합니다.
$$J(w,b) = \frac{1}{2m} \sum\limits_{i = 0}^{m-1} (f_{w,b}(x^{(i)}) - y^{(i)})^2\tag{2}$$

강의에서 *gradient descent*는 다음과 같이 설명되었습니다.

$$\begin{align*} \text{repeat}&\text{ until convergence:} \; \lbrace \newline
\;  w &= w -  \alpha \frac{\partial J(w,b)}{\partial w} \tag{3}  \; \newline
 b &= b -  \alpha \frac{\partial J(w,b)}{\partial b}  \newline \rbrace
\end{align*}$$ 여기서 매개변수 $w$, $b$는 동시에 업데이트됩니다. 기울기은 다음과 같이 정의됩니다.
$$
\begin{align}
\frac{\partial J(w,b)}{\partial w}  &= \frac{1}{m} \sum\limits_{i = 0}^{m-1} (f_{w,b}(x^{(i)}) - y^{(i)})x^{(i)} \tag{4}\\
  \frac{\partial J(w,b)}{\partial b}  &= \frac{1}{m} \sum\limits_{i = 0}^{m-1} (f_{w,b}(x^{(i)}) - y^{(i)}) \tag{5}\\
\end{align}
$$

여기서 *동시에* 매개변수를 업데이트하기 전에 모든 매개변수에 대한 편미분를 계산한다는 의미입니다.

참고 표기: `m`는 예시 개수입니다. `i`는 예시를 선택합니다. `j`는 feature를 선택합니다. 예측값은 모델 출력을 의미합니다. 기울기는 매개변수가 변경될 때 비용이 어떻게 변경되는지를 의미합니다.

1. **예측값을 계산합니다**

하나의 feature에 대해 $\hat y = wx+b$; 여러 feature의 경우 $\hat y = \sum_j x_j w_j+b$. `i`로 인덱싱된 현재 예제를 사용하세요.

2. **이 예의 가중치 기울기를 계산합니다**

$(\hat y_i-y_i)x_i$. `f_wb`, `y[i]` 및 `x[i]`를 사용합니다.

3. **이 예의 편향 기울기를 계산합니다**

$\hat y_i-y_i$. `f_wb` 및 `y[i]`를 사용하십시오.

4. **가중치 기울기 평균을 구합니다**

누적된 `dj_dw`를 예시 개수 `m`로 나눕니다. 배열의 나눗셈은 모든 가중치에 대해 동시에 작동합니다.

5. **편향 기울기 평균화**

누적된 `dj_db`를 예제 `m`의 개수로 나눕니다.

**함수 참조:** [내적](https://numpy.org/doc/stable/reference/generated/numpy.dot.html), [열 평균 및 축](https://numpy.org/doc/stable/reference/generated/numpy.mean.html), [지수](https://numpy.org/doc/stable/reference/generated/numpy.exp.html). 아래 원본 실습에는 전체 수학적 맥락이 포함되어 있습니다.

출처: [C1_W1_Lab05_Gradient_Descent_Soln.ipynb](https://github.com/greyhatguy007/Machine-Learning-Specialization-Coursera/blob/d5c95ffb5b1524367e1a9147707ec4725298d715/C1%20-%20Supervised%20Machine%20Learning%20-%20Regression%20and%20Classification/week1/Optional%20Labs/C1_W1_Lab05_Gradient_Descent_Soln.ipynb)

```python
def compute_gradient(x, y, w, b):
    """
    Computes the gradient for linear regression
    Args:
      x (ndarray (m,)): Data, m examples
      y (ndarray (m,)): target values
      w,b (scalar)    : model parameters
    Returns
      dj_dw (scalar): The gradient of the cost w.r.t. the parameters w
      dj_db (scalar): The gradient of the cost w.r.t. the parameter b
     """

    # Number of training examples
    m = x.shape[0]
    dj_dw = 0
    dj_db = 0

    for i in range(m):
        f_wb = b + x[i] * w
        dj_dw_i = x[i] * (f_wb - y[i])
        dj_db_i = f_wb - y[i]
        dj_db += dj_db_i
        dj_dw += dj_dw_i
    dj_dw = dj_dw / float(m)
    dj_db = dj_db / float(m)

    return dj_dw, dj_db
```

```python
# Verify the reference implementation
test_dw, test_db = compute_gradient(np.array([1., 2.]), np.array([3., 5.]), 1., 0.)
np.testing.assert_allclose([test_dw, test_db], [-4., -2.5])
print("Check passed.")
```

아래의 각 연습은 다음 코드 셀의 한 줄에 하나의 표현식으로 해결됩니다. 번호가 매겨진 참조는 해당 표현식을 코드 순서대로 설명합니다. 학생용 버전과 비교할 때 주변 루프와 반환 문을 유지하세요.

**함수 참조:** [내적](https://numpy.org/doc/stable/reference/generated/numpy.dot.html), [열 평균 및 축](https://numpy.org/doc/stable/reference/generated/numpy.mean.html), [지수](https://numpy.org/doc/stable/reference/generated/numpy.exp.html). 아래 원본 실습에는 전체 수학적 맥락이 포함되어 있습니다.

```python
def gradient_descent(x, y, w_in, b_in, alpha, num_iters, cost_function, gradient_function):
    """
    Performs gradient descent to fit w,b. Updates w,b by taking
    num_iters gradient steps with learning rate alpha

    Args:
      x (ndarray (m,))  : Data, m examples
      y (ndarray (m,))  : target values
      w_in,b_in (scalar): initial values of model parameters
      alpha (float):     Learning rate
      num_iters (int):   number of iterations to run gradient descent
      cost_function:     function to call to produce cost
      gradient_function: function to call to produce gradient

    Returns:
      w (scalar): Updated value of parameter after running gradient descent
      b (scalar): Updated value of parameter after running gradient descent
      J_history (List): History of cost values
      p_history (list): History of parameters [w,b]
      """

    w = copy.deepcopy(w_in) # avoid modifying global w_in
    # An array to store cost J and w's at each iteration primarily for graphing later
    J_history = []
    p_history = []
    b = b_in
    w = w_in

    for i in range(num_iters):
        # Calculate the gradient and update the parameters using gradient_function
        dj_dw, dj_db = gradient_function(x, y, w , b)

        # Update Parameters using equation (3) above
        b = b - dj_db * alpha
        w = w - dj_dw * alpha

        # Save cost J at each iteration
        if i<100000:      # prevent resource exhaustion
            J_history.append( cost_function(x, y, w , b))
            p_history.append([w,b])
        # Print cost every at intervals 10 times or as many iterations if < 10
        if i% math.ceil(num_iters/10) == 0:
            print(f"Iteration {i:4}: Cost {J_history[-1]:0.2e} ",
                  f"dj_dw: {dj_dw: 0.3e}, dj_db: {dj_db: 0.3e}  ",
                  f"w: {w: 0.3e}, b:{b: 0.5e}")

    return w, b, J_history, p_history #return w and J,w history for graphing
```

```python
# Verify the reference implementation
test_w, test_b, *_ = gradient_descent(np.array([1., 2.]), np.array([3., 5.]), 1., 0., 0.1, 1, compute_cost, compute_gradient)
np.testing.assert_allclose([test_w, test_b], [1.4, 0.25])
print("Check passed.")
```

```python
# initialize parameters
w_init = 0
b_init = 0
# some gradient descent settings
iterations = 10000
tmp_alpha = 1.0e-2
# run gradient descent
w_final, b_final, J_hist, p_hist = gradient_descent(x_train ,y_train, w_init, b_init, tmp_alpha,
                                                    iterations, compute_cost, compute_gradient)
print(f"(w,b) found by gradient descent: ({w_final:8.4f},{b_final:8.4f})")
```

### Learning Rate 증가

<figure>
 <img align="left", src="https://raw.githubusercontent.com/greyhatguy007/Machine-Learning-Specialization-Coursera/d5c95ffb5b1524367e1a9147707ec4725298d715/C1%20-%20Supervised%20Machine%20Learning%20-%20Regression%20and%20Classification/week1/Optional%20Labs/images/C1_W1_Lab03_alpha_too_big.PNG"   style="width:340px;height:240px;" >
</figure>
이번 강의에서는 식(3)에서 learning rate, $\alpha$의 적정값에 대한 논의가 있었습니다. $\alpha$가 클수록 gradient descent가 솔루션으로 수렴되는 속도가 빨라집니다. 그러나 너무 크면 gradient descent가 발산합니다. 위에는 훌륭하게 수렴되는 솔루션의 예가 있습니다.

$\alpha$의 값을 늘려서 무슨 일이 일어나는지 살펴보겠습니다.

```python
# initialize parameters
w_init = 0
b_init = 0
# set alpha to a large value
iterations = 10
tmp_alpha = 8.0e-1
# run gradient descent
w_final, b_final, J_hist, p_hist = gradient_descent(x_train ,y_train, w_init, b_init, tmp_alpha,
                                                    iterations, compute_cost, compute_gradient)
```

위에서 $w$ 및 $b$는 각 반복마다 절대값이 증가하면서 양수와 음수 사이를 오가며 움직입니다. 또한 $\frac{\partial J(w,b)}{\partial w}$를 반복할 때마다 부호가 변경되고 비용은 감소하기보다는 증가합니다. 이는 *learning rate가 너무 크고* 발산하고 있다는 분명한 신호입니다. 이를 플롯으로 시각화해 보겠습니다.

```python
plt_divergence(p_hist, J_hist,x_train, y_train)
plt.show()
```

위의 왼쪽 그래프는 gradient descent의 처음 몇 단계에 대한 $w$의 진행 상황을 보여줍니다. $w$는 양수와 음수 사이를 진동하고 비용이 빠르게 증가합니다. Gradient Descent는 $w$와 $b$ 모두에서 동시에 작동하므로 전체 그림을 보려면 오른쪽에 3D 플롯이 필요합니다.
