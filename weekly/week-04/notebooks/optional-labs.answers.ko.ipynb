## 3.1 요약
이 과정에서 사용하게 될 <img align="right" src="https://raw.githubusercontent.com/greyhatguy007/Machine-Learning-Specialization-Coursera/d5c95ffb5b1524367e1a9147707ec4725298d715/C1%20-%20Supervised%20Machine%20Learning%20-%20Regression%20and%20Classification/week2/Optional%20Labs/images/C1_W2_Lab04_Vectors.PNG" style="width:340px;" >벡터는 순서가 지정된 숫자 배열입니다. 표기법에서 벡터는 $\mathbf{x}$와 같이 굵은 소문자로 표시됩니다.  벡터의 요소는 모두 동일한 유형입니다. 예를 들어 벡터에는 문자와 숫자가 모두 포함되지 않습니다. 수학자들은 *순위*를 선호하지만 배열의 요소 수는 종종 *차원*이라고 합니다. 표시된 벡터의 차원은 $n$입니다. 벡터의 요소는 인덱스로 참조될 수 있습니다. 수학 설정에서 인덱스는 일반적으로 1부터 n까지 실행됩니다. 컴퓨터 과학 및 이러한 실습에서 인덱싱은 일반적으로 0에서 n-1까지 실행됩니다.  표기법에서 벡터의 요소는 개별적으로 참조될 때 아래 첨자의 인덱스를 나타냅니다. 예를 들어 벡터 $\mathbf{x}$의 $0^{th}$ 요소는 $x_0$입니다. 이 경우 x는 굵게 표시되지 않습니다.

<a name="toc_40015_3.2"></a>
## 3.2 NumPy 어레이

NumPy의 기본 데이터 구조는 동일한 유형(`dtype`)의 요소를 포함하는 색인 ​​가능한 n차원 *배열*입니다. '차원'이라는 용어가 너무 많이 사용되었음을 바로 알 수 있습니다. 위에서는 벡터의 요소 수였지만 여기서 차원은 배열의 인덱스 수를 나타냅니다. 1차원 또는 1차원 배열에는 하나의 인덱스가 있습니다. 코스 1에서는 벡터를 NumPy 1차원 배열로 표현하겠습니다.

- 1차원 배열, 형태(n,): [0]부터 [n-1]까지 인덱스가 있는 n개 요소

<a name="toc_40015_3.3"></a>
## 3.3 벡터 생성

NumPy의 데이터 생성 루틴은 일반적으로 객체의 모양인 첫 번째 매개변수를 갖습니다. 이는 1차원 결과에 대한 단일 값이거나 결과의 모양을 지정하는 튜플(n,m,...)일 수 있습니다. 다음은 이러한 루틴을 사용하여 벡터를 생성하는 예입니다.

```python
# NumPy routines which allocate memory and fill arrays with value
a = np.zeros(4);                print(f"np.zeros(4) :   a = {a}, a shape = {a.shape}, a data type = {a.dtype}")
a = np.zeros((4,));             print(f"np.zeros(4,) :  a = {a}, a shape = {a.shape}, a data type = {a.dtype}")
a = np.random.random_sample(4); print(f"np.random.random_sample(4): a = {a}, a shape = {a.shape}, a data type = {a.dtype}")
```

일부 데이터 생성 루틴은 모양 튜플을 사용하지 않습니다.

```python
# NumPy routines which allocate memory and fill arrays with value but do not accept shape as input argument
a = np.arange(4.);              print(f"np.arange(4.):     a = {a}, a shape = {a.shape}, a data type = {a.dtype}")
a = np.random.rand(4);          print(f"np.random.rand(4): a = {a}, a shape = {a.shape}, a data type = {a.dtype}")
```

값을 수동으로 지정할 수도 있습니다.

```python
# NumPy routines which allocate memory and fill with user specified values
a = np.array([5,4,3,2]);  print(f"np.array([5,4,3,2]):  a = {a},     a shape = {a.shape}, a data type = {a.dtype}")
a = np.array([5.,4,3,2]); print(f"np.array([5.,4,3,2]): a = {a}, a shape = {a.shape}, a data type = {a.dtype}")
```

이것들은 모두 4개의 요소를 가진 1차원 벡터 `a`를 생성했습니다. `a.shape`는 치수를 반환합니다. 여기서는 4개 요소가 있는 1차원 배열을 나타내는 a.shape = `(4,)`를 볼 수 있습니다.

## 3.4 벡터 연산
벡터를 사용하여 몇 가지 작업을 살펴보겠습니다.
<a name="toc_40015_3.4.1"></a>
### 3.4.1 인덱싱
벡터의 요소는 인덱싱 및 슬라이싱을 통해 액세스할 수 있습니다. NumPy는 매우 완벽한 인덱싱 및 슬라이싱 기능 세트를 제공합니다. 여기서는 강좌에 필요한 기본 사항만 살펴보겠습니다. 자세한 내용은 [슬라이싱 및 인덱싱](https://NumPy.org/doc/stable/reference/arrays.indexing.html)을 참조하세요. **인덱싱**은 배열 내의 위치를 ​​기준으로 배열의 *요소*를 참조하는 것을 의미합니다. **슬라이싱**은 인덱스를 기반으로 배열에서 요소의 *하위 집합*을 가져오는 것을 의미합니다. NumPy는 0부터 인덱싱을 시작하므로 벡터 $\mathbf{a}$의 세 번째 요소는 `a[2]`입니다.

```python
#vector indexing operations on 1-D vectors
a = np.arange(10)
print(a)

#access an element
print(f"a[2].shape: {a[2].shape} a[2]  = {a[2]}, Accessing an element returns a scalar")

# access the last element, negative indexes count from the end
print(f"a[-1] = {a[-1]}")

#indexs must be within the range of the vector or they will produce and error
try:
    c = a[10]
except Exception as e:
    print("The error message you'll see is:")
    print(e)
```

<a name="toc_40015_3.4.2"></a>
### 3.4.2 슬라이싱
슬라이싱은 세 가지 값 세트(`start:stop:step`)를 사용하여 인덱스 배열을 생성합니다. 값의 하위 집합도 유효합니다. 그 사용법은 다음 예를 통해 가장 잘 설명됩니다.

```python
#vector slicing operations
a = np.arange(10)
print(f"a         = {a}")

#access 5 consecutive elements (start:stop:step)
c = a[2:7:1];     print("a[2:7:1] = ", c)

# access 3 elements separated by two
c = a[2:7:2];     print("a[2:7:2] = ", c)

# access all elements index 3 and above
c = a[3:];        print("a[3:]    = ", c)

# access all elements below index 3
c = a[:3];        print("a[:3]    = ", c)

# access all elements
c = a[:];         print("a[:]     = ", c)
```

<a name="toc_40015_3.4.3"></a>
### 3.4.3 단일 벡터 연산
단일 벡터에 대한 연산과 관련된 유용한 연산이 많이 있습니다.

```python
a = np.array([1,2,3,4])
print(f"a             : {a}")
# negate elements of a
b = -a
print(f"b = -a        : {b}")

# sum all elements of a, returns a scalar
b = np.sum(a)
print(f"b = np.sum(a) : {b}")

b = np.mean(a)
print(f"b = np.mean(a): {b}")

b = a**2
print(f"b = a**2      : {b}")
```

<a name="toc_40015_3.4.4"></a>
### 3.4.4 벡터 벡터 요소별 연산
NumPy 산술, 논리 및 비교 연산의 대부분은 벡터에도 적용됩니다. 이러한 연산자는 요소별로 작동합니다. 예를 들어
$$ \mathbf{a} + \mathbf{b} = \sum_{i=0}^{n-1} a_i + b_i $$

```python
a = np.array([ 1, 2, 3, 4])
b = np.array([-1,-2, 3, 4])
print(f"Binary operators work element wise: {a + b}")
```

물론 이것이 올바르게 작동하려면 벡터의 크기가 동일해야 합니다.

```python
#try a mismatched vector operation
c = np.array([1, 2])
try:
    d = a + c
except Exception as e:
    print("The error message you'll see is:")
    print(e)
```

누계 + $a_i b_i$. 누계의 이름은 `x`입니다. 그것을 추가하기 전에 쌍을 곱하십시오.

**함수 참조:** [내적](https://numpy.org/doc/stable/reference/generated/numpy.dot.html), [열 평균 및 축](https://numpy.org/doc/stable/reference/generated/numpy.mean.html), [지수](https://numpy.org/doc/stable/reference/generated/numpy.exp.html). 아래 원본 실습에는 전체 수학적 맥락이 포함되어 있습니다.

출처: [C1_W2_Lab01_Python_Numpy_Vectorization_Soln.ipynb](https://github.com/greyhatguy007/Machine-Learning-Specialization-Coursera/blob/d5c95ffb5b1524367e1a9147707ec4725298d715/C1%20-%20Supervised%20Machine%20Learning%20-%20Regression%20and%20Classification/week2/Optional%20Labs/C1_W2_Lab01_Python_Numpy_Vectorization_Soln.ipynb)

```python
def my_dot(a, b):
    """
   Compute the dot product of two vectors

    Args:
      a (ndarray (n,)):  input vector
      b (ndarray (n,)):  input vector with same dimension as a

    Returns:
      x (scalar):
    """
    x=0
    for i in range(a.shape[0]):
        x = x + b[i] * a[i]
    return x
```

### 3.4.7 속도의 필요성: 벡터 대 for 루프
우리는 속도 메모리 효율성을 향상시키기 위해 NumPy 라이브러리를 활용했습니다. 다음을 보여드리겠습니다.

```python
np.random.seed(1)
a = np.random.rand(10000000)  # very large arrays
b = np.random.rand(10000000)

tic = time.time()  # capture start time
c = np.dot(a, b)
toc = time.time()  # capture end time

print(f"np.dot(a, b) =  {c:.4f}")
print(f"Vectorized version duration: {1000*(toc-tic):.4f} ms ")

tic = time.time()  # capture start time
c = my_dot(a,b)
toc = time.time()  # capture end time

print(f"my_dot(a, b) =  {c:.4f}")
print(f"loop version duration: {1000*(toc-tic):.4f} ms ")

del(a);del(b)  #remove these big arrays from memory
```

따라서 이 예에서는 벡터화를 통해 속도가 크게 향상됩니다. 이는 NumPy가 기본 하드웨어에서 사용 가능한 데이터 병렬성을 더 잘 활용하기 때문입니다. GPU와 최신 CPU는 SIMD(Single Instruction, Multiple Data) 파이프라인을 구현하여 여러 작업을 병렬로 실행할 수 있습니다. 이는 데이터 세트가 매우 큰 경우가 많은 기계 학습에서 매우 중요합니다.

### 4.4.2 슬라이싱
슬라이싱은 세 가지 값 세트(`start:stop:step`)를 사용하여 인덱스 배열을 생성합니다. 값의 하위 집합도 유효합니다. 그 사용법은 다음 예를 통해 가장 잘 설명됩니다.

```python
#vector 2-D slicing operations
a = np.arange(20).reshape(-1, 10)
print(f"a = \n{a}")

#access 5 consecutive elements (start:stop:step)
print("a[0, 2:7:1] = ", a[0, 2:7:1], ",  a[0, 2:7:1].shape =", a[0, 2:7:1].shape, "a 1-D array")

#access 5 consecutive elements (start:stop:step) in two rows
print("a[:, 2:7:1] = \n", a[:, 2:7:1], ",  a[:, 2:7:1].shape =", a[:, 2:7:1].shape, "a 2-D array")

# access all elements
print("a[:,:] = \n", a[:,:], ",  a[:,:].shape =", a[:,:].shape)

# access all elements in one row (very common usage)
print("a[1,:] = ", a[1,:], ",  a[1,:].shape =", a[1,:].shape, "a 1-D array")
# same as
print("a[1]   = ", a[1],   ",  a[1].shape   =", a[1].shape, "a 1-D array")
```

# 3 다중 변수가 있는 모델 예측
여러 변수가 있는 모델의 예측값은 선형 모델에 의해 제공됩니다.

$$ f_{\mathbf{w},b}(\mathbf{x}) =  w_0x_0 + w_1x_1 +... + w_{n-1}x_{n-1} + b \tag{1}$$
또는 벡터 표기법으로:
$$ f_{\mathbf{w},b}(\mathbf{x}) = \mathbf{w} \cdot \mathbf{x} + b  \tag{2} $$
여기서 $\cdot$는 벡터 `dot product`입니다.

내적을 시연하기 위해 (1)과 (2)를 사용하여 예측값을 구현하겠습니다.

<a name="toc_15456_3.1"></a>
## 3.1 요소별 계산으로 예측
이전 예측값은 하나의 feature 값에 하나의 매개변수를 곱하고 편향 매개변수를 추가했습니다. 예측값의 이전 구현을 여러 feature로 직접 확장하는 것은 각 요소에 대한 루프를 사용하여 위의 (1)을 구현하고 해당 매개변수와 곱셈을 수행한 다음 끝에 편향 매개변수를 추가하는 것입니다.

### 참조 솔루션: 한 번에 한 줄씩

아래의 각 연습은 다음 코드 셀의 한 줄에 하나의 표현식으로 해결됩니다. 번호가 매겨진 참조는 해당 표현식을 코드 순서대로 설명합니다. 학생용 버전과 비교할 때 주변 루프와 반환 문을 유지하세요.

참고 표기: `m`는 예시 개수입니다. `i`는 예시를 선택합니다. `j`는 feature를 선택합니다. 예측값은 모델 출력을 의미합니다. 기울기는 매개변수가 변경될 때 비용이 어떻게 변경되는지를 의미합니다.

1. **이 feature 기여도를 계산하세요**

$x_i w_i$. 현재 feature `x[i]`와 가중치 `w[i]`를 사용하세요.

2. **이 feature 기여를 추가하세요**

새로운 예측값 총계 = 이전 `p` + 이번 feature의 `p_i`.

3. **예측값에 편향를 추가합니다**

최종 예측값 = 누적 가중치 feature 총계 `p` + 편향 `b`.

**함수 참조:** [내적](https://numpy.org/doc/stable/reference/generated/numpy.dot.html), [열 평균 및 축](https://numpy.org/doc/stable/reference/generated/numpy.mean.html), [지수](https://numpy.org/doc/stable/reference/generated/numpy.exp.html). 아래 원본 실습에는 전체 수학적 맥락이 포함되어 있습니다.

출처: [C1_W2_Lab02_Multiple_Variable_Soln.ipynb](https://github.com/greyhatguy007/Machine-Learning-Specialization-Coursera/blob/d5c95ffb5b1524367e1a9147707ec4725298d715/C1%20-%20Supervised%20Machine%20Learning%20-%20Regression%20and%20Classification/week2/Optional%20Labs/C1_W2_Lab02_Multiple_Variable_Soln.ipynb)

```python
def predict_single_loop(x, w, b):
    """
    single predict using linear regression

    Args:
      x (ndarray): Shape (n,) example with multiple features
      w (ndarray): Shape (n,) model parameters
      b (scalar):  model parameter

    Returns:
      p (scalar):  prediction
    """
    n = x.shape[0]
    p = 0
    for i in range(n):
        p_i = w[i] * x[i]
        p = p_i + p
    p = b + p
    return p
```

```python
# Verify the reference implementation
np.testing.assert_allclose(predict_single_loop(np.array([2., -1.]), np.array([3., 4.]), 5.), 7.)
print("Check passed.")
```

```python
# get a row from our training data
x_vec = X_train[0,:]
print(f"x_vec shape {x_vec.shape}, x_vec value: {x_vec}")

# make a prediction
f_wb = predict_single_loop(x_vec, w_init, b_init)
print(f"f_wb shape {np.shape(f_wb)}, prediction: {f_wb}")
```

`x_vec`의 모양을 확인하세요. 이는 4개 요소(4,)를 포함하는 1D NumPy 벡터입니다. 결과 `f_wb`는 스칼라입니다.

## 3.2 벡터 연산으로 예측

위의 방정식 (1)은 위의 (2)와 같이 내적을 사용하여 구현될 수 있습니다. 예측값 속도를 높이기 위해 벡터 연산을 사용할 수 있습니다.

Python/Numpy 실습에서 NumPy `np.dot()`[[링크](https://numpy.org/doc/stable/reference/generated/numpy.dot.html)]를 사용하여 벡터 내적을 수행할 수 있다는 점을 상기해 보세요.

### 참조 솔루션: 한 번에 한 줄씩

아래의 각 연습은 다음 코드 셀의 한 줄에 하나의 표현식으로 해결됩니다. 번호가 매겨진 참조는 해당 표현식을 코드 순서대로 설명합니다. 학생용 버전과 비교할 때 주변 루프와 반환 문을 유지하세요.

참고 표기: `m`는 예시 개수입니다. `i`는 예시를 선택합니다. `j`는 feature를 선택합니다. 예측값은 모델 출력을 의미합니다. 기울기는 매개변수가 변경될 때 비용이 어떻게 변경되는지를 의미합니다.

1. **모든 feature에서 예측값을 계산합니다**

$\hat y=\sum_j x_j w_j+b$. `np.dot`는 입력 벡터 `x`와 가중 벡터 `w`를 하나의 숫자로 결합합니다.

**함수 참조:** [내적](https://numpy.org/doc/stable/reference/generated/numpy.dot.html), [열 평균 및 축](https://numpy.org/doc/stable/reference/generated/numpy.mean.html), [지수](https://numpy.org/doc/stable/reference/generated/numpy.exp.html). 아래 원본 실습에는 전체 수학적 맥락이 포함되어 있습니다.

출처: [C1_W2_Lab02_Multiple_Variable_Soln.ipynb](https://github.com/greyhatguy007/Machine-Learning-Specialization-Coursera/blob/d5c95ffb5b1524367e1a9147707ec4725298d715/C1%20-%20Supervised%20Machine%20Learning%20-%20Regression%20and%20Classification/week2/Optional%20Labs/C1_W2_Lab02_Multiple_Variable_Soln.ipynb)

```python
def predict(x, w, b):
    """
    single predict using linear regression
    Args:
      x (ndarray): Shape (n,) example with multiple features
      w (ndarray): Shape (n,) model parameters
      b (scalar):             model parameter

    Returns:
      p (scalar):  prediction
    """
    p = float(x @ w) + b
    return p
```

```python
# Verify the reference implementation
np.testing.assert_allclose(predict(np.array([2., -1.]), np.array([3., 4.]), 5.), 7.)
print("Check passed.")
```

```python
# get a row from our training data
x_vec = X_train[0,:]
print(f"x_vec shape {x_vec.shape}, x_vec value: {x_vec}")

# make a prediction
f_wb = predict(x_vec,w_init, b_init)
print(f"f_wb shape {np.shape(f_wb)}, prediction: {f_wb}")
```

결과와 형태는 Looping을 사용한 이전 버전과 동일합니다. 앞으로는 `np.dot`가 이러한 작업에 사용될 것입니다. 예측값은 이제 단일 문입니다. 대부분의 루틴은 별도의 예측 루틴을 호출하는 대신 이를 직접 구현합니다.

# 4 여러 변수를 사용하여 비용 계산
여러 변수 $J(\mathbf{w},b)$가 있는 cost function의 방정식은 다음과 같습니다.
$$J(\mathbf{w},b) = \frac{1}{2m} \sum\limits_{i = 0}^{m-1} (f_{\mathbf{w},b}(\mathbf{x}^{(i)}) - y^{(i)})^2 \tag{3}$$
여기서:
$$ f_{\mathbf{w},b}(\mathbf{x}^{(i)}) = \mathbf{w} \cdot \mathbf{x}^{(i)} + b  \tag{4} $$


이전 실습과 달리 $\mathbf{w}$ 및 $\mathbf{x}^{(i)}$는 여러 feature를 지원하는 스칼라가 아닌 벡터입니다.

아래는 방정식 (3)과 (4)를 구현한 것입니다. 이는 모든 `m` 예제에 대한 for 루프가 사용되는 *이 과정*의 표준 패턴을 사용한다는 점에 유의하세요.

### 참조 솔루션: 한 번에 한 줄씩

아래의 각 연습은 다음 코드 셀의 한 줄에 하나의 표현식으로 해결됩니다. 번호가 매겨진 참조는 해당 표현식을 코드 순서대로 설명합니다. 학생용 버전과 비교할 때 주변 루프와 반환 문을 유지하세요.

참고 표기: `m`는 예시 개수입니다. `i`는 예시를 선택합니다. `j`는 feature를 선택합니다. 예측값은 모델 출력을 의미합니다. 기울기는 매개변수가 변경될 때 비용이 어떻게 변경되는지를 의미합니다.

1. **예측값을 계산합니다**

하나의 feature에 대해 $\hat y = wx+b$; 여러 feature의 경우 $\hat y = \sum_j x_j w_j+b$. `i`로 인덱싱된 현재 예제를 사용하세요.

2. **합계에 제곱오차를 더합니다**

새로운 합계 = 기존 `cost` + $(\hat y_i-y_i)^2$. 이전 줄에서 예측값 변수를 사용합니다.

3. **평균 비용 계산**

$J = \frac{1}{2m}\sum_i(\hat y_i-y_i)^2$. 누적된 오류를 `m`의 두 배로 나눕니다.

**함수 참조:** [내적](https://numpy.org/doc/stable/reference/generated/numpy.dot.html), [열 평균 및 축](https://numpy.org/doc/stable/reference/generated/numpy.mean.html), [지수](https://numpy.org/doc/stable/reference/generated/numpy.exp.html). 아래 원본 실습에는 전체 수학적 맥락이 포함되어 있습니다.

출처: [C1_W2_Lab02_Multiple_Variable_Soln.ipynb](https://github.com/greyhatguy007/Machine-Learning-Specialization-Coursera/blob/d5c95ffb5b1524367e1a9147707ec4725298d715/C1%20-%20Supervised%20Machine%20Learning%20-%20Regression%20and%20Classification/week2/Optional%20Labs/C1_W2_Lab02_Multiple_Variable_Soln.ipynb)

```python
def compute_cost(X, y, w, b):
    """
    compute cost
    Args:
      X (ndarray (m,n)): Data, m examples with n features
      y (ndarray (m,)) : target values
      w (ndarray (n,)) : model parameters
      b (scalar)       : model parameter

    Returns:
      cost (scalar): cost
    """
    m = X.shape[0]
    cost = 0.0
    for i in range(m):
        f_wb_i = float(X[i] @ w) + b           #(n,)(n,) = scalar (see np.dot)
        cost = cost + np.square(y[i] - f_wb_i)       #scalar
    cost = cost / (2.0 * m)                      #scalar
    return cost
```

```python
# Verify the reference implementation
np.testing.assert_allclose(compute_cost(np.array([[1., 2.], [3., 4.]]), np.array([2., 4.]), np.zeros(2), 0.), 5.)
print("Check passed.")
```

```python
# Compute and display cost using our pre-chosen optimal parameters.
cost = compute_cost(X_train, y_train, w_init, b_init)
print(f'Cost at optimal w : {cost}')
```

# 5 다중 변수가 있는 Gradient Descent
다중 변수의 경우 Gradient descent:

$$\begin{align*} \text{repeat}&\text{ until convergence:} \; \lbrace \newline\;
& w_j = w_j -  \alpha \frac{\partial J(\mathbf{w},b)}{\partial w_j} \tag{5}  \; & \text{for j = 0..n-1}\newline
&b\ \ = b -  \alpha \frac{\partial J(\mathbf{w},b)}{\partial b}  \newline \rbrace
\end{align*}$$

여기서 n은 feature의 수이고, 매개변수 $w_j$, $b$는 동시에 업데이트됩니다.

$$
\begin{align}
\frac{\partial J(\mathbf{w},b)}{\partial w_j}  &= \frac{1}{m} \sum\limits_{i = 0}^{m-1} (f_{\mathbf{w},b}(\mathbf{x}^{(i)}) - y^{(i)})x_{j}^{(i)} \tag{6}  \\
\frac{\partial J(\mathbf{w},b)}{\partial b}  &= \frac{1}{m} \sum\limits_{i = 0}^{m-1} (f_{\mathbf{w},b}(\mathbf{x}^{(i)}) - y^{(i)}) \tag{7}
\end{align}
$$
* m은 데이터 세트의 학습 예제 수입니다.

    
* $f_{\mathbf{w},b}(\mathbf{x}^{(i)})$는 해당 모델의 예측값이고, $y^{(i)}$는 target 값입니다.

<a name="toc_15456_5.1"></a>
## 5.1 여러 변수를 사용하여 기울기 계산
방정식 (6)과 (7)을 계산하기 위한 구현은 다음과 같습니다. 이를 구현하는 방법에는 여러 가지가 있습니다. 이 버전에는
- 모든 m개 예제에 대한 외부 루프.
    - 예시의 $\frac{\partial J(\mathbf{w},b)}{\partial b}$는 직접 계산하여 누적할 수 있습니다.
    - 모든 n개의 feature에 대한 두 번째 루프에서:
        - $\frac{\partial J(\mathbf{w},b)}{\partial w_j}$는 각 $w_j$에 대해 계산됩니다.

### 참조 솔루션: 한 번에 한 줄씩

아래의 각 연습은 다음 코드 셀의 한 줄에 하나의 표현식으로 해결됩니다. 번호가 매겨진 참조는 해당 표현식을 코드 순서대로 설명합니다. 학생용 버전과 비교할 때 주변 루프와 반환 문을 유지하세요.

참고 표기: `m`는 예시 개수입니다. `i`는 예시를 선택합니다. `j`는 feature를 선택합니다. 예측값은 모델 출력을 의미합니다. 기울기는 매개변수가 변경될 때 비용이 어떻게 변경되는지를 의미합니다.

1. **예측값에서 실제 값을 뺀 값을 계산합니다**

$e_i=(\sum_j X_{ij}w_j+b)-y_i$. 현재 행 `X[i]`, 가중치 `w`, 편향 `b` 및 레이블 `y[i]`를 사용합니다.

2. **이 가중치 기울기를 누적합니다**

새로운 $g_j$ = 기존 $g_j + e_i X_{ij}$. 오류 이름은 이전 줄에 `err` 또는 `err_i`로 표시됩니다. 누적 합계는 `dj_dw[j]`입니다.

3. **편향 기울기 누적**

새로운 $g_b$ = 기존 $g_b+e_i$. `dj_db`에 `err` 또는 `err_i`를 추가합니다.

4. **가중치 기울기 평균을 구합니다**

누적된 `dj_dw`를 예시 개수 `m`로 나눕니다. 배열의 나눗셈은 모든 가중치에 대해 동시에 작동합니다.

5. **편향 기울기 평균화**

누적된 `dj_db`를 예제 `m`의 개수로 나눕니다.

**함수 참조:** [내적](https://numpy.org/doc/stable/reference/generated/numpy.dot.html), [열 평균 및 축](https://numpy.org/doc/stable/reference/generated/numpy.mean.html), [지수](https://numpy.org/doc/stable/reference/generated/numpy.exp.html). 아래 원본 실습에는 전체 수학적 맥락이 포함되어 있습니다.

출처: [C1_W2_Lab02_Multiple_Variable_Soln.ipynb](https://github.com/greyhatguy007/Machine-Learning-Specialization-Coursera/blob/d5c95ffb5b1524367e1a9147707ec4725298d715/C1%20-%20Supervised%20Machine%20Learning%20-%20Regression%20and%20Classification/week2/Optional%20Labs/C1_W2_Lab02_Multiple_Variable_Soln.ipynb)

```python
def compute_gradient(X, y, w, b):
    """
    Computes the gradient for linear regression
    Args:
      X (ndarray (m,n)): Data, m examples with n features
      y (ndarray (m,)) : target values
      w (ndarray (n,)) : model parameters
      b (scalar)       : model parameter

    Returns:
      dj_dw (ndarray (n,)): The gradient of the cost w.r.t. the parameters w.
      dj_db (scalar):       The gradient of the cost w.r.t. the parameter b.
    """
    m,n = X.shape           #(number of examples, number of features)
    dj_dw = np.zeros((n,))
    dj_db = 0.

    for i in range(m):
        err = float(X[i] @ w) + b - y[i]
        for j in range(n):
            dj_dw[j] = dj_dw[j] + X[i, j] * err
        dj_db = dj_db + err
    dj_dw = dj_dw / float(m)
    dj_db = dj_db / float(m)

    return dj_db, dj_dw
```

```python
# Verify the reference implementation
test_db, test_dw = compute_gradient(np.array([[1., 2.], [3., 4.]]), np.array([2., 4.]), np.zeros(2), 0.)
np.testing.assert_allclose(test_db, -3.)
np.testing.assert_allclose(test_dw, [-7., -10.])
print("Check passed.")
```

## 5.2 다중 변수가 있는 Gradient Descent
아래 루틴은 위의 방정식 (5)를 구현합니다.

### 참조 솔루션: 한 번에 한 줄씩

아래의 각 연습은 다음 코드 셀의 한 줄에 하나의 표현식으로 해결됩니다. 번호가 매겨진 참조는 해당 표현식을 코드 순서대로 설명합니다. 학생용 버전과 비교할 때 주변 루프와 반환 문을 유지하세요.

참고 표기: `m`는 예시 개수입니다. `i`는 예시를 선택합니다. `j`는 feature를 선택합니다. 예측값은 모델 출력을 의미합니다. 기울기는 매개변수가 변경될 때 비용이 어떻게 변경되는지를 의미합니다.

1. **가중치 업데이트**

$w_{new}=w-\alpha\nabla_w J$. `w`, `alpha` 및 `dj_dw`를 사용하십시오. 배열 산술은 모든 가중치를 업데이트합니다.

2. **편향 업데이트**

$b_{new}=b-\alpha\,\partial J/\partial b$. `b`, `alpha` 및 `dj_db`를 사용합니다.

**함수 참조:** [내적](https://numpy.org/doc/stable/reference/generated/numpy.dot.html), [열 평균 및 축](https://numpy.org/doc/stable/reference/generated/numpy.mean.html), [지수](https://numpy.org/doc/stable/reference/generated/numpy.exp.html). 아래 원본 실습에는 전체 수학적 맥락이 포함되어 있습니다.

출처: [C1_W2_Lab02_Multiple_Variable_Soln.ipynb](https://github.com/greyhatguy007/Machine-Learning-Specialization-Coursera/blob/d5c95ffb5b1524367e1a9147707ec4725298d715/C1%20-%20Supervised%20Machine%20Learning%20-%20Regression%20and%20Classification/week2/Optional%20Labs/C1_W2_Lab02_Multiple_Variable_Soln.ipynb)

```python
def gradient_descent(X, y, w_in, b_in, cost_function, gradient_function, alpha, num_iters):
    """
    Performs batch gradient descent to learn theta. Updates theta by taking
    num_iters gradient steps with learning rate alpha

    Args:
      X (ndarray (m,n))   : Data, m examples with n features
      y (ndarray (m,))    : target values
      w_in (ndarray (n,)) : initial model parameters
      b_in (scalar)       : initial model parameter
      cost_function       : function to compute cost
      gradient_function   : function to compute the gradient
      alpha (float)       : Learning rate
      num_iters (int)     : number of iterations to run gradient descent

    Returns:
      w (ndarray (n,)) : Updated values of parameters
      b (scalar)       : Updated value of parameter
      """

    # An array to store cost J and w's at each iteration primarily for graphing later
    J_history = []
    w = copy.deepcopy(w_in)  #avoid modifying global w within function
    b = b_in

    for i in range(num_iters):

        # Calculate the gradient and update the parameters
        dj_db,dj_dw = gradient_function(X, y, w, b)   ##None

        # Update Parameters using w, b, alpha and gradient
        w = w - dj_dw * alpha               ##None
        b = b - dj_db * alpha               ##None

        # Save cost J at each iteration
        if i<100000:      # prevent resource exhaustion
            J_history.append( cost_function(X, y, w, b))

        # Print cost every at intervals 10 times or as many iterations if < 10
        if i% math.ceil(num_iters / 10) == 0:
            print(f"Iteration {i:4d}: Cost {J_history[-1]:8.2f}   ")

    return w, b, J_history #return final w,b and J history for graphing
```

```python
# Verify the reference implementation
test_w, test_b, *_ = gradient_descent(np.array([[1., 2.], [3., 4.]]), np.array([2., 4.]), np.zeros(2), 0., compute_cost, compute_gradient, 0.1, 1)
np.testing.assert_allclose(test_w, [0.7, 1.])
np.testing.assert_allclose(test_b, 0.3)
print("Check passed.")
```

## 여러 변수가 있는 Gradient Descent
다음은 다중 변수에 대해 gradient descent의 마지막 실습에서 개발한 방정식입니다.:

$$\begin{align*} \text{repeat}&\text{ until convergence:} \; \lbrace \newline\;
& w_j := w_j -  \alpha \frac{\partial J(\mathbf{w},b)}{\partial w_j} \tag{1}  \; & \text{for j = 0..n-1}\newline
&b\ \ := b -  \alpha \frac{\partial J(\mathbf{w},b)}{\partial b}  \newline \rbrace
\end{align*}$$

여기서 n은 feature의 수이고, 매개변수 $w_j$, $b$는 동시에 업데이트됩니다.

$$
\begin{align}
\frac{\partial J(\mathbf{w},b)}{\partial w_j}  &= \frac{1}{m} \sum\limits_{i = 0}^{m-1} (f_{\mathbf{w},b}(\mathbf{x}^{(i)}) - y^{(i)})x_{j}^{(i)} \tag{2}  \\
\frac{\partial J(\mathbf{w},b)}{\partial b}  &= \frac{1}{m} \sum\limits_{i = 0}^{m-1} (f_{\mathbf{w},b}(\mathbf{x}^{(i)}) - y^{(i)}) \tag{3}
\end{align}
$$
* m은 데이터 세트의 학습 예제 수입니다.

    
* $f_{\mathbf{w},b}(\mathbf{x}^{(i)})$는 해당 모델의 예측값이고, $y^{(i)}$는 target 값입니다.

해결책은 Feature Scaling입니다.

강의에서는 세 가지 다른 기술에 대해 논의했습니다.
- Feature Scaling: 양수인 feature를 최대값으로 나누거나, 일반적으로 (x-min)/(max-min)을 사용합니다. 이때 값은 0~1 범위가 됩니다. 첫 번째 방법은 양수인 feature에, 두 번째 방법은 음수가 포함된 feature에도 사용할 수 있습니다. 두 방법 모두 최대값과 최소값이 같은 상수 feature는 별도 처리가 필요합니다.
- 평균 정규화: $x_i := \dfrac{x_i - \mu_i}{max - min} $
- 아래에서 살펴볼 Z-score 정규화.

### z-score 정규화
z-score 정규화 후에는 모든 feature의 평균은 0, 표준편차는 1이 됩니다.

Z-score 정규화를 구현하려면 다음 수식에 표시된 대로 입력 값을 조정하세요.
$$x^{(i)}_j = \dfrac{x^{(i)}_j - \mu_j}{\sigma_j} \tag{4}$$
여기서 $j$는 feature 또는 $\mathbf{X}$ 행렬의 열을 선택합니다. $µ_j$는 feature(j)에 대한 모든 값의 평균이고 $\sigma_j$는 feature(j)의 표준편차입니다.
$$
\begin{align}
\mu_j &= \frac{1}{m} \sum_{i=0}^{m-1} x^{(i)}_j \tag{5}\\
\sigma^2_j &= \frac{1}{m} \sum_{i=0}^{m-1} (x^{(i)}_j - \mu_j)^2  \tag{6}
\end{align}
$$

>**구현 참고:** feature를 정규화할 때 정규화에 사용된 값, 즉 계산에 사용된 평균값과 표준 편차를 저장하는 것이 중요합니다. 모델에서 매개변수를 학습한 후 이전에 본 적이 없는 주택 가격을 예측하려는 경우가 많습니다. 새로운 x 값(거실 면적 및 침실 수)이 주어지면 먼저 이전에 학습 세트에서 계산한 평균 및 표준 편차를 사용하여 x를 정규화해야 합니다.

### 참조 솔루션: 한 번에 한 줄씩

아래의 각 연습은 다음 코드 셀의 한 줄에 하나의 표현식으로 해결됩니다. 번호가 매겨진 참조는 해당 표현식을 코드 순서대로 설명합니다. 학생용 버전과 비교할 때 주변 루프와 반환 문을 유지하세요.

참고 표기: `m`는 예시 개수입니다. `i`는 예시를 선택합니다. `j`는 feature를 선택합니다. 예측값은 모델 출력을 의미합니다. 기울기는 매개변수가 변경될 때 비용이 어떻게 변경되는지를 의미합니다.

1. **각 feature 평균을 계산합니다**

$\mu_j=\frac{1}{m}\sum_i X_{ij}$. `np.mean`와 `axis=0`는 열당 하나의 평균을 반환합니다.

2. **각 feature 표준편차를 계산합니다**

$\sigma_j=\sqrt{\frac{1}{m}\sum_i(X_{ij}-\mu_j)^2}$. `np.std`를 `axis=0` 및 기본 `ddof=0`와 함께 사용하세요.

3. **feature 표준화**

$X_{ij}^{scaled}=(X_{ij}-\mu_j)/\sigma_j$. `X`, `mu` 및 `sigma`를 사용하십시오. NumPy는 모든 행에 작업을 적용합니다.

**함수 참조:** [내적](https://numpy.org/doc/stable/reference/generated/numpy.dot.html), [열 평균 및 축](https://numpy.org/doc/stable/reference/generated/numpy.mean.html), [지수](https://numpy.org/doc/stable/reference/generated/numpy.exp.html). 아래 원본 실습에는 전체 수학적 맥락이 포함되어 있습니다.

출처: [C1_W2_Lab03_Feature_Scaling_and_Learning_Rate_Soln.ipynb](https://github.com/greyhatguy007/Machine-Learning-Specialization-Coursera/blob/d5c95ffb5b1524367e1a9147707ec4725298d715/C1%20-%20Supervised%20Machine%20Learning%20-%20Regression%20and%20Classification/week2/Optional%20Labs/C1_W2_Lab03_Feature_Scaling_and_Learning_Rate_Soln.ipynb)

```python
def zscore_normalize_features(X):
    """
    computes  X, zcore normalized by column

    Args:
      X (ndarray (m,n))     : input data, m examples, n features

    Returns:
      X_norm (ndarray (m,n)): input normalized by column
      mu (ndarray (n,))     : mean of each feature
      sigma (ndarray (n,))  : standard deviation of each feature
    """
    # find the mean of each column/feature
    mu     = X.mean(axis=0)                 # mu will have shape (n,)
    # find the standard deviation of each column/feature
    sigma  = X.std(axis=0)                  # sigma will have shape (n,)
    # element-wise, subtract mu for that column from each example, divide by std for that column
    X_norm = np.divide(X - mu, sigma)

    return (X_norm, mu, sigma)

#check our work
#from sklearn.preprocessing import scale
#scale(X_orig, axis=0, with_mean=True, with_std=True, copy=True)
```

```python
# Verify the reference implementation
test_norm, test_mu, test_sigma = zscore_normalize_features(np.array([[1., 2.], [3., 6.]]))
np.testing.assert_allclose(test_mu, [2., 4.])
np.testing.assert_allclose(test_sigma, [1., 2.])
np.testing.assert_allclose(test_norm, [[-1., -1.], [1., 1.]])
print("Check passed.")
```

- 중간: 첫 번째 단계에서는 각 feature에서 평균 또는 평균값을 제거합니다. 그러면 feature가 0을 중심으로 남게 됩니다. '연령' feature의 차이를 확인하기는 어렵지만 '크기(sqft)'는 확실히 0에 가깝습니다.

# Feature Engineering 및 다항식 회귀 개요

기본적으로 Linear Regression은 다음 형식의 모델을 구축하는 수단을 제공합니다.
$$f_{\mathbf{w},b} = w_0x_0 + w_1x_1+ ... + w_{n-1}x_{n-1} + b \tag{1}$$
feature/데이터가 비선형이거나 feature의 조합인 경우에는 어떻게 됩니까? 예를 들어, 주택 가격은 생활 면적에 선형적인 경향이 없지만 매우 작거나 큰 주택에 불이익을 가하여 위 그래픽에 표시된 곡선을 만듭니다. 이 곡선에 맞게 linear regression 기계를 어떻게 사용할 수 있습니까? 우리가 가지고 있는 '기계'는 방정식을 학습 데이터에 '맞추기' 위해 (1)의 매개변수 $\mathbf{w}$, $\mathbf{b}$를 수정하는 기능입니다. 그러나 (1)의 $\mathbf{w}$, $\mathbf{b}$를 아무리 조정해도 비선형 곡선에 적합하지 않습니다.

## 다항식 Features

위에서 우리는 데이터가 비선형인 시나리오를 고려했습니다. 비선형 곡선을 맞추기 위해 지금까지 우리가 알고 있는 것을 사용해 보겠습니다. 간단한 이차식부터 시작하겠습니다: $y = 1+x^2$

당신은 우리가 사용하는 모든 루틴에 익숙합니다. 검토를 위해 lab_utils.py 파일에서 사용할 수 있습니다. 열 경계를 따라 연결하기 위해 NumPy 루틴인 [`np.c_[..]`](https://numpy.org/doc/stable/reference/generated/numpy.c_.html)를 사용합니다.

```python
# create target data
x = np.arange(0, 20, 1)
y = 1 + x**2
X = x.reshape(-1, 1)

model_w,model_b = run_gradient_descent_feng(X,y,iterations=1000, alpha = 1e-2)

plt.scatter(x, y, marker='x', c='r', label="Actual Value"); plt.title("no feature engineering")
plt.plot(x,X@model_w + model_b, label="Predicted Value");  plt.xlabel("X"); plt.ylabel("y"); plt.legend(); plt.show()
```

뭐, 예상대로 별로 맞지는 않네요. 필요한 것은 $y= w_0x_0^2 + b$ 또는 **다항식 feature**와 같은 것입니다. 이를 달성하려면 *입력 데이터*를 수정하여 필요한 feature를 *엔지니어링*할 수 있습니다. 원본 데이터를 $x$ 값을 제곱하는 버전으로 바꾸면 $y= w_0x_0^2 + b$를 얻을 수 있습니다. 한번 해보자. 아래에서 `X`를 `X**2`로 교체하세요.

1. **3승까지 feature를 생성합니다**

$[x,x^2,x^3]$ 열을 해당 순서대로 생성합니다. 출력에는 `len(x)` 행과 3개의 열이 있습니다.

**Python 패턴, 다른 예:** `np.column_stack([t, t**2])`는 두 개의 feature 열을 만듭니다. 13개의 거듭제곱을 직접 작성하는 것을 피하기 위해 한 줄 패턴 `np.column_stack([t**k for k in range(2, 6)])`는 2에서 5까지의 거듭제곱에 대한 열을 만듭니다. 입력 및 지수 범위를 요청된 열에 적용합니다. `range`의 상한은 제외됩니다.

**함수 참조:** [내적](https://numpy.org/doc/stable/reference/generated/numpy.dot.html), [열 평균 및 축](https://numpy.org/doc/stable/reference/generated/numpy.mean.html), [지수](https://numpy.org/doc/stable/reference/generated/numpy.exp.html). 아래 원본 실습에는 전체 수학적 맥락이 포함되어 있습니다.

```python
# create target data
x = np.arange(0, 20, 1)
y = x**2

# engineer features .
X = np.column_stack([x ** power for power in range(1, 4)])   #<-- added engineered feature
```

```python
model_w,model_b = run_gradient_descent_feng(X, y, iterations=10000, alpha=1e-7)

plt.scatter(x, y, marker='x', c='r', label="Actual Value"); plt.title("x, x**2, x**3 features")
plt.plot(x, X@model_w + model_b, label="Predicted Value"); plt.xlabel("x"); plt.ylabel("y"); plt.legend(); plt.show()
```

$\mathbf{w}$, `[0.08 0.54 0.03]` 및 b의 값은 `0.0106`입니다. 이는 피팅/훈련 후 모델이 다음과 같음을 의미합니다.
$$ 0.08x + 0.54x^2 + 0.03x^3 + 0.0106 $$
Gradient descent는 $w_1$ 항을 다른 항에 비해 늘려 $x^2$ 데이터에 가장 적합한 데이터를 강조했습니다.  아주 오랜 시간 동안 실행한다면 다른 용어의 영향은 계속해서 줄어들 것입니다.
>Gradient descent는 ​​관련 매개변수를 강조하여 '올바른' feature를 선택합니다.

이 아이디어를 검토해 보겠습니다.
- 처음에 feature는 서로 비교할 수 있도록 크기가 조정되었습니다.
- 가중치 값이 적다는 것은 feature가 덜 중요하거나 정확하다는 것을 의미하며, 극단적으로 가중치가 0이 되거나 0에 매우 가까워지면 해당 feature는 현재 모델의 예측에 거의 기여하지 않습니다.
- 위의 피팅 후 $x^2$ feature와 관련된 가중치는 데이터 피팅에 가장 유용하므로 $x$ 또는 $x^3$의 가중치보다 훨씬 큽니다.

# Gradient Descent
Scikit-learn에는 gradient descent 회귀 모델 [sklearn.linear_model.SGDRegressor](https://scikit-learn.org/stable/modules/generated/sklearn.linear_model.SGDRegressor.html#examples-using-sklearn-linear-model-sgdregressor)이 있습니다.  gradient descent의 이전 구현과 마찬가지로 이 모델은 정규화된 입력에서 가장 잘 작동합니다. [sklearn.preprocessing.StandardScaler](https://scikit-learn.org/stable/modules/generated/sklearn.preprocessing.StandardScaler.html#sklearn.preprocessing.StandardScaler)는 이전 실습에서처럼 z-score 정규화를 수행합니다. 여기서는 '표준점수'라고 합니다.

### 데이터 세트 로드

```python
X_train, y_train = load_house_data()
X_features = ['size(sqft)','bedrooms','floors','age']
```

```python
scaler = StandardScaler()
X_norm = scaler.fit_transform(X_train)
print(f"Peak to Peak range by column in Raw        X:{np.ptp(X_train,axis=0)}")
print(f"Peak to Peak range by column in Normalized X:{np.ptp(X_norm,axis=0)}")
```

### 매개변수 보기
매개변수는 *정규화된* 입력 데이터와 연관되어 있습니다. 적합 매개변수는 이 데이터를 사용하여 이전 실습에서 찾은 매개변수와 매우 유사합니다.

```python
b_norm = sgdr.intercept_
w_norm = sgdr.coef_
print(f"model parameters:                   w: {w_norm}, b:{b_norm}")
print( "model parameters from previous lab: w: [110.56 -21.27 -32.71 -37.97], b: 363.16")
```

### 예측값 만들기
학습 데이터의 목표를 예측합니다. `predict` 루틴을 모두 사용하고 $w$ 및 $b$를 사용하여 계산합니다.

```python
# make a prediction using sgdr.predict()
y_pred_sgd = sgdr.predict(X_norm)
# make a prediction using w,b.
y_pred = np.dot(X_norm, w_norm) + b_norm
print(f"prediction using np.dot() and sgdr.predict match: {(y_pred == y_pred_sgd).all()}")

print(f"Prediction on training set:\n{y_pred[:4]}" )
print(f"Target values \n{y_train[:4]}")
```

# Linear Regression, Closed-form Solution

두 번째 예는 여러 개의 feature를 사용한 이전 실습의 예입니다. 최종 매개변수 값과 예측값은 해당 실습의 정규화되지 않은 '장기 실행' 결과와 매우 유사합니다. 정규화되지 않은 실행은 결과를 생성하는 데 몇 시간이 걸렸지만 이는 거의 즉각적입니다. Closed-form Solution은 이와 같은 소규모 데이터 세트에서는 잘 작동하지만 대규모 데이터 세트에서는 계산이 까다로울 수 있습니다.

```python
# load the dataset
X_train, y_train = load_house_data()
X_features = ['size(sqft)','bedrooms','floors','age']
```

```python
b = linear_model.intercept_
w = linear_model.coef_
print(f"w = {w:}, b = {b:0.2f}")
```

```python
print(f"Prediction on training set:\n {linear_model.predict(X_train)[:4]}" )
print(f"prediction using w,b:\n {(X_train @ w + b)[:4]}")
print(f"Target values \n {y_train[:4]}")

x_house = np.array([1200, 3,1, 40]).reshape(-1,4)
x_house_predict = linear_model.predict(x_house)[0]
print(f" predicted price of a house with 1200 sqft, 3 bedrooms, 1 floor, 40 years old = ${x_house_predict*1000:0.2f}")
```
