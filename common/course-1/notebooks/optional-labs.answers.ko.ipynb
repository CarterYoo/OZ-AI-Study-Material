## 목표
이 실습에서는 다음을 수행합니다.
- 하나의 변수를 사용하여 linear regression에 대한 모델 $f_{w,b}$를 구현하는 방법을 알아봅니다.

## 표기법
다음은 여러분이 보게 될 표기법 중 일부를 요약한 것입니다.

|일반 <img width=70/> <br /> 표기법 <img width=70/> | 설명<img width=350/>| Python(해당하는 경우) |
|: ------------|: -----------------------------------------------||
| $a$ | 스칼라, 굵지 않게 ||
| $\mathbf{a}$ | 벡터, 굵은 글씨 ||
| **회귀** |         |    |     |
|  $\mathbf{x}$ | 학습 예제 feature 값(이 실습에서는 크기(1000평방피트)) | `x_train` |   
|  $\mathbf{y}$ | 학습 예제 목표(이 실습에서는 가격(1,000달러)).  | `y_train`
|  $x^{(i)}$, $y^{(i)}$ | $i_{th}$학습 예제 | `x_i`, `y_i`|
| m | 학습 예제 수 | `m`|
|  $w$ |  매개변수: 가중치 | `w` |
|  $b$ |  매개변수: 편향 | `b` |     
| $f_{w,b}(x^{(i)})$ | $w,b$에 의해 매개변수화된 $x^{(i)}$에서의 모델 평가 결과: $f_{w,b}(x^{(i)}) = wx^{(i)}+b$ | `f_wb` |

## 도구
이 실습에서는 다음을 활용합니다.
- 과학 컴퓨팅을 위한 인기 라이브러리인 NumPy
- Matplotlib, 데이터 플로팅에 널리 사용되는 라이브러리

````python
import numpy as np
import matplotlib.pyplot as plt
plt.style.use('./deeplearning.mplstyle')
````

# 문제 진술
<img align="left" src="https://raw.githubusercontent.com/greyhatguy007/Machine-Learning-Specialization-Coursera/d5c95ffb5b1524367e1a9147707ec4725298d715/C1%20-%20Supervised%20Machine%20Learning%20-%20Regression%20and%20Classification/week1/Optional%20Labs/images/C1_W1_L3_S1_trainingdata.png"    style=" width:380px; padding: 10px;  " />

강의에서 다룬 주택 가격 예측 예제를 사용합니다. 면적이 1000제곱피트인 집은 300,000달러, 2000제곱피트인 집은 500,000달러에 팔렸습니다. 이 두 점이 학습 데이터입니다. 아래 표에서는 면적을 1000제곱피트, 가격을 1000달러 단위로 표시합니다.

| 규모(1000평방피트) | 가격(천달러) |
| ------| ----------- |
| 1.0 | 300 |
| 2.0 | 500 |

이 두 점을 통해 linear regression 모델(위의 파란색 직선으로 표시)을 적합하게 하여 다른 주택(예: 1200평방피트의 주택)의 가격을 예측할 수 있습니다.

다음 코드 셀을 실행하여 `x_train` 및 `y_train` 변수를 생성하십시오. 데이터는 1차원 NumPy 배열에 저장됩니다.

````python
# x_train is the input variable (size in 1000 square feet)
# y_train is the target (price in 1000s of dollars)
x_train = np.array([1.0, 2.0])
y_train = np.array([300.0, 500.0])
print(f"x_train = {x_train}")
print(f"y_train = {y_train}")
````

>**참고**: 이 과정에서는 인쇄할 때 [여기](https://docs.python.org/3/tutorial/inputoutput.html)에 설명된 Python 'f-string' 출력 형식을 자주 활용합니다. 중괄호 사이의 내용은 출력을 생성할 때 평가됩니다.

### 학습 예제 수 `m`
`m`를 사용하여 학습 예제의 수를 나타냅니다. Numpy 어레이에는 `.shape` 매개변수가 있습니다. `x_train.shape`는 각 차원에 대한 항목이 포함된 Python 튜플을 반환합니다. `x_train.shape[0]`는 아래와 같이 배열의 길이와 예제의 개수입니다.

````python
# m is the number of training examples
print(f"x_train.shape: {x_train.shape}")
m = x_train.shape[0]
print(f"Number of training examples is: {m}")
````

아래와 같이 Python `len()` 함수를 사용할 수도 있습니다.

## 모델 함수

<img align="left" src="https://raw.githubusercontent.com/greyhatguy007/Machine-Learning-Specialization-Coursera/d5c95ffb5b1524367e1a9147707ec4725298d715/C1%20-%20Supervised%20Machine%20Learning%20-%20Regression%20and%20Classification/week1/Optional%20Labs/images/C1_W1_L3_S1_model.png"     style=" width:380px; padding: 10px; " > 강의에서 설명한 것처럼 linear regression(`x`에서 `y`로 매핑하는 함수)에 대한 모델 함수는 다음과 같이 표현됩니다.

$$ f_{w,b}(x^{(i)}) = wx^{(i)} + b \tag{1}$$

위 공식은 직선을 표현하는 방법입니다. $w$와 $b$의 값에 따라 플롯에 다른 직선이 표시됩니다. <br/> <br/> <br/> <br/> <br/>

아래 코드 블록을 통해 이에 대한 더 나은 직관을 얻으려고 노력해 보겠습니다. $w = 100$ 및 $b = 100$부터 시작하겠습니다.

**참고: 이 셀로 돌아와서 모델의 w 및 b 매개변수를 조정할 수 있습니다**

````python
w = 100
b = 100
print(f"w: {w}")
print(f"b: {b}")
````

이제 두 데이터 포인트에 대한 $f_{w,b}(x^{(i)})$ 값을 계산해 보겠습니다. 각 데이터 포인트에 대해 다음과 같이 명시적으로 작성할 수 있습니다.

$x^{(0)}$, `f_wb = w * x[0] + b`의 경우

$x^{(1)}$, `f_wb = w * x[1] + b`의 경우

데이터 포인트 수가 많은 경우 이는 다루기 힘들고 반복적일 수 있습니다. 따라서 대신 아래 `compute_model_output` 함수에 표시된 대로 `for` 루프의 함수 출력을 계산할 수 있습니다.
> **참고**: 인수 설명 `(ndarray (m,))`는 모양(m,)의 Numpy n차원 배열을 설명합니다. `(scalar)`는 차원 없이 크기만 설명하는 인수를 설명합니다.  
> **참고**: `np.zero(n)`는 $n$ 항목이 포함된 1차원 numpy 배열을 반환합니다.

### 참조 솔루션: 한 번에 한 줄씩

아래의 각 연습은 다음 코드 셀의 한 줄에 하나의 표현식으로 해결됩니다. 번호가 매겨진 참조는 해당 표현식을 코드 순서대로 설명합니다. 학생용 버전과 비교할 때 주변 루프와 반환 문을 유지하세요.

참고 표기: `m`는 예시 개수입니다. `i`는 예시를 선택합니다. `j`는 feature를 선택합니다. 예측값은 모델 출력을 의미합니다. 기울기는 매개변수가 변경될 때 비용이 어떻게 변경되는지를 의미합니다.

1. **예측값을 계산합니다**

하나의 feature에 대해 $\hat y = wx+b$; 여러 feature의 경우 $\hat y = \sum_j x_j w_j+b$. `i`로 인덱싱된 현재 예제를 사용하세요.

**함수 참조:** [내적](https://numpy.org/doc/stable/reference/generated/numpy.dot.html), [열 평균 및 축](https://numpy.org/doc/stable/reference/generated/numpy.mean.html), [지수](https://numpy.org/doc/stable/reference/generated/numpy.exp.html). 아래 원본 실습에는 전체 수학적 맥락이 포함되어 있습니다.

출처: [C1_W1_Lab03_Model_Representation_Soln.ipynb](https://github.com/greyhatguy007/Machine-Learning-Specialization-Coursera/blob/d5c95ffb5b1524367e1a9147707ec4725298d715/C1%20-%20Supervised%20Machine%20Learning%20-%20Regression%20and%20Classification/week1/Optional%20Labs/C1_W1_Lab03_Model_Representation_Soln.ipynb)

````python
def compute_model_output(x, w, b):
    """
    Computes the prediction of a linear model
    Args:
      x (ndarray (m,)): Data, m examples
      w,b (scalar)    : model parameters
    Returns
      y (ndarray (m,)): target values
    """
    m = x.shape[0]
    f_wb = np.zeros(m)
    for i in range(m):
        f_wb[i] = b + x[i] * w

    return f_wb
````

````python
# Verify the reference implementation
np.testing.assert_allclose(compute_model_output(np.array([0., 2.]), 3., 1.), [1., 7.])
print("Check passed.")
````

이제 `compute_model_output` 함수를 호출하고 출력을 플로팅해 보겠습니다.

````python
tmp_f_wb = compute_model_output(x_train, w, b,)

# Plot our model prediction
plt.plot(x_train, tmp_f_wb, c='b',label='Our Prediction')

# Plot the data points
plt.scatter(x_train, y_train, marker='x', c='r',label='Actual Values')

# Set the title
plt.title("Housing Prices")
# Set the y-axis label
plt.ylabel('Price (in 1000s of dollars)')
# Set the x-axis label
plt.xlabel('Size (1000 sqft)')
plt.legend()
plt.show()
````

## Cost 계산
이 과제에서 '비용'이라는 용어는 데이터가 주택 비용이므로 약간 혼란스러울 수 있습니다. 여기서 비용은 모델이 집의 target 가격을 얼마나 잘 예측하는지 측정하는 것입니다. 주택 데이터에는 '가격'이라는 용어가 사용됩니다.

변수가 하나인 비용 방정식은 다음과 같습니다.
  $$J(w,b) = \frac{1}{2m} \sum\limits_{i = 0}^{m-1} (f_{w,b}(x^{(i)}) - y^{(i)})^2 \tag{1}$$

어디서
  $$f_{w,b}(x^{(i)}) = wx^{(i)} + b \tag{2}$$
  
- $f_{w,b}(x^{(i)})$는 예측값입니다(예: $w,b$ 매개변수를 사용하는 $i$).  
- $(f_{w,b}(x^{(i)}) -y^{(i)})^2$는 target 값과 예측값 사이의 제곱 차이입니다.   
- 이러한 차이는 모든 $m$ 예에 대해 합산되고 `2m`로 나누어 비용 $J(w,b)$를 생성합니다.  
>강의에서 합산하는 인덱스 범위는 일반적으로 1에서 m까지이고 코드는 0에서 m-1까지입니다.

아래 코드는 각 예를 반복하여 비용을 계산합니다. 각 루프에서:
- `f_wb`, 예측값가 계산됩니다.
- target과 예측값의 차이를 계산하여 제곱합니다.
- 총 비용에 추가됩니다.

### 참조 솔루션: 한 번에 한 줄씩

아래의 각 연습은 다음 코드 셀의 한 줄에 하나의 표현식으로 해결됩니다. 번호가 매겨진 참조는 해당 표현식을 코드 순서대로 설명합니다. 학생용 버전과 비교할 때 주변 루프와 반환 문을 유지하세요.

참고 표기: `m`는 예시 개수입니다. `i`는 예시를 선택합니다. `j`는 feature를 선택합니다. 예측값은 모델 출력을 의미합니다. 기울기는 매개변수가 변경될 때 비용이 어떻게 변경되는지를 의미합니다.

1. **예측값을 계산합니다**

하나의 feature에 대해 $\hat y = wx+b$; 여러 feature의 경우 $\hat y = \sum_j x_j w_j+b$. `i`로 인덱싱된 현재 예제를 사용하세요.

2. **제곱 오차 계산**

$(\hat y-y)^2$. `f_wb`를 `y[i]`와 비교하세요. 제곱에는 `** 2`를 사용하세요.

3. **전체에 이 오류를 추가하세요**

새 합계 = 이전 `cost_sum` + 이 예의 `cost`.

4. **평균 비용 계산**

$J = \frac{1}{2m}\sum_i(\hat y_i-y_i)^2$. 누적된 오류를 `m`의 두 배로 나눕니다.

**함수 참조:** [내적](https://numpy.org/doc/stable/reference/generated/numpy.dot.html), [열 평균 및 축](https://numpy.org/doc/stable/reference/generated/numpy.mean.html), [지수](https://numpy.org/doc/stable/reference/generated/numpy.exp.html). 아래 원본 실습에는 전체 수학적 맥락이 포함되어 있습니다.

출처: [C1_W1_Lab04_Cost_function_Soln.ipynb](https://github.com/greyhatguy007/Machine-Learning-Specialization-Coursera/blob/d5c95ffb5b1524367e1a9147707ec4725298d715/C1%20-%20Supervised%20Machine%20Learning%20-%20Regression%20and%20Classification/week1/Optional%20Labs/C1_W1_Lab04_Cost_function_Soln.ipynb)

````python
def compute_cost(x, y, w, b):
    """
    Computes the cost function for linear regression.

    Args:
      x (ndarray (m,)): Data, m examples
      y (ndarray (m,)): target values
      w,b (scalar)    : model parameters

    Returns
        total_cost (float): The cost of using w,b as the parameters for linear regression
               to fit the data points in x and y
    """
    # number of training examples
    m = x.shape[0]

    cost_sum = 0
    for i in range(m):
        f_wb = b + x[i] * w
        cost = np.square(y[i] - f_wb)
        cost_sum = cost + cost_sum
    total_cost = cost_sum / (2.0 * m)

    return total_cost
````

````python
# Verify the reference implementation
np.testing.assert_allclose(compute_cost(np.array([1., 2.]), np.array([3., 5.]), 1., 0.), 3.25)
print("Check passed.")
````

## Cost Function 직관

<img align="left" src="https://raw.githubusercontent.com/greyhatguy007/Machine-Learning-Specialization-Coursera/d5c95ffb5b1524367e1a9147707ec4725298d715/C1%20-%20Supervised%20Machine%20Learning%20-%20Regression%20and%20Classification/week1/Optional%20Labs/images/C1_W1_Lab02_GoalOfRegression.PNG"    style=" width:380px; padding: 10px;  " /> 귀하의 목표는 $x$ 입력을 통해 주택 가치를 정확하게 예측하는 매개변수 $w,b$를 사용하여 모델 $f_{w,b}(x) = wx + b$를 찾는 것입니다. 비용은 학습 데이터에서 모델이 얼마나 정확한지를 측정한 것입니다.

위의 비용 방정식 (1)은 예측값 $f_{w,b}(x)$가 target 데이터 $y$와 일치하도록 $w$ 및 $b$를 선택할 수 있는 경우 $(f_{w,b}(x^{(i)}) - y^{(i)})^2 $ 항은 0이 되고 비용은 최소화된다는 것을 보여줍니다. 이 간단한 2점 예를 통해 이를 달성할 수 있습니다!

이전 실습에서는 $b=100$가 최적의 솔루션을 제공한다고 판단했으므로 $b$를 100으로 설정하고 $w$에 집중하겠습니다.

<br/>
아래에서는 슬라이더 컨트롤을 사용하여 비용을 최소화하는 $w$ 값을 선택합니다. 플롯이 업데이트되는 데 몇 초 정도 걸릴 수 있습니다.

### Compute_Cost
이는 지난 실습에서 개발되었습니다. 여기서 다시 필요합니다.

### 참조 솔루션: 한 번에 한 줄씩

아래의 각 연습은 다음 코드 셀의 한 줄에 하나의 표현식으로 해결됩니다. 번호가 매겨진 참조는 해당 표현식을 코드 순서대로 설명합니다. 학생용 버전과 비교할 때 주변 루프와 반환 문을 유지하세요.

참고 표기: `m`는 예시 개수입니다. `i`는 예시를 선택합니다. `j`는 feature를 선택합니다. 예측값은 모델 출력을 의미합니다. 기울기는 매개변수가 변경될 때 비용이 어떻게 변경되는지를 의미합니다.

1. **예측값을 계산합니다**

하나의 feature에 대해 $\hat y = wx+b$; 여러 feature의 경우 $\hat y = \sum_j x_j w_j+b$. `i`로 인덱싱된 현재 예제를 사용하세요.

2. **합계에 제곱오차를 더합니다**

새로운 합계 = 기존 `cost` + $(\hat y_i-y_i)^2$. 이전 줄에서 예측값 변수를 사용합니다.

3. **평균 비용 계산**

$J = \frac{1}{2m}\sum_i(\hat y_i-y_i)^2$. 누적된 오류를 `m`의 두 배로 나눕니다.

**함수 참조:** [내적](https://numpy.org/doc/stable/reference/generated/numpy.dot.html), [열 평균 및 축](https://numpy.org/doc/stable/reference/generated/numpy.mean.html), [지수](https://numpy.org/doc/stable/reference/generated/numpy.exp.html). 아래 원본 실습에는 전체 수학적 맥락이 포함되어 있습니다.

출처: [C1_W1_Lab05_Gradient_Descent_Soln.ipynb](https://github.com/greyhatguy007/Machine-Learning-Specialization-Coursera/blob/d5c95ffb5b1524367e1a9147707ec4725298d715/C1%20-%20Supervised%20Machine%20Learning%20-%20Regression%20and%20Classification/week1/Optional%20Labs/C1_W1_Lab05_Gradient_Descent_Soln.ipynb)

````python
#Function to calculate the cost
def compute_cost(x, y, w, b):

    m = x.shape[0]
    cost = 0

    for i in range(m):
        f_wb = b + x[i] * w
        cost = cost + np.square(y[i] - f_wb)
    total_cost = cost / (2.0 * m)

    return total_cost
````

````python
# Verify the reference implementation
np.testing.assert_allclose(compute_cost(np.array([1., 2.]), np.array([3., 5.]), 1., 0.), 3.25)
print("Check passed.")
````

<a name="toc_40291_2.1"></a>
## Gradient descent 요약
지금까지 이 과정에서는 $f_{w,b}(x^{(i)})$를 예측하는 선형 모델을 개발했습니다.
$$f_{w,b}(x^{(i)}) = wx^{(i)} + b \tag{1}$$
linear regression에서는 예측값 $f_{w,b}(x^{(i)})$와 실제 데이터 $y^{(i)}$ 사이의 오류 측정을 최소화하여 매개변수 $w$,$b$에 맞게 입력 학습 데이터를 활용합니다. 측정값은 $cost$, $J(w,b)$라고 합니다. 교육에서는 모든 교육 샘플 $x^{(i)},y^{(i)}$에 대한 비용을 측정합니다.
$$J(w,b) = \frac{1}{2m} \sum\limits_{i = 0}^{m-1} (f_{w,b}(x^{(i)}) - y^{(i)})^2\tag{2}$$

강의에서 *gradient descent*는 다음과 같이 설명되었습니다.

$$\begin{align*} \text{repeat}&\text{ until convergence:} \; \lbrace \newline
\;  w &= w -  \alpha \frac{\partial J(w,b)}{\partial w} \tag{3}  \; \newline
 b &= b -  \alpha \frac{\partial J(w,b)}{\partial b}  \newline \rbrace
\end{align*}$$ 여기서 매개변수 $w$, $b$는 동시에 업데이트됩니다. 기울기은 다음과 같이 정의됩니다.
$$
\begin{align}
\frac{\partial J(w,b)}{\partial w}  &= \frac{1}{m} \sum\limits_{i = 0}^{m-1} (f_{w,b}(x^{(i)}) - y^{(i)})x^{(i)} \tag{4}\\
  \frac{\partial J(w,b)}{\partial b}  &= \frac{1}{m} \sum\limits_{i = 0}^{m-1} (f_{w,b}(x^{(i)}) - y^{(i)}) \tag{5}\\
\end{align}
$$

여기서 *동시에* 매개변수를 업데이트하기 전에 모든 매개변수에 대한 편미분를 계산한다는 의미입니다.

<a name="toc_40291_2.2"></a>
## Gradient Descent 구현
하나의 feature에 대해 gradient descent 알고리즘을 구현합니다. 함수 세 개가 필요합니다.
- 위의 방정식 (4)와 (5)를 구현하는 `compute_gradient`
- 위의 방정식 (2)를 구현하는 `compute_cost`(이전 실습의 코드)
- `gradient_descent`, 계산_기울기 및 compute_cost 활용

규칙:
- 편미분를 포함하는 Python 변수의 이름은 이 패턴을 따르며 $\frac{\partial J(w,b)}{\partial b}$는 `dj_db`가 됩니다.
- w.r.t는 With Respect To의 약자로, $b$에 대해 $J(wb)$를 미분한다는 뜻입니다.

<a name="toc_40291_2.3"></a>
### 계산_기울기
<a name='ex-01'></a>
`compute_gradient`는 위의 (4)와 (5)를 구현하고 $\frac{\partial J(w,b)}{\partial w}$,$\frac{\partial J(w,b)}{\partial b}$를 반환합니다. 포함된 주석은 작업을 설명합니다.

### 참조 솔루션: 한 번에 한 줄씩

아래의 각 연습은 다음 코드 셀의 한 줄에 하나의 표현식으로 해결됩니다. 번호가 매겨진 참조는 해당 표현식을 코드 순서대로 설명합니다. 학생용 버전과 비교할 때 주변 루프와 반환 문을 유지하세요.

참고 표기: `m`는 예시 개수입니다. `i`는 예시를 선택합니다. `j`는 feature를 선택합니다. 예측값은 모델 출력을 의미합니다. 기울기는 매개변수가 변경될 때 비용이 어떻게 변경되는지를 의미합니다.

1. **예측값을 계산합니다**

하나의 feature에 대해 $\hat y = wx+b$; 여러 feature의 경우 $\hat y = \sum_j x_j w_j+b$. `i`로 인덱싱된 현재 예제를 사용하세요.

2. **이 예의 가중치 기울기를 계산합니다**

$(\hat y_i-y_i)x_i$. `f_wb`, `y[i]` 및 `x[i]`를 사용합니다.

3. **이 예의 편향 기울기를 계산합니다**

$\hat y_i-y_i$. `f_wb` 및 `y[i]`를 사용하십시오.

4. **가중치 기울기 평균을 구합니다**

누적된 `dj_dw`를 예시 개수 `m`로 나눕니다. 배열의 나눗셈은 모든 가중치에 대해 동시에 작동합니다.

5. **편향 기울기 평균화**

누적된 `dj_db`를 예제 `m`의 개수로 나눕니다.

**함수 참조:** [내적](https://numpy.org/doc/stable/reference/generated/numpy.dot.html), [열 평균 및 축](https://numpy.org/doc/stable/reference/generated/numpy.mean.html), [지수](https://numpy.org/doc/stable/reference/generated/numpy.exp.html). 아래 원본 실습에는 전체 수학적 맥락이 포함되어 있습니다.

출처: [C1_W1_Lab05_Gradient_Descent_Soln.ipynb](https://github.com/greyhatguy007/Machine-Learning-Specialization-Coursera/blob/d5c95ffb5b1524367e1a9147707ec4725298d715/C1%20-%20Supervised%20Machine%20Learning%20-%20Regression%20and%20Classification/week1/Optional%20Labs/C1_W1_Lab05_Gradient_Descent_Soln.ipynb)

````python
def compute_gradient(x, y, w, b):
    """
    Computes the gradient for linear regression
    Args:
      x (ndarray (m,)): Data, m examples
      y (ndarray (m,)): target values
      w,b (scalar)    : model parameters
    Returns
      dj_dw (scalar): The gradient of the cost w.r.t. the parameters w
      dj_db (scalar): The gradient of the cost w.r.t. the parameter b
     """

    # Number of training examples
    m = x.shape[0]
    dj_dw = 0
    dj_db = 0

    for i in range(m):
        f_wb = b + x[i] * w
        dj_dw_i = x[i] * (f_wb - y[i])
        dj_db_i = f_wb - y[i]
        dj_db += dj_db_i
        dj_dw += dj_dw_i
    dj_dw = dj_dw / float(m)
    dj_db = dj_db / float(m)

    return dj_dw, dj_db
````

````python
# Verify the reference implementation
test_dw, test_db = compute_gradient(np.array([1., 2.]), np.array([3., 5.]), 1., 0.)
np.testing.assert_allclose([test_dw, test_db], [-4., -2.5])
print("Check passed.")
````

`compute_gradient` 함수를 사용하여 매개변수 중 하나인 $w_0$에 상대적인 cost function의 일부 편미분를 찾아 플로팅해 보겠습니다.

````python
plt_gradients(x_train,y_train, compute_cost, compute_gradient)
plt.show()
````

위의 왼쪽 플롯은 $\frac{\partial J(w,b)}{\partial w}$ 또는 세 지점에서 $w$를 기준으로 한 비용 곡선의 기울기를 보여줍니다. 플롯의 오른쪽에서는 도함수가 양수이고 왼쪽에서는 음수입니다. '그릇 모양'으로 인해 도함수은 항상 gradient descent를 기울기가 0인 아래쪽으로 유도합니다.

왼쪽 플롯은 $b=100$으로 고정했습니다. Gradient descent는 $\frac{\partial J(w,b)}{\partial w}$ 및 $\frac{\partial J(w,b)}{\partial b}$를 모두 활용하여 매개변수를 업데이트합니다. 오른쪽의 '퀴버 플롯'은 두 매개변수의 기울기를 볼 수 있는 수단을 제공합니다. 화살표 크기는 해당 지점의 기울기 크기를 반영합니다. 화살표의 방향과 기울기는 해당 지점의 $\frac{\partial J(w,b)}{\partial w}$와 $\frac{\partial J(w,b)}{\partial b}$의 비율을 반영합니다. 기울기는 최소값에서 *멀리* 지점을 가리킵니다. 위의 방정식 (3)을 검토하십시오. 조정된 기울기는 $w$ 또는 $b$의 현재 값에서 *뺄셈*됩니다. 그러면 비용이 절감되는 방향으로 매개변수가 이동됩니다.

<a name="toc_40291_2.5"></a>
### Gradient Descent
이제 기울기를 계산할 수 있으므로 위의 방정식 (3)에 설명된 gradient descent를 아래 `gradient_descent`에서 구현할 수 있습니다. 구현 세부 사항은 주석에 설명되어 있습니다. 아래에서는 이 함수를 사용하여 학습 데이터에서 $w$ 및 $b$의 최적 값을 찾습니다.

### 참조 솔루션: 한 번에 한 줄씩

아래의 각 연습은 다음 코드 셀의 한 줄에 하나의 표현식으로 해결됩니다. 번호가 매겨진 참조는 해당 표현식을 코드 순서대로 설명합니다. 학생용 버전과 비교할 때 주변 루프와 반환 문을 유지하세요.

참고 표기: `m`는 예시 개수입니다. `i`는 예시를 선택합니다. `j`는 feature를 선택합니다. 예측값은 모델 출력을 의미합니다. 기울기는 매개변수가 변경될 때 비용이 어떻게 변경되는지를 의미합니다.

1. **편향 업데이트**

$b_{new}=b-\alpha\,\partial J/\partial b$. `b`, `alpha` 및 `dj_db`를 사용합니다.

2. **가중치 업데이트**

$w_{new}=w-\alpha\,\partial J/\partial w$. `w`, `alpha` 및 `dj_dw`를 사용합니다.

**함수 참조:** [내적](https://numpy.org/doc/stable/reference/generated/numpy.dot.html), [열 평균 및 축](https://numpy.org/doc/stable/reference/generated/numpy.mean.html), [지수](https://numpy.org/doc/stable/reference/generated/numpy.exp.html). 아래 원본 실습에는 전체 수학적 맥락이 포함되어 있습니다.

출처: [C1_W1_Lab05_Gradient_Descent_Soln.ipynb](https://github.com/greyhatguy007/Machine-Learning-Specialization-Coursera/blob/d5c95ffb5b1524367e1a9147707ec4725298d715/C1%20-%20Supervised%20Machine%20Learning%20-%20Regression%20and%20Classification/week1/Optional%20Labs/C1_W1_Lab05_Gradient_Descent_Soln.ipynb)

````python
def gradient_descent(x, y, w_in, b_in, alpha, num_iters, cost_function, gradient_function):
    """
    Performs gradient descent to fit w,b. Updates w,b by taking
    num_iters gradient steps with learning rate alpha

    Args:
      x (ndarray (m,))  : Data, m examples
      y (ndarray (m,))  : target values
      w_in,b_in (scalar): initial values of model parameters
      alpha (float):     Learning rate
      num_iters (int):   number of iterations to run gradient descent
      cost_function:     function to call to produce cost
      gradient_function: function to call to produce gradient

    Returns:
      w (scalar): Updated value of parameter after running gradient descent
      b (scalar): Updated value of parameter after running gradient descent
      J_history (List): History of cost values
      p_history (list): History of parameters [w,b]
      """

    w = copy.deepcopy(w_in) # avoid modifying global w_in
    # An array to store cost J and w's at each iteration primarily for graphing later
    J_history = []
    p_history = []
    b = b_in
    w = w_in

    for i in range(num_iters):
        # Calculate the gradient and update the parameters using gradient_function
        dj_dw, dj_db = gradient_function(x, y, w , b)

        # Update Parameters using equation (3) above
        b = b - dj_db * alpha
        w = w - dj_dw * alpha

        # Save cost J at each iteration
        if i<100000:      # prevent resource exhaustion
            J_history.append( cost_function(x, y, w , b))
            p_history.append([w,b])
        # Print cost every at intervals 10 times or as many iterations if < 10
        if i% math.ceil(num_iters/10) == 0:
            print(f"Iteration {i:4}: Cost {J_history[-1]:0.2e} ",
                  f"dj_dw: {dj_dw: 0.3e}, dj_db: {dj_db: 0.3e}  ",
                  f"w: {w: 0.3e}, b:{b: 0.5e}")

    return w, b, J_history, p_history #return w and J,w history for graphing
````

````python
# Verify the reference implementation
test_w, test_b, *_ = gradient_descent(np.array([1., 2.]), np.array([3., 5.]), 1., 0., 0.1, 1, compute_cost, compute_gradient)
np.testing.assert_allclose([test_w, test_b], [1.4, 0.25])
print("Check passed.")
````

````python
# initialize parameters
w_init = 0
b_init = 0
# some gradient descent settings
iterations = 10000
tmp_alpha = 1.0e-2
# run gradient descent
w_final, b_final, J_hist, p_hist = gradient_descent(x_train ,y_train, w_init, b_init, tmp_alpha,
                                                    iterations, compute_cost, compute_gradient)
print(f"(w,b) found by gradient descent: ({w_final:8.4f},{b_final:8.4f})")
````

잠시 시간을 내어 위에 출력된 gradient descent 프로세스의 몇 가지 특징을 살펴보세요.

- 강의 슬라이드에 설명된 것처럼 비용은 크게 시작되었다가 급격히 감소합니다.
- 편미분 `dj_dw`, `dj_db`도 처음에는 빠르게 작아지고 나중에는 느려집니다. 강의 다이어그램에서 볼 수 있듯이 프로세스가 '그릇 바닥'에 가까워질수록 해당 지점의 미분 값이 작아지기 때문에 진행 속도가 느려집니다.
- learning rate, 알파는 고정되어 있지만 진행 속도가 느려집니다.

### 비용과 gradient descent의 반복 비교
비용 대 반복의 도표는 gradient descent의 진행 상황을 나타내는 유용한 척도입니다. 성공적인 실행에서는 비용이 항상 감소해야 합니다. 비용의 변화는 처음에는 매우 빠르므로 초기 구간과 마지막 구간을 서로 다른 축 범위로 표시하면 변화를 보기 쉽습니다. 아래 도표에서 축의 비용 규모와 반복 단계를 확인하세요.

````python
# plot cost versus iteration
fig, (ax1, ax2) = plt.subplots(1, 2, constrained_layout=True, figsize=(12,4))
ax1.plot(J_hist[:100])
ax2.plot(1000 + np.arange(len(J_hist[1000:])), J_hist[1000:])
ax1.set_title("Cost vs. iteration(start)");  ax2.set_title("Cost vs. iteration (end)")
ax1.set_ylabel('Cost')            ;  ax2.set_ylabel('Cost')
ax1.set_xlabel('iteration step')  ;  ax2.set_xlabel('iteration step')
plt.show()
````

### Learning Rate 증가

<figure>
 <img align="left", src="https://raw.githubusercontent.com/greyhatguy007/Machine-Learning-Specialization-Coursera/d5c95ffb5b1524367e1a9147707ec4725298d715/C1%20-%20Supervised%20Machine%20Learning%20-%20Regression%20and%20Classification/week1/Optional%20Labs/images/C1_W1_Lab03_alpha_too_big.PNG"   style="width:340px;height:240px;" >
</figure>
이번 강의에서는 식(3)에서 learning rate, $\alpha$의 적정값에 대한 논의가 있었습니다. $\alpha$가 클수록 gradient descent가 솔루션으로 수렴되는 속도가 빨라집니다. 그러나 너무 크면 gradient descent가 발산합니다. 위에는 훌륭하게 수렴되는 솔루션의 예가 있습니다.

$\alpha$의 값을 늘려서 무슨 일이 일어나는지 살펴보겠습니다.

````python
# initialize parameters
w_init = 0
b_init = 0
# set alpha to a large value
iterations = 10
tmp_alpha = 8.0e-1
# run gradient descent
w_final, b_final, J_hist, p_hist = gradient_descent(x_train ,y_train, w_init, b_init, tmp_alpha,
                                                    iterations, compute_cost, compute_gradient)
````

위에서 $w$ 및 $b$는 각 반복마다 절대값이 증가하면서 양수와 음수 사이를 오가며 움직입니다. 또한 $\frac{\partial J(w,b)}{\partial w}$를 반복할 때마다 부호가 변경되고 비용은 감소하기보다는 증가합니다. 이는 *learning rate가 너무 크고* 발산하고 있다는 분명한 신호입니다. 이를 플롯으로 시각화해 보겠습니다.

````python
plt_divergence(p_hist, J_hist,x_train, y_train)
plt.show()
````

위의 왼쪽 그래프는 gradient descent의 처음 몇 단계에 대한 $w$의 진행 상황을 보여줍니다. $w$는 양수와 음수 사이를 진동하고 비용이 빠르게 증가합니다. Gradient Descent는 $w$와 $b$ 모두에서 동시에 작동하므로 전체 그림을 보려면 오른쪽에 3D 플롯이 필요합니다.

## 3.1 요약
이 과정에서 사용하게 될 <img align="right" src="https://raw.githubusercontent.com/greyhatguy007/Machine-Learning-Specialization-Coursera/d5c95ffb5b1524367e1a9147707ec4725298d715/C1%20-%20Supervised%20Machine%20Learning%20-%20Regression%20and%20Classification/week2/Optional%20Labs/images/C1_W2_Lab04_Vectors.PNG" style="width:340px;" >벡터는 순서가 지정된 숫자 배열입니다. 표기법에서 벡터는 $\mathbf{x}$와 같이 굵은 소문자로 표시됩니다.  벡터의 요소는 모두 동일한 유형입니다. 예를 들어 벡터에는 문자와 숫자가 모두 포함되지 않습니다. 수학자들은 *순위*를 선호하지만 배열의 요소 수는 종종 *차원*이라고 합니다. 표시된 벡터의 차원은 $n$입니다. 벡터의 요소는 인덱스로 참조될 수 있습니다. 수학 설정에서 인덱스는 일반적으로 1부터 n까지 실행됩니다. 컴퓨터 과학 및 이러한 실습에서 인덱싱은 일반적으로 0에서 n-1까지 실행됩니다.  표기법에서 벡터의 요소는 개별적으로 참조될 때 아래 첨자의 인덱스를 나타냅니다. 예를 들어 벡터 $\mathbf{x}$의 $0^{th}$ 요소는 $x_0$입니다. 이 경우 x는 굵게 표시되지 않습니다.

<a name="toc_40015_3.2"></a>
## 3.2 NumPy 어레이

NumPy의 기본 데이터 구조는 동일한 유형(`dtype`)의 요소를 포함하는 색인 ​​가능한 n차원 *배열*입니다. '차원'이라는 용어가 너무 많이 사용되었음을 바로 알 수 있습니다. 위에서는 벡터의 요소 수였지만 여기서 차원은 배열의 인덱스 수를 나타냅니다. 1차원 또는 1차원 배열에는 하나의 인덱스가 있습니다. 코스 1에서는 벡터를 NumPy 1차원 배열로 표현하겠습니다.

- 1차원 배열, 형태(n,): [0]부터 [n-1]까지 인덱스가 있는 n개 요소

<a name="toc_40015_3.3"></a>
## 3.3 벡터 생성

NumPy의 데이터 생성 루틴은 일반적으로 객체의 모양인 첫 번째 매개변수를 갖습니다. 이는 1차원 결과에 대한 단일 값이거나 결과의 모양을 지정하는 튜플(n,m,...)일 수 있습니다. 다음은 이러한 루틴을 사용하여 벡터를 생성하는 예입니다.

````python
# NumPy routines which allocate memory and fill arrays with value
a = np.zeros(4);                print(f"np.zeros(4) :   a = {a}, a shape = {a.shape}, a data type = {a.dtype}")
a = np.zeros((4,));             print(f"np.zeros(4,) :  a = {a}, a shape = {a.shape}, a data type = {a.dtype}")
a = np.random.random_sample(4); print(f"np.random.random_sample(4): a = {a}, a shape = {a.shape}, a data type = {a.dtype}")
````

일부 데이터 생성 루틴은 모양 튜플을 사용하지 않습니다.

````python
# NumPy routines which allocate memory and fill arrays with value but do not accept shape as input argument
a = np.arange(4.);              print(f"np.arange(4.):     a = {a}, a shape = {a.shape}, a data type = {a.dtype}")
a = np.random.rand(4);          print(f"np.random.rand(4): a = {a}, a shape = {a.shape}, a data type = {a.dtype}")
````

값을 수동으로 지정할 수도 있습니다.

````python
# NumPy routines which allocate memory and fill with user specified values
a = np.array([5,4,3,2]);  print(f"np.array([5,4,3,2]):  a = {a},     a shape = {a.shape}, a data type = {a.dtype}")
a = np.array([5.,4,3,2]); print(f"np.array([5.,4,3,2]): a = {a}, a shape = {a.shape}, a data type = {a.dtype}")
````

이것들은 모두 4개의 요소를 가진 1차원 벡터 `a`를 생성했습니다. `a.shape`는 치수를 반환합니다. 여기서는 4개 요소가 있는 1차원 배열을 나타내는 a.shape = `(4,)`를 볼 수 있습니다.

## 3.4 벡터 연산
벡터를 사용하여 몇 가지 작업을 살펴보겠습니다.
<a name="toc_40015_3.4.1"></a>
### 3.4.1 인덱싱
벡터의 요소는 인덱싱 및 슬라이싱을 통해 액세스할 수 있습니다. NumPy는 매우 완벽한 인덱싱 및 슬라이싱 기능 세트를 제공합니다. 여기서는 강좌에 필요한 기본 사항만 살펴보겠습니다. 자세한 내용은 [슬라이싱 및 인덱싱](https://NumPy.org/doc/stable/reference/arrays.indexing.html)을 참조하세요. **인덱싱**은 배열 내의 위치를 ​​기준으로 배열의 *요소*를 참조하는 것을 의미합니다. **슬라이싱**은 인덱스를 기반으로 배열에서 요소의 *하위 집합*을 가져오는 것을 의미합니다. NumPy는 0부터 인덱싱을 시작하므로 벡터 $\mathbf{a}$의 세 번째 요소는 `a[2]`입니다.

````python
#vector indexing operations on 1-D vectors
a = np.arange(10)
print(a)

#access an element
print(f"a[2].shape: {a[2].shape} a[2]  = {a[2]}, Accessing an element returns a scalar")

# access the last element, negative indexes count from the end
print(f"a[-1] = {a[-1]}")

#indexs must be within the range of the vector or they will produce and error
try:
    c = a[10]
except Exception as e:
    print("The error message you'll see is:")
    print(e)
````

<a name="toc_40015_3.4.2"></a>
### 3.4.2 슬라이싱
슬라이싱은 세 가지 값 세트(`start:stop:step`)를 사용하여 인덱스 배열을 생성합니다. 값의 하위 집합도 유효합니다. 그 사용법은 다음 예를 통해 가장 잘 설명됩니다.

````python
#vector slicing operations
a = np.arange(10)
print(f"a         = {a}")

#access 5 consecutive elements (start:stop:step)
c = a[2:7:1];     print("a[2:7:1] = ", c)

# access 3 elements separated by two
c = a[2:7:2];     print("a[2:7:2] = ", c)

# access all elements index 3 and above
c = a[3:];        print("a[3:]    = ", c)

# access all elements below index 3
c = a[:3];        print("a[:3]    = ", c)

# access all elements
c = a[:];         print("a[:]     = ", c)
````

<a name="toc_40015_3.4.3"></a>
### 3.4.3 단일 벡터 연산
단일 벡터에 대한 연산과 관련된 유용한 연산이 많이 있습니다.

````python
a = np.array([1,2,3,4])
print(f"a             : {a}")
# negate elements of a
b = -a
print(f"b = -a        : {b}")

# sum all elements of a, returns a scalar
b = np.sum(a)
print(f"b = np.sum(a) : {b}")

b = np.mean(a)
print(f"b = np.mean(a): {b}")

b = a**2
print(f"b = a**2      : {b}")
````

### 3.4.4 벡터 벡터 요소별 연산
NumPy 산술, 논리 및 비교 연산의 대부분은 벡터에도 적용됩니다. 이러한 연산자는 요소별로 작동합니다. 예를 들어
$$ \mathbf{a} + \mathbf{b} = \sum_{i=0}^{n-1} a_i + b_i $$

````python
a = np.array([ 1, 2, 3, 4])
b = np.array([-1,-2, 3, 4])
print(f"Binary operators work element wise: {a + b}")
````

물론 이것이 올바르게 작동하려면 벡터의 크기가 동일해야 합니다.

````python
#try a mismatched vector operation
c = np.array([1, 2])
try:
    d = a + c
except Exception as e:
    print("The error message you'll see is:")
    print(e)
````

### 3.4.6 벡터 벡터 내적
내적은 선형대수학과 NumPy의 핵심입니다. 이는 이 과정에서 광범위하게 사용되는 작업이므로 잘 이해해야 합니다. 내적은 아래와 같습니다.

<img src="https://raw.githubusercontent.com/greyhatguy007/Machine-Learning-Specialization-Coursera/d5c95ffb5b1524367e1a9147707ec4725298d715/C1%20-%20Supervised%20Machine%20Learning%20-%20Regression%20and%20Classification/week2/Optional%20Labs/images/C1_W2_Lab04_dot_notrans.gif" width=800>

내적은 두 벡터의 값을 요소별로 곱한 다음 결과를 합산합니다. 벡터 내적을 위해서는 두 벡터의 차원이 동일해야 합니다.

아래 내적의 자체 버전을 구현해 보겠습니다.

**for 루프를 사용하여** 두 벡터의 내적을 반환하는 함수를 구현합니다. 주어진 입력 $a$ 및 $b$를 반환하는 함수:
$$ x = \sum_{i=0}^{n-1} a_i b_i $$
`a`와 `b`가 모두 동일한 모양이라고 가정합니다.

### 참조 솔루션: 한 번에 한 줄씩

아래의 각 연습은 다음 코드 셀의 한 줄에 하나의 표현식으로 해결됩니다. 번호가 매겨진 참조는 해당 표현식을 코드 순서대로 설명합니다. 학생용 버전과 비교할 때 주변 루프와 반환 문을 유지하세요.

참고 표기: `m`는 예시 개수입니다. `i`는 예시를 선택합니다. `j`는 feature를 선택합니다. 예측값은 모델 출력을 의미합니다. 기울기는 매개변수가 변경될 때 비용이 어떻게 변경되는지를 의미합니다.

1. **이 값 쌍을 내적에 추가하세요**

누계 + $a_i b_i$. 누계의 이름은 `x`입니다. 그것을 추가하기 전에 쌍을 곱하십시오.

**함수 참조:** [내적](https://numpy.org/doc/stable/reference/generated/numpy.dot.html), [열 평균 및 축](https://numpy.org/doc/stable/reference/generated/numpy.mean.html), [지수](https://numpy.org/doc/stable/reference/generated/numpy.exp.html). 아래 원본 실습에는 전체 수학적 맥락이 포함되어 있습니다.

출처: [C1_W2_Lab01_Python_Numpy_Vectorization_Soln.ipynb](https://github.com/greyhatguy007/Machine-Learning-Specialization-Coursera/blob/d5c95ffb5b1524367e1a9147707ec4725298d715/C1%20-%20Supervised%20Machine%20Learning%20-%20Regression%20and%20Classification/week2/Optional%20Labs/C1_W2_Lab01_Python_Numpy_Vectorization_Soln.ipynb)

````python
def my_dot(a, b):
    """
   Compute the dot product of two vectors

    Args:
      a (ndarray (n,)):  input vector
      b (ndarray (n,)):  input vector with same dimension as a

    Returns:
      x (scalar):
    """
    x=0
    for i in range(a.shape[0]):
        x = x + b[i] * a[i]
    return x
````

````python
# Verify the reference implementation
np.testing.assert_allclose(my_dot(np.array([2., -1., 3.]), np.array([4., 5., -2.])), -3.)
print("Check passed.")
````

````python
# test 1-D
a = np.array([1, 2, 3, 4])
b = np.array([-1, 4, 3, 2])
print(f"my_dot(a, b) = {my_dot(a, b)}")
````

내적은 스칼라 값을 반환할 것으로 예상됩니다.

`np.dot`를 사용하여 동일한 작업을 시도해 보겠습니다.

````python
# test 1-D
a = np.array([1, 2, 3, 4])
b = np.array([-1, 4, 3, 2])
c = np.dot(a, b)
print(f"NumPy 1-D np.dot(a, b) = {c}, np.dot(a, b).shape = {c.shape} ")
c = np.dot(b, a)
print(f"NumPy 1-D np.dot(b, a) = {c}, np.dot(a, b).shape = {c.shape} ")
````

위에서 1D 결과가 우리의 구현과 일치한다는 것을 알 수 있습니다.

### 3.4.7 속도의 필요성: 벡터 대 for 루프
우리는 속도 메모리 효율성을 향상시키기 위해 NumPy 라이브러리를 활용했습니다. 다음을 보여드리겠습니다.

````python
np.random.seed(1)
a = np.random.rand(10000000)  # very large arrays
b = np.random.rand(10000000)

tic = time.time()  # capture start time
c = np.dot(a, b)
toc = time.time()  # capture end time

print(f"np.dot(a, b) =  {c:.4f}")
print(f"Vectorized version duration: {1000*(toc-tic):.4f} ms ")

tic = time.time()  # capture start time
c = my_dot(a,b)
toc = time.time()  # capture end time

print(f"my_dot(a, b) =  {c:.4f}")
print(f"loop version duration: {1000*(toc-tic):.4f} ms ")

del(a);del(b)  #remove these big arrays from memory
````

따라서 이 예에서는 벡터화를 통해 속도가 크게 향상됩니다. 이는 NumPy가 기본 하드웨어에서 사용 가능한 데이터 병렬성을 더 잘 활용하기 때문입니다. GPU와 최신 CPU는 SIMD(Single Instruction, Multiple Data) 파이프라인을 구현하여 여러 작업을 병렬로 실행할 수 있습니다. 이는 데이터 세트가 매우 큰 경우가 많은 기계 학습에서 매우 중요합니다.

### 3.4.8 코스 1의 벡터 벡터 연산
벡터 벡터 연산은 강좌 1에서 자주 등장합니다. 그 이유는 다음과 같습니다.
- 앞으로 예제는 (m,n) 차원의 배열 `X_train`에 저장됩니다. 이에 대해서는 문맥을 통해 더 자세히 설명하겠지만 여기서는 이것이 2차원 배열 또는 행렬이라는 점에 유의하는 것이 중요합니다(행렬에 대한 다음 섹션 참조).
- `w`는 (n,) 모양의 1차원 벡터가 됩니다.
- 예제를 반복하면서 작업을 수행하고 X를 인덱싱하여 개별적으로 작업할 각 예제를 추출합니다. 예:`X[i]`

설명이 다소 길지만 벡터 연산을 수행할 때는 피연산자의 모양을 정렬하고 이해하는 것이 중요합니다.

````python
# show common Course 1 example
X = np.array([[1],[2],[3],[4]])
w = np.array([2])
c = np.dot(X[1], w)

print(f"X[1] has shape {X[1].shape}")
print(f"w has shape {w.shape}")
print(f"c has shape {c.shape}")
````

## 4.3 행렬 생성
1차원 벡터를 생성한 동일한 함수가 2차원 또는 n차원 배열을 생성합니다. 다음은 몇 가지 예입니다.

아래에서는 2차원 결과를 얻기 위해 모양 튜플이 제공됩니다. NumPy가 대괄호를 사용하여 각 치수를 표시하는 방법을 확인하세요. NumPy보다 더 주의할 점은 출력할 때 한 줄에 한 행씩 출력된다는 점입니다.

## 4.4 행렬 연산
행렬을 사용하여 몇 가지 연산을 살펴보겠습니다.

### 4.4.1 인덱싱

행렬에는 두 번째 인덱스가 포함됩니다. 두 인덱스는 [행, 열]을 설명합니다. 인덱싱으로는 요소나 행/열을 반환할 수 있습니다. 아래를 참조하세요:

````python
#vector indexing operations on matrices
a = np.arange(6).reshape(-1, 2)   #reshape is a convenient way to create matrices
print(f"a.shape: {a.shape}, \na= {a}")

#access an element
print(f"\na[2,0].shape:   {a[2, 0].shape}, a[2,0] = {a[2, 0]},     type(a[2,0]) = {type(a[2, 0])} Accessing an element returns a scalar\n")

#access a row
print(f"a[2].shape:   {a[2].shape}, a[2]   = {a[2]}, type(a[2])   = {type(a[2])}")
````

마지막 예에 주목할 가치가 있습니다. 행을 지정하여 행렬에 액세스하면 *1D 벡터*가 반환됩니다.

**Reshape** 이전 예제에서는 [reshape](https://numpy.org/doc/stable/reference/generated/numpy.reshape.html)를 사용하여 배열의 모양을 만들었습니다. `a = np.arange(6).reshape(-1, 2) ` 이 코드 줄은 먼저 6개 요소의 *1-D 벡터*를 생성했습니다. 그런 다음 reshape 명령을 사용하여 해당 벡터를 *2D* 배열로 재구성했습니다. 다음과 같이 작성할 수 있습니다. `a = np.arange(6).reshape(3, 2) ` 동일한 3행, 2열 배열에 도달합니다. -1 인수는 배열 크기와 열 수에 따라 행 수를 계산하도록 루틴에 지시합니다.

<a name="toc_40015_4.4.2"></a>
### 4.4.2 슬라이싱
슬라이싱은 세 가지 값 세트(`start:stop:step`)를 사용하여 인덱스 배열을 생성합니다. 값의 하위 집합도 유효합니다. 그 사용법은 다음 예를 통해 가장 잘 설명됩니다.

````python
#vector 2-D slicing operations
a = np.arange(20).reshape(-1, 10)
print(f"a = \n{a}")

#access 5 consecutive elements (start:stop:step)
print("a[0, 2:7:1] = ", a[0, 2:7:1], ",  a[0, 2:7:1].shape =", a[0, 2:7:1].shape, "a 1-D array")

#access 5 consecutive elements (start:stop:step) in two rows
print("a[:, 2:7:1] = \n", a[:, 2:7:1], ",  a[:, 2:7:1].shape =", a[:, 2:7:1].shape, "a 2-D array")

# access all elements
print("a[:,:] = \n", a[:,:], ",  a[:,:].shape =", a[:,:].shape)

# access all elements in one row (very common usage)
print("a[1,:] = ", a[1,:], ",  a[1,:].shape =", a[1,:].shape, "a 1-D array")
# same as
print("a[1]   = ", a[1],   ",  a[1].shape   =", a[1].shape, "a 1-D array")
````

# 2 문제 설명

주택 가격 예측값의 동기를 부여하는 예를 사용하게 됩니다. 학습 데이터 세트에는 아래 표에 표시된 4개의 feature(크기, 침실, 층 및 연령)가 있는 3개의 예가 포함되어 있습니다.  이전 실습과 달리 크기는 1000평방피트가 아닌 평방피트 단위입니다. 이로 인해 문제가 발생하며 다음 실습에서 이 문제를 해결할 것입니다!

| 크기(평방피트) | 침실 수 | 층수 | 집의 나이 | 가격(천달러) |   
| ---| ------ |---- |---------------|--------------- |  
| 2104 | 5 | 1 | 45 | 460 |  
| 1416 | 3 | 2 | 40 | 232 |  
| 852 | 2 | 1 | 35 | 178 |

이러한 값을 사용하여 linear regression 모델을 구축하면 다른 주택의 가격을 예측할 수 있습니다. 예를 들어, 1200제곱피트, 침실 3개, 1층, 40년 된 주택입니다.

다음 코드 셀을 실행하여 `X_train` 및 `y_train` 변수를 생성하십시오.

````python
X_train = np.array([[2104, 5, 1, 45], [1416, 3, 2, 40], [852, 2, 1, 35]])
y_train = np.array([460, 232, 178])
````

<a name="toc_15456_2.1"></a>
## 2.1 예제가 포함된 행렬 X
위의 표와 유사하게 예제는 NumPy 행렬 `X_train`에 저장됩니다. 행렬의 각 행은 하나의 예를 나타냅니다. $m$ 학습 예제(이 예제에서는 $m$는 3개)가 있고 $n$ feature(이 예제에서는 4개)가 있는 경우 $\mathbf{X}$는 차원($m$, $n$)(m행, n열)이 있는 행렬입니다.


$$\mathbf{X} =
\begin{pmatrix}
 x^{(0)}_0 & x^{(0)}_1 & \cdots & x^{(0)}_{n-1} \\
 x^{(1)}_0 & x^{(1)}_1 & \cdots & x^{(1)}_{n-1} \\
 \cdots \\
 x^{(m-1)}_0 & x^{(m-1)}_1 & \cdots & x^{(m-1)}_{n-1}
\end{pmatrix}
$$
표기법:
- $\mathbf{x}^{(i)}$는 예시 i를 포함하는 벡터입니다. $\mathbf{x}^{(i)}$ $ = (x^{(i)}_0, x^{(i)}_1, \cdots,x^{(i)}_{n-1})$
- $x^{(i)}_j$는 예 i의 요소 j입니다. 괄호 안의 위 첨자는 예제 번호를 나타내고 아래 첨자는 요소를 나타냅니다.

입력 데이터를 표시합니다.

````python
# data is stored in numpy array/matrix
print(f"X Shape: {X_train.shape}, X Type:{type(X_train)})")
print(X_train)
print(f"y Shape: {y_train.shape}, y Type:{type(y_train)})")
print(y_train)
````

<a name="toc_15456_2.2"></a>
## 2.2 매개변수 벡터 w, b

* $\mathbf{w}$는 $n$ 요소로 구성된 벡터입니다.
  - 각 요소에는 하나의 feature와 관련된 매개변수가 포함되어 있습니다.
  - 우리 데이터세트에서 n은 4입니다.
  - 개념적으로는 이것을 열 벡터로 그립니다.

$$\mathbf{w} = \begin{pmatrix}
w_0 \\
w_1 \\
\cdots\\
w_{n-1}
\end{pmatrix}
$$
* $b$는 스칼라 매개변수입니다.

데모를 위해 $\mathbf{w}$ 및 $b$는 최적에 가까운 일부 초기 선택된 값으로 로드됩니다. $\mathbf{w}$는 1차원 NumPy 벡터입니다.

````python
b_init = 785.1811367994083
w_init = np.array([ 0.39133535, 18.75376741, -53.36032453, -26.42131618])
print(f"w_init shape: {w_init.shape}, b_init type: {type(b_init)}")
````

<a name="toc_15456_3"></a>
# 3 다중 변수가 있는 모델 예측
여러 변수가 있는 모델의 예측값은 선형 모델에 의해 제공됩니다.

$$ f_{\mathbf{w},b}(\mathbf{x}) =  w_0x_0 + w_1x_1 +... + w_{n-1}x_{n-1} + b \tag{1}$$
또는 벡터 표기법으로:
$$ f_{\mathbf{w},b}(\mathbf{x}) = \mathbf{w} \cdot \mathbf{x} + b  \tag{2} $$
여기서 $\cdot$는 벡터 `dot product`입니다.

내적을 시연하기 위해 (1)과 (2)를 사용하여 예측값을 구현하겠습니다.

<a name="toc_15456_3.1"></a>
## 3.1 요소별 계산으로 예측
이전 예측값은 하나의 feature 값에 하나의 매개변수를 곱하고 편향 매개변수를 추가했습니다. 예측값의 이전 구현을 여러 feature로 직접 확장하는 것은 각 요소에 대한 루프를 사용하여 위의 (1)을 구현하고 해당 매개변수와 곱셈을 수행한 다음 끝에 편향 매개변수를 추가하는 것입니다.

### 참조 솔루션: 한 번에 한 줄씩

아래의 각 연습은 다음 코드 셀의 한 줄에 하나의 표현식으로 해결됩니다. 번호가 매겨진 참조는 해당 표현식을 코드 순서대로 설명합니다. 학생용 버전과 비교할 때 주변 루프와 반환 문을 유지하세요.

참고 표기: `m`는 예시 개수입니다. `i`는 예시를 선택합니다. `j`는 feature를 선택합니다. 예측값은 모델 출력을 의미합니다. 기울기는 매개변수가 변경될 때 비용이 어떻게 변경되는지를 의미합니다.

1. **이 feature 기여도를 계산하세요**

$x_i w_i$. 현재 feature `x[i]`와 가중치 `w[i]`를 사용하세요.

2. **이 feature 기여를 추가하세요**

새로운 예측값 총계 = 이전 `p` + 이번 feature의 `p_i`.

3. **예측값에 편향를 추가합니다**

최종 예측값 = 누적 가중치 feature 총계 `p` + 편향 `b`.

**함수 참조:** [내적](https://numpy.org/doc/stable/reference/generated/numpy.dot.html), [열 평균 및 축](https://numpy.org/doc/stable/reference/generated/numpy.mean.html), [지수](https://numpy.org/doc/stable/reference/generated/numpy.exp.html). 아래 원본 실습에는 전체 수학적 맥락이 포함되어 있습니다.

출처: [C1_W2_Lab02_Multiple_Variable_Soln.ipynb](https://github.com/greyhatguy007/Machine-Learning-Specialization-Coursera/blob/d5c95ffb5b1524367e1a9147707ec4725298d715/C1%20-%20Supervised%20Machine%20Learning%20-%20Regression%20and%20Classification/week2/Optional%20Labs/C1_W2_Lab02_Multiple_Variable_Soln.ipynb)

````python
def predict_single_loop(x, w, b):
    """
    single predict using linear regression

    Args:
      x (ndarray): Shape (n,) example with multiple features
      w (ndarray): Shape (n,) model parameters
      b (scalar):  model parameter

    Returns:
      p (scalar):  prediction
    """
    n = x.shape[0]
    p = 0
    for i in range(n):
        p_i = w[i] * x[i]
        p = p_i + p
    p = b + p
    return p
````

````python
# Verify the reference implementation
np.testing.assert_allclose(predict_single_loop(np.array([2., -1.]), np.array([3., 4.]), 5.), 7.)
print("Check passed.")
````

````python
# get a row from our training data
x_vec = X_train[0,:]
print(f"x_vec shape {x_vec.shape}, x_vec value: {x_vec}")

# make a prediction
f_wb = predict_single_loop(x_vec, w_init, b_init)
print(f"f_wb shape {np.shape(f_wb)}, prediction: {f_wb}")
````

`x_vec`의 모양을 확인하세요. 이는 4개 요소(4,)를 포함하는 1D NumPy 벡터입니다. 결과 `f_wb`는 스칼라입니다.

<a name="toc_15456_3.2"></a>
## 3.2 벡터 연산으로 예측

위의 방정식 (1)은 위의 (2)와 같이 내적을 사용하여 구현될 수 있습니다. 예측값 속도를 높이기 위해 벡터 연산을 사용할 수 있습니다.

Python/Numpy 실습에서 NumPy `np.dot()`[[링크](https://numpy.org/doc/stable/reference/generated/numpy.dot.html)]를 사용하여 벡터 내적을 수행할 수 있다는 점을 상기해 보세요.

### 참조 솔루션: 한 번에 한 줄씩

아래의 각 연습은 다음 코드 셀의 한 줄에 하나의 표현식으로 해결됩니다. 번호가 매겨진 참조는 해당 표현식을 코드 순서대로 설명합니다. 학생용 버전과 비교할 때 주변 루프와 반환 문을 유지하세요.

참고 표기: `m`는 예시 개수입니다. `i`는 예시를 선택합니다. `j`는 feature를 선택합니다. 예측값은 모델 출력을 의미합니다. 기울기는 매개변수가 변경될 때 비용이 어떻게 변경되는지를 의미합니다.

1. **모든 feature에서 예측값을 계산합니다**

$\hat y=\sum_j x_j w_j+b$. `np.dot`는 입력 벡터 `x`와 가중 벡터 `w`를 하나의 숫자로 결합합니다.

**함수 참조:** [내적](https://numpy.org/doc/stable/reference/generated/numpy.dot.html), [열 평균 및 축](https://numpy.org/doc/stable/reference/generated/numpy.mean.html), [지수](https://numpy.org/doc/stable/reference/generated/numpy.exp.html). 아래 원본 실습에는 전체 수학적 맥락이 포함되어 있습니다.

출처: [C1_W2_Lab02_Multiple_Variable_Soln.ipynb](https://github.com/greyhatguy007/Machine-Learning-Specialization-Coursera/blob/d5c95ffb5b1524367e1a9147707ec4725298d715/C1%20-%20Supervised%20Machine%20Learning%20-%20Regression%20and%20Classification/week2/Optional%20Labs/C1_W2_Lab02_Multiple_Variable_Soln.ipynb)

````python
def predict(x, w, b):
    """
    single predict using linear regression
    Args:
      x (ndarray): Shape (n,) example with multiple features
      w (ndarray): Shape (n,) model parameters
      b (scalar):             model parameter

    Returns:
      p (scalar):  prediction
    """
    p = float(x @ w) + b
    return p
````

````python
# Verify the reference implementation
np.testing.assert_allclose(predict(np.array([2., -1.]), np.array([3., 4.]), 5.), 7.)
print("Check passed.")
````

````python
# get a row from our training data
x_vec = X_train[0,:]
print(f"x_vec shape {x_vec.shape}, x_vec value: {x_vec}")

# make a prediction
f_wb = predict(x_vec,w_init, b_init)
print(f"f_wb shape {np.shape(f_wb)}, prediction: {f_wb}")
````

결과와 형태는 Looping을 사용한 이전 버전과 동일합니다. 앞으로는 `np.dot`가 이러한 작업에 사용될 것입니다. 예측값은 이제 단일 문입니다. 대부분의 루틴은 별도의 예측 루틴을 호출하는 대신 이를 직접 구현합니다.

<a name="toc_15456_4"></a>
# 4 여러 변수를 사용하여 비용 계산
여러 변수 $J(\mathbf{w},b)$가 있는 cost function의 방정식은 다음과 같습니다.
$$J(\mathbf{w},b) = \frac{1}{2m} \sum\limits_{i = 0}^{m-1} (f_{\mathbf{w},b}(\mathbf{x}^{(i)}) - y^{(i)})^2 \tag{3}$$
여기서:
$$ f_{\mathbf{w},b}(\mathbf{x}^{(i)}) = \mathbf{w} \cdot \mathbf{x}^{(i)} + b  \tag{4} $$


이전 실습과 달리 $\mathbf{w}$ 및 $\mathbf{x}^{(i)}$는 여러 feature를 지원하는 스칼라가 아닌 벡터입니다.

아래는 방정식 (3)과 (4)를 구현한 것입니다. 이는 모든 `m` 예제에 대한 for 루프가 사용되는 *이 과정*의 표준 패턴을 사용한다는 점에 유의하세요.

### 참조 솔루션: 한 번에 한 줄씩

아래의 각 연습은 다음 코드 셀의 한 줄에 하나의 표현식으로 해결됩니다. 번호가 매겨진 참조는 해당 표현식을 코드 순서대로 설명합니다. 학생용 버전과 비교할 때 주변 루프와 반환 문을 유지하세요.

참고 표기: `m`는 예시 개수입니다. `i`는 예시를 선택합니다. `j`는 feature를 선택합니다. 예측값은 모델 출력을 의미합니다. 기울기는 매개변수가 변경될 때 비용이 어떻게 변경되는지를 의미합니다.

1. **예측값을 계산합니다**

하나의 feature에 대해 $\hat y = wx+b$; 여러 feature의 경우 $\hat y = \sum_j x_j w_j+b$. `i`로 인덱싱된 현재 예제를 사용하세요.

2. **합계에 제곱오차를 더합니다**

새로운 합계 = 기존 `cost` + $(\hat y_i-y_i)^2$. 이전 줄에서 예측값 변수를 사용합니다.

3. **평균 비용 계산**

$J = \frac{1}{2m}\sum_i(\hat y_i-y_i)^2$. 누적된 오류를 `m`의 두 배로 나눕니다.

**함수 참조:** [내적](https://numpy.org/doc/stable/reference/generated/numpy.dot.html), [열 평균 및 축](https://numpy.org/doc/stable/reference/generated/numpy.mean.html), [지수](https://numpy.org/doc/stable/reference/generated/numpy.exp.html). 아래 원본 실습에는 전체 수학적 맥락이 포함되어 있습니다.

출처: [C1_W2_Lab02_Multiple_Variable_Soln.ipynb](https://github.com/greyhatguy007/Machine-Learning-Specialization-Coursera/blob/d5c95ffb5b1524367e1a9147707ec4725298d715/C1%20-%20Supervised%20Machine%20Learning%20-%20Regression%20and%20Classification/week2/Optional%20Labs/C1_W2_Lab02_Multiple_Variable_Soln.ipynb)

````python
def compute_cost(X, y, w, b):
    """
    compute cost
    Args:
      X (ndarray (m,n)): Data, m examples with n features
      y (ndarray (m,)) : target values
      w (ndarray (n,)) : model parameters
      b (scalar)       : model parameter

    Returns:
      cost (scalar): cost
    """
    m = X.shape[0]
    cost = 0.0
    for i in range(m):
        f_wb_i = float(X[i] @ w) + b           #(n,)(n,) = scalar (see np.dot)
        cost = cost + np.square(y[i] - f_wb_i)       #scalar
    cost = cost / (2.0 * m)                      #scalar
    return cost
````

````python
# Verify the reference implementation
np.testing.assert_allclose(compute_cost(np.array([[1., 2.], [3., 4.]]), np.array([2., 4.]), np.zeros(2), 0.), 5.)
print("Check passed.")
````

````python
# Compute and display cost using our pre-chosen optimal parameters.
cost = compute_cost(X_train, y_train, w_init, b_init)
print(f'Cost at optimal w : {cost}')
````

**예상 결과**: 최적의 비용 w : 1.5578904045996674e-12

# 5 다중 변수가 있는 Gradient Descent
다중 변수의 경우 Gradient descent:

$$\begin{align*} \text{repeat}&\text{ until convergence:} \; \lbrace \newline\;
& w_j = w_j -  \alpha \frac{\partial J(\mathbf{w},b)}{\partial w_j} \tag{5}  \; & \text{for j = 0..n-1}\newline
&b\ \ = b -  \alpha \frac{\partial J(\mathbf{w},b)}{\partial b}  \newline \rbrace
\end{align*}$$

여기서 n은 feature의 수이고, 매개변수 $w_j$, $b$는 동시에 업데이트됩니다.

$$
\begin{align}
\frac{\partial J(\mathbf{w},b)}{\partial w_j}  &= \frac{1}{m} \sum\limits_{i = 0}^{m-1} (f_{\mathbf{w},b}(\mathbf{x}^{(i)}) - y^{(i)})x_{j}^{(i)} \tag{6}  \\
\frac{\partial J(\mathbf{w},b)}{\partial b}  &= \frac{1}{m} \sum\limits_{i = 0}^{m-1} (f_{\mathbf{w},b}(\mathbf{x}^{(i)}) - y^{(i)}) \tag{7}
\end{align}
$$
* m은 데이터 세트의 학습 예제 수입니다.

    
* $f_{\mathbf{w},b}(\mathbf{x}^{(i)})$는 해당 모델의 예측값이고, $y^{(i)}$는 target 값입니다.

<a name="toc_15456_5.1"></a>
## 5.1 여러 변수를 사용하여 기울기 계산
방정식 (6)과 (7)을 계산하기 위한 구현은 다음과 같습니다. 이를 구현하는 방법에는 여러 가지가 있습니다. 이 버전에는
- 모든 m개 예제에 대한 외부 루프.
    - 예시의 $\frac{\partial J(\mathbf{w},b)}{\partial b}$는 직접 계산하여 누적할 수 있습니다.
    - 모든 n개의 feature에 대한 두 번째 루프에서:
        - $\frac{\partial J(\mathbf{w},b)}{\partial w_j}$는 각 $w_j$에 대해 계산됩니다.

### 참조 솔루션: 한 번에 한 줄씩

아래의 각 연습은 다음 코드 셀의 한 줄에 하나의 표현식으로 해결됩니다. 번호가 매겨진 참조는 해당 표현식을 코드 순서대로 설명합니다. 학생용 버전과 비교할 때 주변 루프와 반환 문을 유지하세요.

참고 표기: `m`는 예시 개수입니다. `i`는 예시를 선택합니다. `j`는 feature를 선택합니다. 예측값은 모델 출력을 의미합니다. 기울기는 매개변수가 변경될 때 비용이 어떻게 변경되는지를 의미합니다.

1. **예측값에서 실제 값을 뺀 값을 계산합니다**

$e_i=(\sum_j X_{ij}w_j+b)-y_i$. 현재 행 `X[i]`, 가중치 `w`, 편향 `b` 및 레이블 `y[i]`를 사용합니다.

2. **이 가중치 기울기를 누적합니다**

새로운 $g_j$ = 기존 $g_j + e_i X_{ij}$. 오류 이름은 이전 줄에 `err` 또는 `err_i`로 표시됩니다. 누적 합계는 `dj_dw[j]`입니다.

3. **편향 기울기 누적**

새로운 $g_b$ = 기존 $g_b+e_i$. `dj_db`에 `err` 또는 `err_i`를 추가합니다.

4. **가중치 기울기 평균을 구합니다**

누적된 `dj_dw`를 예시 개수 `m`로 나눕니다. 배열의 나눗셈은 모든 가중치에 대해 동시에 작동합니다.

5. **편향 기울기 평균화**

누적된 `dj_db`를 예제 `m`의 개수로 나눕니다.

**함수 참조:** [내적](https://numpy.org/doc/stable/reference/generated/numpy.dot.html), [열 평균 및 축](https://numpy.org/doc/stable/reference/generated/numpy.mean.html), [지수](https://numpy.org/doc/stable/reference/generated/numpy.exp.html). 아래 원본 실습에는 전체 수학적 맥락이 포함되어 있습니다.

출처: [C1_W2_Lab02_Multiple_Variable_Soln.ipynb](https://github.com/greyhatguy007/Machine-Learning-Specialization-Coursera/blob/d5c95ffb5b1524367e1a9147707ec4725298d715/C1%20-%20Supervised%20Machine%20Learning%20-%20Regression%20and%20Classification/week2/Optional%20Labs/C1_W2_Lab02_Multiple_Variable_Soln.ipynb)

````python
def compute_gradient(X, y, w, b):
    """
    Computes the gradient for linear regression
    Args:
      X (ndarray (m,n)): Data, m examples with n features
      y (ndarray (m,)) : target values
      w (ndarray (n,)) : model parameters
      b (scalar)       : model parameter

    Returns:
      dj_dw (ndarray (n,)): The gradient of the cost w.r.t. the parameters w.
      dj_db (scalar):       The gradient of the cost w.r.t. the parameter b.
    """
    m,n = X.shape           #(number of examples, number of features)
    dj_dw = np.zeros((n,))
    dj_db = 0.

    for i in range(m):
        err = float(X[i] @ w) + b - y[i]
        for j in range(n):
            dj_dw[j] = dj_dw[j] + X[i, j] * err
        dj_db = dj_db + err
    dj_dw = dj_dw / float(m)
    dj_db = dj_db / float(m)

    return dj_db, dj_dw
````

````python
# Verify the reference implementation
test_db, test_dw = compute_gradient(np.array([[1., 2.], [3., 4.]]), np.array([2., 4.]), np.zeros(2), 0.)
np.testing.assert_allclose(test_db, -3.)
np.testing.assert_allclose(test_dw, [-7., -10.])
print("Check passed.")
````

````python
#Compute and display gradient
tmp_dj_db, tmp_dj_dw = compute_gradient(X_train, y_train, w_init, b_init)
print(f'dj_db at initial w,b: {tmp_dj_db}')
print(f'dj_dw at initial w,b: \n {tmp_dj_dw}')
````

## 5.2 다중 변수가 있는 Gradient Descent
아래 루틴은 위의 방정식 (5)를 구현합니다.

### 참조 솔루션: 한 번에 한 줄씩

아래의 각 연습은 다음 코드 셀의 한 줄에 하나의 표현식으로 해결됩니다. 번호가 매겨진 참조는 해당 표현식을 코드 순서대로 설명합니다. 학생용 버전과 비교할 때 주변 루프와 반환 문을 유지하세요.

참고 표기: `m`는 예시 개수입니다. `i`는 예시를 선택합니다. `j`는 feature를 선택합니다. 예측값은 모델 출력을 의미합니다. 기울기는 매개변수가 변경될 때 비용이 어떻게 변경되는지를 의미합니다.

1. **가중치 업데이트**

$w_{new}=w-\alpha\nabla_w J$. `w`, `alpha` 및 `dj_dw`를 사용하십시오. 배열 산술은 모든 가중치를 업데이트합니다.

2. **편향 업데이트**

$b_{new}=b-\alpha\,\partial J/\partial b$. `b`, `alpha` 및 `dj_db`를 사용합니다.

**함수 참조:** [내적](https://numpy.org/doc/stable/reference/generated/numpy.dot.html), [열 평균 및 축](https://numpy.org/doc/stable/reference/generated/numpy.mean.html), [지수](https://numpy.org/doc/stable/reference/generated/numpy.exp.html). 아래 원본 실습에는 전체 수학적 맥락이 포함되어 있습니다.

출처: [C1_W2_Lab02_Multiple_Variable_Soln.ipynb](https://github.com/greyhatguy007/Machine-Learning-Specialization-Coursera/blob/d5c95ffb5b1524367e1a9147707ec4725298d715/C1%20-%20Supervised%20Machine%20Learning%20-%20Regression%20and%20Classification/week2/Optional%20Labs/C1_W2_Lab02_Multiple_Variable_Soln.ipynb)

````python
def gradient_descent(X, y, w_in, b_in, cost_function, gradient_function, alpha, num_iters):
    """
    Performs batch gradient descent to learn theta. Updates theta by taking
    num_iters gradient steps with learning rate alpha

    Args:
      X (ndarray (m,n))   : Data, m examples with n features
      y (ndarray (m,))    : target values
      w_in (ndarray (n,)) : initial model parameters
      b_in (scalar)       : initial model parameter
      cost_function       : function to compute cost
      gradient_function   : function to compute the gradient
      alpha (float)       : Learning rate
      num_iters (int)     : number of iterations to run gradient descent

    Returns:
      w (ndarray (n,)) : Updated values of parameters
      b (scalar)       : Updated value of parameter
      """

    # An array to store cost J and w's at each iteration primarily for graphing later
    J_history = []
    w = copy.deepcopy(w_in)  #avoid modifying global w within function
    b = b_in

    for i in range(num_iters):

        # Calculate the gradient and update the parameters
        dj_db,dj_dw = gradient_function(X, y, w, b)   ##None

        # Update Parameters using w, b, alpha and gradient
        w = w - dj_dw * alpha               ##None
        b = b - dj_db * alpha               ##None

        # Save cost J at each iteration
        if i<100000:      # prevent resource exhaustion
            J_history.append( cost_function(X, y, w, b))

        # Print cost every at intervals 10 times or as many iterations if < 10
        if i% math.ceil(num_iters / 10) == 0:
            print(f"Iteration {i:4d}: Cost {J_history[-1]:8.2f}   ")

    return w, b, J_history #return final w,b and J history for graphing
````

````python
# Verify the reference implementation
test_w, test_b, *_ = gradient_descent(np.array([[1., 2.], [3., 4.]]), np.array([2., 4.]), np.zeros(2), 0., compute_cost, compute_gradient, 0.1, 1)
np.testing.assert_allclose(test_w, [0.7, 1.])
np.testing.assert_allclose(test_b, 0.3)
print("Check passed.")
````

다음 셀에서는 구현을 테스트합니다.

````python
# initialize parameters
initial_w = np.zeros_like(w_init)
initial_b = 0.
# some gradient descent settings
iterations = 1000
alpha = 5.0e-7
# run gradient descent
w_final, b_final, J_hist = gradient_descent(X_train, y_train, initial_w, initial_b,
                                                    compute_cost, compute_gradient,
                                                    alpha, iterations)
print(f"b,w found by gradient descent: {b_final:0.2f},{w_final} ")
m,_ = X_train.shape
for i in range(m):
    print(f"prediction: {np.dot(X_train[i], w_final) + b_final:0.2f}, target value: {y_train[i]}")
````

````python
# plot cost versus iteration
fig, (ax1, ax2) = plt.subplots(1, 2, constrained_layout=True, figsize=(12, 4))
ax1.plot(J_hist)
ax2.plot(100 + np.arange(len(J_hist[100:])), J_hist[100:])
ax1.set_title("Cost vs. iteration");  ax2.set_title("Cost vs. iteration (tail)")
ax1.set_ylabel('Cost')             ;  ax2.set_ylabel('Cost')
ax1.set_xlabel('iteration step')   ;  ax2.set_xlabel('iteration step')
plt.show()
````

<a name="toc_15456_5"></a>
## 여러 변수가 있는 Gradient Descent
다음은 다중 변수에 대해 gradient descent의 마지막 실습에서 개발한 방정식입니다.:

$$\begin{align*} \text{repeat}&\text{ until convergence:} \; \lbrace \newline\;
& w_j := w_j -  \alpha \frac{\partial J(\mathbf{w},b)}{\partial w_j} \tag{1}  \; & \text{for j = 0..n-1}\newline
&b\ \ := b -  \alpha \frac{\partial J(\mathbf{w},b)}{\partial b}  \newline \rbrace
\end{align*}$$

여기서 n은 feature의 수이고, 매개변수 $w_j$, $b$는 동시에 업데이트됩니다.

$$
\begin{align}
\frac{\partial J(\mathbf{w},b)}{\partial w_j}  &= \frac{1}{m} \sum\limits_{i = 0}^{m-1} (f_{\mathbf{w},b}(\mathbf{x}^{(i)}) - y^{(i)})x_{j}^{(i)} \tag{2}  \\
\frac{\partial J(\mathbf{w},b)}{\partial b}  &= \frac{1}{m} \sum\limits_{i = 0}^{m-1} (f_{\mathbf{w},b}(\mathbf{x}^{(i)}) - y^{(i)}) \tag{3}
\end{align}
$$
* m은 데이터 세트의 학습 예제 수입니다.

    
* $f_{\mathbf{w},b}(\mathbf{x}^{(i)})$는 해당 모델의 예측값이고, $y^{(i)}$는 target 값입니다.

## Learning Rate

위 그림은 $w$가 고르지 않게 업데이트되는 이유를 보여줍니다.
- $\alpha$는 모든 매개변수 업데이트($w$ 및 $b$)에서 공유됩니다.
- $w$의 경우 공통 오류 항에 feature를 곱합니다. ($b$ 아님).
- feature는 ​​크기가 상당히 다양하여 일부 feature 업데이트가 다른 것보다 훨씬 빠릅니다. 이 경우 $w_0$에는 'size(sqft)'를 곱하는데, 일반적으로 > 1000이고, $w_1$에는 '침실 수'(일반적으로 2~4)를 곱합니다.
    
해결책은 Feature Scaling입니다.

강의에서는 세 가지 다른 기술에 대해 논의했습니다.
- Feature Scaling: 양수인 feature를 최대값으로 나누거나, 일반적으로 (x-min)/(max-min)을 사용합니다. 이때 값은 0~1 범위가 됩니다. 첫 번째 방법은 양수인 feature에, 두 번째 방법은 음수가 포함된 feature에도 사용할 수 있습니다. 두 방법 모두 최대값과 최소값이 같은 상수 feature는 별도 처리가 필요합니다.
- 평균 정규화: $x_i := \dfrac{x_i - \mu_i}{max - min} $
- 아래에서 살펴볼 Z-score 정규화.

### z-score 정규화
z-score 정규화 후에는 모든 feature의 평균은 0, 표준편차는 1이 됩니다.

Z-score 정규화를 구현하려면 다음 수식에 표시된 대로 입력 값을 조정하세요.
$$x^{(i)}_j = \dfrac{x^{(i)}_j - \mu_j}{\sigma_j} \tag{4}$$
여기서 $j$는 feature 또는 $\mathbf{X}$ 행렬의 열을 선택합니다. $µ_j$는 feature(j)에 대한 모든 값의 평균이고 $\sigma_j$는 feature(j)의 표준편차입니다.
$$
\begin{align}
\mu_j &= \frac{1}{m} \sum_{i=0}^{m-1} x^{(i)}_j \tag{5}\\
\sigma^2_j &= \frac{1}{m} \sum_{i=0}^{m-1} (x^{(i)}_j - \mu_j)^2  \tag{6}
\end{align}
$$

>**구현 참고:** feature를 정규화할 때 정규화에 사용된 값, 즉 계산에 사용된 평균값과 표준 편차를 저장하는 것이 중요합니다. 모델에서 매개변수를 학습한 후 이전에 본 적이 없는 주택 가격을 예측하려는 경우가 많습니다. 새로운 x 값(거실 면적 및 침실 수)이 주어지면 먼저 이전에 학습 세트에서 계산한 평균 및 표준 편차를 사용하여 x를 정규화해야 합니다.

**구현**

### 참조 솔루션: 한 번에 한 줄씩

아래의 각 연습은 다음 코드 셀의 한 줄에 하나의 표현식으로 해결됩니다. 번호가 매겨진 참조는 해당 표현식을 코드 순서대로 설명합니다. 학생용 버전과 비교할 때 주변 루프와 반환 문을 유지하세요.

참고 표기: `m`는 예시 개수입니다. `i`는 예시를 선택합니다. `j`는 feature를 선택합니다. 예측값은 모델 출력을 의미합니다. 기울기는 매개변수가 변경될 때 비용이 어떻게 변경되는지를 의미합니다.

1. **각 feature 평균을 계산합니다**

$\mu_j=\frac{1}{m}\sum_i X_{ij}$. `np.mean`와 `axis=0`는 열당 하나의 평균을 반환합니다.

2. **각 feature 표준편차를 계산합니다**

$\sigma_j=\sqrt{\frac{1}{m}\sum_i(X_{ij}-\mu_j)^2}$. `np.std`를 `axis=0` 및 기본 `ddof=0`와 함께 사용하세요.

3. **feature 표준화**

$X_{ij}^{scaled}=(X_{ij}-\mu_j)/\sigma_j$. `X`, `mu` 및 `sigma`를 사용하십시오. NumPy는 모든 행에 작업을 적용합니다.

**함수 참조:** [내적](https://numpy.org/doc/stable/reference/generated/numpy.dot.html), [열 평균 및 축](https://numpy.org/doc/stable/reference/generated/numpy.mean.html), [지수](https://numpy.org/doc/stable/reference/generated/numpy.exp.html). 아래 원본 실습에는 전체 수학적 맥락이 포함되어 있습니다.

출처: [C1_W2_Lab03_Feature_Scaling_and_Learning_Rate_Soln.ipynb](https://github.com/greyhatguy007/Machine-Learning-Specialization-Coursera/blob/d5c95ffb5b1524367e1a9147707ec4725298d715/C1%20-%20Supervised%20Machine%20Learning%20-%20Regression%20and%20Classification/week2/Optional%20Labs/C1_W2_Lab03_Feature_Scaling_and_Learning_Rate_Soln.ipynb)

````python
def zscore_normalize_features(X):
    """
    computes  X, zcore normalized by column

    Args:
      X (ndarray (m,n))     : input data, m examples, n features

    Returns:
      X_norm (ndarray (m,n)): input normalized by column
      mu (ndarray (n,))     : mean of each feature
      sigma (ndarray (n,))  : standard deviation of each feature
    """
    # find the mean of each column/feature
    mu     = X.mean(axis=0)                 # mu will have shape (n,)
    # find the standard deviation of each column/feature
    sigma  = X.std(axis=0)                  # sigma will have shape (n,)
    # element-wise, subtract mu for that column from each example, divide by std for that column
    X_norm = np.divide(X - mu, sigma)

    return (X_norm, mu, sigma)

#check our work
#from sklearn.preprocessing import scale
#scale(X_orig, axis=0, with_mean=True, with_std=True, copy=True)
````

````python
# Verify the reference implementation
test_norm, test_mu, test_sigma = zscore_normalize_features(np.array([[1., 2.], [3., 6.]]))
np.testing.assert_allclose(test_mu, [2., 4.])
np.testing.assert_allclose(test_sigma, [1., 2.])
np.testing.assert_allclose(test_norm, [[-1., -1.], [1., 1.]])
print("Check passed.")
````

Z-score 정규화와 관련된 단계를 살펴보겠습니다. 아래 플롯은 변환을 단계별로 보여줍니다.

````python
X_norm, mu, sigma = zscore_normalize_features(X_train)
X_mean = X_train - mu

fig,ax=plt.subplots(1, 3, figsize=(12, 3))
ax[0].scatter(X_train[:,0], X_train[:,3])
ax[0].set_xlabel(X_features[0]); ax[0].set_ylabel(X_features[3]);
ax[0].set_title("unnormalized")
ax[0].axis('equal')

ax[1].scatter(X_mean[:,0], X_mean[:,3])
ax[1].set_xlabel(X_features[0]); ax[0].set_ylabel(X_features[3]);
ax[1].set_title(r"X - $\mu$")
ax[1].axis('equal')

ax[2].scatter(X_norm[:,0], X_norm[:,3])
ax[2].set_xlabel(X_features[0]); ax[0].set_ylabel(X_features[3]);
ax[2].set_title(r"Z-score normalized")
ax[2].axis('equal')
plt.tight_layout(rect=[0, 0.03, 1, 0.95])
fig.suptitle("distribution of features before, during, after normalization")
plt.show()
````

위의 플롯은 학습 데이터의 두 feature인 "age"와 "size(sqft)" 사이의 관계를 보여줍니다. *동일 척도로 표시됩니다*.
- 왼쪽: 비정규화: '크기(sqft)' feature의 값 범위 또는 분산이 연령에 비해 훨씬 큼
- 중간: 첫 번째 단계에서는 각 feature에서 평균 또는 평균값을 제거합니다. 그러면 feature가 0을 중심으로 남게 됩니다. '연령' feature의 차이를 확인하기는 어렵지만 '크기(sqft)'는 확실히 0에 가깝습니다.
- 오른쪽: 두 번째 단계에서는 표준편차로 나눕니다. 이로 인해 feature는 모두 비슷한 눈금으로 0의 중심에 있게 됩니다.

데이터를 정규화하여 원본 데이터와 비교해 보겠습니다.

````python
# normalize the original features
X_norm, X_mu, X_sigma = zscore_normalize_features(X_train)
print(f"X_mu = {X_mu}, \nX_sigma = {X_sigma}")
print(f"Peak to Peak range by column in Raw        X:{np.ptp(X_train,axis=0)}")
print(f"Peak to Peak range by column in Normalized X:{np.ptp(X_norm,axis=0)}")
````

각 열의 최댓값과 최솟값 차이를 비교해보시오. 정규화 전에는 feature 사이에 수천 배 차이가 났지만, 정규화 후에는 그 차이가 2~3배 정도로 줄었습니다.

````python
fig,ax=plt.subplots(1, 4, figsize=(12, 3))
for i in range(len(ax)):
    norm_plot(ax[i],X_train[:,i],)
    ax[i].set_xlabel(X_features[i])
ax[0].set_ylabel("count");
fig.suptitle("distribution of features before normalization")
plt.show()
fig,ax=plt.subplots(1,4,figsize=(12,3))
for i in range(len(ax)):
    norm_plot(ax[i],X_norm[:,i],)
    ax[i].set_xlabel(X_features[i])
ax[0].set_ylabel("count");
fig.suptitle("distribution of features after normalization")

plt.show()
````

위에서 정규화된 데이터(x축)의 범위는 0을 중심으로 하며 대략 +/- 2입니다. 가장 중요한 것은 범위가 각 feature에 대해 비슷하다는 것입니다.

정규화된 데이터를 사용하여 gradient descent 알고리즘을 다시 실행해 보겠습니다. **알파 값이 훨씬 더 큽니다**. 이렇게 하면 gradient descent 속도가 빨라집니다.

````python
w_norm, b_norm, hist = run_gradient_descent(X_norm, y_train, 1000, 1.0e-1, )
````

스케일을 맞춘 feature는 매우 정확한 결과를 **훨씬, 훨씬 빠르게!** 얻습니다. 이 상당히 짧은 실행이 끝나면 각 매개변수의 기울기가 매우 작아집니다. 0.1의 learning rate는 정규화된 feature를 사용한 회귀 분석의 좋은 시작입니다. 예측값 대 target 값을 플롯해 보겠습니다. 예측값은 정규화된 feature를 사용하여 만들어지고 플롯은 원래 feature 값을 사용하여 표시됩니다.

````python
#predict target using normalized features
m = X_norm.shape[0]
yp = np.zeros(m)
for i in range(m):
    yp[i] = np.dot(X_norm[i], w_norm) + b_norm

    # plot predictions and targets versus original features
fig,ax=plt.subplots(1,4,figsize=(12, 3),sharey=True)
for i in range(len(ax)):
    ax[i].scatter(X_train[:,i],y_train, label = 'target')
    ax[i].set_xlabel(X_features[i])
    ax[i].scatter(X_train[:,i],yp,color=dlc["dlorange"], label = 'predict')
ax[0].set_ylabel("Price"); ax[0].legend();
fig.suptitle("target versus prediction using z-score normalized model")
plt.show()
````

결과는 좋아 보입니다. 참고할 몇 가지 사항:
- 여러 개의 feature를 사용하면 더 이상 feature에 대한 결과를 보여주는 단일 플롯을 가질 수 없습니다.
- 예측에는 정규화된 입력을 사용했습니다. 새 입력도 학습 세트에서 구한 평균과 표준편차로 정규화해야 합니다. 이 과정에서 예측 대상인 집값 자체를 정규화하는 것은 아닙니다.

**예측** 우리 모델을 생성하는 목적은 이를 사용하여 데이터 세트에 없는 주택 가격을 예측하는 것입니다. 40년 된 1200제곱피트, 침실 3개, 1층 주택의 가격을 예측해 보겠습니다. 학습 데이터를 정규화할 때 파생된 평균 및 표준 편차를 사용하여 데이터를 정규화해야 한다는 점을 기억하세요.

### 참조 솔루션: 한 번에 한 줄씩

아래의 각 연습은 다음 코드 셀의 한 줄에 하나의 표현식으로 해결됩니다. 번호가 매겨진 참조는 해당 표현식을 코드 순서대로 설명합니다. 학생용 버전과 비교할 때 주변 루프와 반환 문을 유지하세요.

참고 표기: `m`는 예시 개수입니다. `i`는 예시를 선택합니다. `j`는 feature를 선택합니다. 예측값은 모델 출력을 의미합니다. 기울기는 매개변수가 변경될 때 비용이 어떻게 변경되는지를 의미합니다.

1. **새 집 feature 표준화**

$x_j^{scaled}=(x_j-\mu_j)/\sigma_j$. `x_house` 및 훈련 통계 `X_mu`, `X_sigma`를 사용합니다.

**함수 참조:** [내적](https://numpy.org/doc/stable/reference/generated/numpy.dot.html), [열 평균 및 축](https://numpy.org/doc/stable/reference/generated/numpy.mean.html), [지수](https://numpy.org/doc/stable/reference/generated/numpy.exp.html). 아래 원본 실습에는 전체 수학적 맥락이 포함되어 있습니다.

# Feature Engineering 및 다항식 회귀 개요

기본적으로 Linear Regression은 다음 형식의 모델을 구축하는 수단을 제공합니다.
$$f_{\mathbf{w},b} = w_0x_0 + w_1x_1+ ... + w_{n-1}x_{n-1} + b \tag{1}$$
feature/데이터가 비선형이거나 feature의 조합인 경우에는 어떻게 됩니까? 예를 들어, 주택 가격은 생활 면적에 선형적인 경향이 없지만 매우 작거나 큰 주택에 불이익을 가하여 위 그래픽에 표시된 곡선을 만듭니다. 이 곡선에 맞게 linear regression 기계를 어떻게 사용할 수 있습니까? 우리가 가지고 있는 '기계'는 방정식을 학습 데이터에 '맞추기' 위해 (1)의 매개변수 $\mathbf{w}$, $\mathbf{b}$를 수정하는 기능입니다. 그러나 (1)의 $\mathbf{w}$, $\mathbf{b}$를 아무리 조정해도 비선형 곡선에 적합하지 않습니다.

<a name='PolynomialFeatures'></a>
## 다항식 Features

위에서 우리는 데이터가 비선형인 시나리오를 고려했습니다. 비선형 곡선을 맞추기 위해 지금까지 우리가 알고 있는 것을 사용해 보겠습니다. 간단한 이차식부터 시작하겠습니다: $y = 1+x^2$

당신은 우리가 사용하는 모든 루틴에 익숙합니다. 검토를 위해 lab_utils.py 파일에서 사용할 수 있습니다. 열 경계를 따라 연결하기 위해 NumPy 루틴인 [`np.c_[..]`](https://numpy.org/doc/stable/reference/generated/numpy.c_.html)를 사용합니다.

````python
# create target data
x = np.arange(0, 20, 1)
y = 1 + x**2
X = x.reshape(-1, 1)

model_w,model_b = run_gradient_descent_feng(X,y,iterations=1000, alpha = 1e-2)

plt.scatter(x, y, marker='x', c='r', label="Actual Value"); plt.title("no feature engineering")
plt.plot(x,X@model_w + model_b, label="Predicted Value");  plt.xlabel("X"); plt.ylabel("y"); plt.legend(); plt.show()
````

뭐, 예상대로 별로 맞지는 않네요. 필요한 것은 $y= w_0x_0^2 + b$ 또는 **다항식 feature**와 같은 것입니다. 이를 달성하려면 *입력 데이터*를 수정하여 필요한 feature를 *엔지니어링*할 수 있습니다. 원본 데이터를 $x$ 값을 제곱하는 버전으로 바꾸면 $y= w_0x_0^2 + b$를 얻을 수 있습니다. 한번 해보자. 아래에서 `X`를 `X**2`로 교체하세요.

### 참조 솔루션: 한 번에 한 줄씩

아래의 각 연습은 다음 코드 셀의 한 줄에 하나의 표현식으로 해결됩니다. 번호가 매겨진 참조는 해당 표현식을 코드 순서대로 설명합니다. 학생용 버전과 비교할 때 주변 루프와 반환 문을 유지하세요.

참고 표기: `m`는 예시 개수입니다. `i`는 예시를 선택합니다. `j`는 feature를 선택합니다. 예측값은 모델 출력을 의미합니다. 기울기는 매개변수가 변경될 때 비용이 어떻게 변경되는지를 의미합니다.

1. **제곱된 feature를 생성합니다**

$x^2$. `**`를 사용하여 `x`의 모든 요소를 ​​강화하세요.

**함수 참조:** [내적](https://numpy.org/doc/stable/reference/generated/numpy.dot.html), [열 평균 및 축](https://numpy.org/doc/stable/reference/generated/numpy.mean.html), [지수](https://numpy.org/doc/stable/reference/generated/numpy.exp.html). 아래 원본 실습에는 전체 수학적 맥락이 포함되어 있습니다.

출처: [C1_W2_Lab04_FeatEng_PolyReg_Soln.ipynb](https://github.com/greyhatguy007/Machine-Learning-Specialization-Coursera/blob/d5c95ffb5b1524367e1a9147707ec4725298d715/C1%20-%20Supervised%20Machine%20Learning%20-%20Regression%20and%20Classification/week2/Optional%20Labs/C1_W2_Lab04_FeatEng_PolyReg_Soln.ipynb)

````python
# create target data
x = np.arange(0, 20, 1)
y = 1 + x**2

# Engineer features
X = np.square(x)      #<-- added engineered feature
````

````python
X = X.reshape(-1, 1)  #X should be a 2-D Matrix
model_w,model_b = run_gradient_descent_feng(X, y, iterations=10000, alpha = 1e-5)

plt.scatter(x, y, marker='x', c='r', label="Actual Value"); plt.title("Added x**2 feature")
plt.plot(x, np.dot(X,model_w) + model_b, label="Predicted Value"); plt.xlabel("x"); plt.ylabel("y"); plt.legend(); plt.show()
````

데이터에 훨씬 잘 맞습니다 거의 완벽에 가깝습니다. 그래프 바로 위에 출력된 $\mathbf{w}$ 및 b 값: `w,b found by gradient descent: w: [1.], b: 0.0490`를 확인하세요. Gradient descent는 $\mathbf{w},b $의 초기 값을 (1.0,0.049) 또는 $y=1*x_0^2+0.049$의 모델로 수정하여 $y=1*x_0^2+1$의 target에 매우 가깝습니다. 반복 횟수를 늘리면 데이터에 더 잘 맞출 수 있습니다.

### Features 선택
<a name='GDF'></a>
위에서 우리는 $x^2$ 용어가 필요하다는 것을 알고 있었습니다. 어떤 feature가 필요한지 항상 명확하지 않을 수도 있습니다. 다양한 잠재적인 feature를 추가하여 가장 유용한 것을 찾을 수 있습니다. 예를 들어 $y=w_0x_0 + w_1x_1^2 + w_2x_2^3+b$를 대신 시도했다면 어떨까요?

다음 셀을 실행합니다.

### 참조 솔루션: 한 번에 한 줄씩

아래의 각 연습은 다음 코드 셀의 한 줄에 하나의 표현식으로 해결됩니다. 번호가 매겨진 참조는 해당 표현식을 코드 순서대로 설명합니다. 학생용 버전과 비교할 때 주변 루프와 반환 문을 유지하세요.

참고 표기: `m`는 예시 개수입니다. `i`는 예시를 선택합니다. `j`는 feature를 선택합니다. 예측값은 모델 출력을 의미합니다. 기울기는 매개변수가 변경될 때 비용이 어떻게 변경되는지를 의미합니다.

1. **3승까지 feature를 생성합니다**

$[x,x^2,x^3]$ 열을 해당 순서대로 생성합니다. 출력에는 `len(x)` 행과 3개의 열이 있습니다.

**Python 패턴, 다른 예:** `np.column_stack([t, t**2])`는 두 개의 feature 열을 만듭니다. 13개의 거듭제곱을 직접 작성하는 것을 피하기 위해 한 줄 패턴 `np.column_stack([t**k for k in range(2, 6)])`는 2에서 5까지의 거듭제곱에 대한 열을 만듭니다. 입력 및 지수 범위를 요청된 열에 적용합니다. `range`의 상한은 제외됩니다.

[NumPy 컬럼 스택 참조](https://numpy.org/doc/stable/reference/generated/numpy.column_stack.html)

**함수 참조:** [내적](https://numpy.org/doc/stable/reference/generated/numpy.dot.html), [열 평균 및 축](https://numpy.org/doc/stable/reference/generated/numpy.mean.html), [지수](https://numpy.org/doc/stable/reference/generated/numpy.exp.html). 아래 원본 실습에는 전체 수학적 맥락이 포함되어 있습니다.

출처: [C1_W2_Lab04_FeatEng_PolyReg_Soln.ipynb](https://github.com/greyhatguy007/Machine-Learning-Specialization-Coursera/blob/d5c95ffb5b1524367e1a9147707ec4725298d715/C1%20-%20Supervised%20Machine%20Learning%20-%20Regression%20and%20Classification/week2/Optional%20Labs/C1_W2_Lab04_FeatEng_PolyReg_Soln.ipynb)

````python
# create target data
x = np.arange(0, 20, 1)
y = x**2

# engineer features .
X = np.column_stack([x ** power for power in range(1, 4)])   #<-- added engineered feature
````

````python
model_w,model_b = run_gradient_descent_feng(X, y, iterations=10000, alpha=1e-7)

plt.scatter(x, y, marker='x', c='r', label="Actual Value"); plt.title("x, x**2, x**3 features")
plt.plot(x, X@model_w + model_b, label="Predicted Value"); plt.xlabel("x"); plt.ylabel("y"); plt.legend(); plt.show()
````

$\mathbf{w}$, `[0.08 0.54 0.03]` 및 b의 값은 `0.0106`입니다. 이는 피팅/훈련 후 모델이 다음과 같음을 의미합니다.
$$ 0.08x + 0.54x^2 + 0.03x^3 + 0.0106 $$
Gradient descent는 $w_1$ 항을 다른 항에 비해 늘려 $x^2$ 데이터에 가장 적합한 데이터를 강조했습니다.  아주 오랜 시간 동안 실행한다면 다른 용어의 영향은 계속해서 줄어들 것입니다.
>Gradient descent는 ​​관련 매개변수를 강조하여 '올바른' feature를 선택합니다.

이 아이디어를 검토해 보겠습니다.
- 처음에 feature는 서로 비교할 수 있도록 크기가 조정되었습니다.
- 가중치 값이 적다는 것은 feature가 덜 중요하거나 정확하다는 것을 의미하며, 극단적으로 가중치가 0이 되거나 0에 매우 가까워지면 해당 feature는 현재 모델의 예측에 거의 기여하지 않습니다.
- 위의 피팅 후 $x^2$ feature와 관련된 가중치는 데이터 피팅에 가장 유용하므로 $x$ 또는 $x^3$의 가중치보다 훨씬 큽니다.

### 대체 보기
위에서는 target 데이터와 얼마나 잘 일치하는지에 따라 다항식 feature가 선택되었습니다. 이에 대해 생각하는 또 다른 방법은 새로운 feature를 생성한 후에도 여전히 Linear Regression을 사용하고 있다는 점에 유의하는 것입니다. 이를 감안할 때 최고의 feature는 target에 비해 선형적입니다. 이는 예를 통해 가장 잘 이해됩니다.

### 참조 솔루션: 한 번에 한 줄씩

아래의 각 연습은 다음 코드 셀의 한 줄에 하나의 표현식으로 해결됩니다. 번호가 매겨진 참조는 해당 표현식을 코드 순서대로 설명합니다. 학생용 버전과 비교할 때 주변 루프와 반환 문을 유지하세요.

참고 표기: `m`는 예시 개수입니다. `i`는 예시를 선택합니다. `j`는 feature를 선택합니다. 예측값은 모델 출력을 의미합니다. 기울기는 매개변수가 변경될 때 비용이 어떻게 변경되는지를 의미합니다.

1. **3승까지 feature를 생성합니다**

$[x,x^2,x^3]$ 열을 해당 순서대로 생성합니다. 출력에는 `len(x)` 행과 3개의 열이 있습니다.

**Python 패턴, 다른 예:** `np.column_stack([t, t**2])`는 두 개의 feature 열을 만듭니다. 13개의 거듭제곱을 직접 작성하는 것을 피하기 위해 한 줄 패턴 `np.column_stack([t**k for k in range(2, 6)])`는 2에서 5까지의 거듭제곱에 대한 열을 만듭니다. 입력 및 지수 범위를 요청된 열에 적용합니다. `range`의 상한은 제외됩니다.

[NumPy 컬럼 스택 참조](https://numpy.org/doc/stable/reference/generated/numpy.column_stack.html)

**함수 참조:** [내적](https://numpy.org/doc/stable/reference/generated/numpy.dot.html), [열 평균 및 축](https://numpy.org/doc/stable/reference/generated/numpy.mean.html), [지수](https://numpy.org/doc/stable/reference/generated/numpy.exp.html). 아래 원본 실습에는 전체 수학적 맥락이 포함되어 있습니다.

````python
# create target data
x = np.arange(0, 20, 1)
y = x**2

# engineer features .
X = np.column_stack([x ** power for power in range(1, 4)])   #<-- added engineered feature
X_features = ['x','x^2','x^3']
````

````python
fig,ax=plt.subplots(1, 3, figsize=(12, 3), sharey=True)
for i in range(len(ax)):
    ax[i].scatter(X[:,i],y)
    ax[i].set_xlabel(X_features[i])
ax[0].set_ylabel("y")
plt.show()
````

위에서 target 값 $y$에 대해 매핑된 $x^2$ feature는 선형이라는 것이 분명합니다. 그러면 Linear Regression은 해당 feature를 사용하여 모델을 쉽게 생성할 수 있습니다.

# Optional Lab: scikit-learn을 사용하는 Linear Regression

[scikit-learn](https://scikit-learn.org/stable/index.html)라는 오픈 소스, 상업적으로 사용 가능한 기계 학습 도구 키트가 있습니다. 이 툴킷에는 이 과정에서 작업하게 될 다양한 알고리즘의 구현이 포함되어 있습니다.



## 목표
이 실습에서는 다음을 수행합니다.
- scikit-learn을 활용하여 Gradient Descent를 사용하여 Linear Regression을 구현합니다.

## 도구
scikit-learn뿐만 아니라 matplotlib 및 NumPy의 기능도 활용하게 됩니다.

````python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import SGDRegressor
from sklearn.preprocessing import StandardScaler
from lab_utils_multi import  load_house_data
from lab_utils_common import dlc
np.set_printoptions(precision=2)
plt.style.use('./deeplearning.mplstyle')
````

# Gradient Descent
Scikit-learn에는 gradient descent 회귀 모델 [sklearn.linear_model.SGDRegressor](https://scikit-learn.org/stable/modules/generated/sklearn.linear_model.SGDRegressor.html#examples-using-sklearn-linear-model-sgdregressor)이 있습니다.  gradient descent의 이전 구현과 마찬가지로 이 모델은 정규화된 입력에서 가장 잘 작동합니다. [sklearn.preprocessing.StandardScaler](https://scikit-learn.org/stable/modules/generated/sklearn.preprocessing.StandardScaler.html#sklearn.preprocessing.StandardScaler)는 이전 실습에서처럼 z-score 정규화를 수행합니다. 여기서는 '표준점수'라고 합니다.

### 데이터 세트 로드

````python
X_train, y_train = load_house_data()
X_features = ['size(sqft)','bedrooms','floors','age']
````

### 학습 데이터 확장/정규화

````python
scaler = StandardScaler()
X_norm = scaler.fit_transform(X_train)
print(f"Peak to Peak range by column in Raw        X:{np.ptp(X_train,axis=0)}")
print(f"Peak to Peak range by column in Normalized X:{np.ptp(X_norm,axis=0)}")
````

### 회귀 모델 생성 및 적합

````python
sgdr = SGDRegressor(max_iter=1000)
sgdr.fit(X_norm, y_train)
print(sgdr)
print(f"number of iterations completed: {sgdr.n_iter_}, number of weight updates: {sgdr.t_}")
````

### 매개변수 보기
매개변수는 *정규화된* 입력 데이터와 연관되어 있습니다. 적합 매개변수는 이 데이터를 사용하여 이전 실습에서 찾은 매개변수와 매우 유사합니다.

````python
b_norm = sgdr.intercept_
w_norm = sgdr.coef_
print(f"model parameters:                   w: {w_norm}, b:{b_norm}")
print( "model parameters from previous lab: w: [110.56 -21.27 -32.71 -37.97], b: 363.16")
````

### 예측값 만들기
학습 데이터의 목표를 예측합니다. `predict` 루틴을 모두 사용하고 $w$ 및 $b$를 사용하여 계산합니다.

````python
# make a prediction using sgdr.predict()
y_pred_sgd = sgdr.predict(X_norm)
# make a prediction using w,b.
y_pred = np.dot(X_norm, w_norm) + b_norm
print(f"prediction using np.dot() and sgdr.predict match: {(y_pred == y_pred_sgd).all()}")

print(f"Prediction on training set:\n{y_pred[:4]}" )
print(f"Target values \n{y_train[:4]}")
````

### 플롯 결과
예측값 대 target 값을 플롯해 보겠습니다.

````python
# plot predictions and targets vs original features
fig,ax=plt.subplots(1,4,figsize=(12,3),sharey=True)
for i in range(len(ax)):
    ax[i].scatter(X_train[:,i],y_train, label = 'target')
    ax[i].set_xlabel(X_features[i])
    ax[i].scatter(X_train[:,i],y_pred,color=dlc["dlorange"], label = 'predict')
ax[0].set_ylabel("Price"); ax[0].legend();
fig.suptitle("target versus prediction using z-score normalized model")
plt.show()
````

- 오픈 소스 기계 학습 툴킷인 scikit-learn을 활용했습니다.

# Linear Regression, Closed-form Solution
Scikit-learn에는 폐쇄형 Linear Regression을 구현하는 [linear regression 모델](https://scikit-learn.org/stable/modules/generated/sklearn.linear_model.LinearRegression.html#sklearn.linear_model.LinearRegression)이 있습니다.

앞에서 사용한 두 집으로 시작합니다. 면적이 1000제곱피트인 집은 300,000달러, 2000제곱피트인 집은 500,000달러에 팔렸습니다.

| 규모(1000평방피트) | 가격(천달러) |
| ---| ----------- |
| 1 | 300 |
| 2 | 500 |

### 데이터 세트 로드

````python
X_train = np.array([1.0, 2.0])   #features
y_train = np.array([300, 500])   #target value
````

### 모델 생성 및 피팅

>Closed-form Solution에는 정규화가 필요하지 않습니다.

## Sigmoid 또는 Logistic 함수

sigmoid 함수를 구현하고 이를 직접 살펴보겠습니다.

## Sigmoid 함수의 공식

sigmoid 함수의 공식은 다음과 같습니다.

$g(z) = \frac{1}{1+e^{-z}}\tag{1}$

logistic regression의 경우 z(Sigmoid 함수에 대한 입력)는 linear regression 모델의 출력입니다.
- 단일 예의 경우 $z$는 스칼라입니다.
- 예제가 여러 개인 경우 $z$는 각 예제마다 하나씩 $m$ 값으로 구성된 벡터일 수 있습니다.
- Sigmoid 함수의 구현은 이러한 잠재적인 입력 형식을 모두 포함해야 합니다. 이것을 Python으로 구현해보자.

NumPy에는 [`exp()`](https://numpy.org/doc/stable/reference/generated/numpy.exp.html)라는 함수가 있는데, 이 함수는 입력 배열(`z`)에 있는 모든 요소의 지수($e^{z}$)를 계산하는 편리한 방법을 제공합니다.

또한 아래와 같이 단일 숫자를 입력으로 사용하여 작동합니다.

````python
# Input is an array.
input_array = np.array([1,2,3])
exp_array = np.exp(input_array)

print("Input to exp:", input_array)
print("Output of exp:", exp_array)

# Input is a single number
input_val = 1
exp_val = np.exp(input_val)

print("Input to exp:", input_val)
print("Output of exp:", exp_val)
````

`sigmoid` 함수는 아래 셀과 같이 Python으로 구현됩니다.

### 참조 솔루션: 한 번에 한 줄씩

아래의 각 연습은 다음 코드 셀의 한 줄에 하나의 표현식으로 해결됩니다. 번호가 매겨진 참조는 해당 표현식을 코드 순서대로 설명합니다. 학생용 버전과 비교할 때 주변 루프와 반환 문을 유지하세요.

참고 표기: `m`는 예시 개수입니다. `i`는 예시를 선택합니다. `j`는 feature를 선택합니다. 예측값은 모델 출력을 의미합니다. 기울기는 매개변수가 변경될 때 비용이 어떻게 변경되는지를 의미합니다.

1. **sigmoid 확률을 계산합니다**

$\sigma(z)=1/(1+e^{-z})$. `np.exp`는 스칼라 또는 배열의 지수를 평가합니다.

**함수 참조:** [내적](https://numpy.org/doc/stable/reference/generated/numpy.dot.html), [열 평균 및 축](https://numpy.org/doc/stable/reference/generated/numpy.mean.html), [지수](https://numpy.org/doc/stable/reference/generated/numpy.exp.html). 아래 원본 실습에는 전체 수학적 맥락이 포함되어 있습니다.

출처: [C1_W3_Lab02_Sigmoid_function_Soln.ipynb](https://github.com/greyhatguy007/Machine-Learning-Specialization-Coursera/blob/d5c95ffb5b1524367e1a9147707ec4725298d715/C1%20-%20Supervised%20Machine%20Learning%20-%20Regression%20and%20Classification/week3/Optional%20Labs/C1_W3_Lab02_Sigmoid_function_Soln.ipynb)

````python
def sigmoid(z):
    """
    Compute the sigmoid of z

    Args:
        z (ndarray): A scalar, numpy array of any size.

    Returns:
        g (ndarray): sigmoid(z), with the same shape as z

    """

    g = np.reciprocal(1.0 + np.exp(-z))

    return g
````

````python
# Verify the reference implementation
np.testing.assert_allclose(sigmoid(np.array([-2., 0., 2.])), [0.119202922, 0.5, 0.880797078], rtol=1e-7)
print("Check passed.")
````

`z`의 다양한 값에 대해 이 함수의 출력이 무엇인지 살펴보겠습니다.

````python
# Generate an array of evenly spaced values between -10 and 10
z_tmp = np.arange(-10,11)

# Use the function implemented above to get the sigmoid values
y = sigmoid(z_tmp)

# Code for pretty printing the two arrays next to each other
np.set_printoptions(precision=3)
print("Input (z), Output (sigmoid(z))")
print(np.c_[z_tmp, y])
````

왼쪽 열의 값은 `z`이고 오른쪽 열의 값은 `sigmoid(z)`입니다. 보시다시피 sigmoid에 대한 입력 값은 -10~10이고, 출력 값은 0~1입니다.

이제 `matplotlib` 라이브러리를 사용하여 이 함수를 플롯해 보겠습니다.

````python
# Plot z vs sigmoid(z)
fig,ax = plt.subplots(1,1,figsize=(5,3))
ax.plot(z_tmp, y, c="b")

ax.set_title("Sigmoid function")
ax.set_ylabel('sigmoid(z)')
ax.set_xlabel('z')
draw_vthresh(ax,0)
````

보시다시피, sigmoid 함수는 `z`가 큰 음수 값으로 갈 때 `0`에 접근하고 `z`가 큰 양수 값으로 갈 때 `1`에 접근합니다.

## Logistic Regression
<img align="left" src="https://raw.githubusercontent.com/greyhatguy007/Machine-Learning-Specialization-Coursera/d5c95ffb5b1524367e1a9147707ec4725298d715/C1%20-%20Supervised%20Machine%20Learning%20-%20Regression%20and%20Classification/week3/Optional%20Labs/images/C1_W3_LogisticRegression_right.png"     style=" width:300px; padding: 10px; " > logistic regression 모델은 아래와 같이 익숙한 linear regression 모델에 sigmoid를 적용합니다.

$$ f_{\mathbf{w},b}(\mathbf{x}^{(i)}) = g(\mathbf{w} \cdot \mathbf{x}^{(i)} + b ) \tag{2} $$

어디

  $g(z) = \frac{1}{1+e^{-z}}\tag{3}$


종양 분류의 범주형 데이터 예시에 Logistic Regression을 적용해 보겠습니다. 먼저 매개변수의 예제와 초기값을 로드합니다.
  


````python
x_train = np.array([0., 1, 2, 3, 4, 5])
y_train = np.array([0,  0, 0, 1, 1, 1])

w_in = np.zeros((1))
b_in = 0
````

다음 단계를 시도해 보세요.
- 주어진 학습 데이터에 가장 적합한 logistic regression 모델을 찾으려면 'Logistic Regression 실행'을 클릭하십시오.
    - 결과 모델이 데이터에 매우 잘 맞는다는 점에 유의하세요.
    - 참고로 주황색 선은 위의 '$z$' 또는 $\mathbf{w} \cdot \mathbf{x}^{(i)} + b$입니다. linear regression 모델의 라인과 일치하지 않습니다. *임계값*을 적용하여 이러한 결과를 더욱 개선하세요.
- 임계값이 적용되면 예측값을 표시하려면 '토글 0.5 임계값' 상자를 선택합니다.
    - 이 예측값은 좋아 보입니다. 예측값은 데이터와 일치합니다.
    - 이제 큰 종양 크기 범위(거의 10)에 추가 데이터 포인트를 추가하고 Logistic Regression을 다시 실행합니다.
    - linear regression 모델과 달리 이 모델은 계속해서 올바른 예측값을 만듭니다.

````python
plt.close('all')
addpt = plt_one_addpt_onclick( x_train,y_train, w_in, b_in, logistic=True)
````

# Optional Lab: Logistic Regression, 결정 경계

## 목표
이 실습에서는 다음을 수행합니다.
- logistic regression 모델의 결정 경계를 플롯합니다. 이를 통해 모델이 예측하는 내용을 더 잘 이해할 수 있습니다.

````python
import numpy as np

import matplotlib.pyplot as plt
from lab_utils_common import plot_data, sigmoid, draw_vthresh
plt.style.use('./deeplearning.mplstyle')
````

## 데이터세트

다음과 같은 학습 데이터 세트가 있다고 가정해 보겠습니다.
- 입력 변수 `X`는 각각 2개의 feature를 포함하는 6개의 학습 예제가 있는 numpy 배열입니다.
- 출력 변수 `y`도 6개의 예제가 있는 numpy 배열이고, `y`는 `0` 또는 `1`입니다.

````python
X = np.array([[0.5, 1.5], [1,1], [1.5, 0.5], [3, 0.5], [2, 2], [1, 2.5]])
y = np.array([0, 0, 0, 1, 1, 1]).reshape(-1,1)
````

### 플롯 데이터

도우미 함수를 사용하여 이 데이터를 그려보겠습니다. $y=1$ 라벨이 있는 데이터 포인트는 빨간색 십자가로 표시되고, $y=0$ 라벨이 있는 데이터 포인트는 파란색 원으로 표시됩니다.

````python
fig,ax = plt.subplots(1,1,figsize=(4,4))
plot_data(X, y, ax)

ax.axis([0, 4, 0, 3.5])
ax.set_ylabel('$x_1$')
ax.set_xlabel('$x_0$')
plt.show()
````

## Logistic regression 모델


* 다음 형식의 이 데이터에 대해 logistic regression 모델을 훈련한다고 가정해 보겠습니다.

  $f(x) = g(w_0x_0+w_1x_1 + b)$
  
여기서 $g(z) = \frac{1}{1+e^{-z}}$는 sigmoid 함수입니다.


* 모델을 훈련하고 매개변수를 $b = -3, w_0 = 1, w_1 = 1$로 얻었다고 가정해 보겠습니다. 즉,

  $f(x) = g(x_0+x_1-3)$

(과정에서 이러한 매개변수를 데이터에 맞추는 방법을 배우게 됩니다.)
  
  
결정 경계를 그려 훈련된 모델이 무엇을 예측하는지 이해해 보겠습니다.

### logistic regression 및 결정 경계에 대한 복습

* logistic regression의 경우 모델은 다음과 같이 표시됩니다.

  $$f_{\mathbf{w},b}(\mathbf{x}^{(i)}) = g(\mathbf{w} \cdot \mathbf{x}^{(i)} + b) \tag{1}$$

여기서 $g(z)$는 sigmoid 함수로 알려져 있으며 모든 입력 값을 0과 1 사이의 값으로 매핑합니다.

$g(z) = \frac{1}{1+e^{-z}}\tag{2}$
  $\mathbf{w} \cdot \mathbf{x}$는 벡터 내적입니다.
  
  $$\mathbf{w} \cdot \mathbf{x} = w_0 x_0 + w_1 x_1$$
  
  
* 모델($f_{\mathbf{w},b}(x)$)의 출력은 $y=1$가 $\mathbf{x}$에 주어지고 $\mathbf{w}$ 및 $b$에 의해 매개변수화될 확률로 해석됩니다.
* 따라서 logistic regression 모델에서 최종 예측값($y=0$ 또는 $y=1$)를 얻으려면 다음 휴리스틱을 사용할 수 있습니다.

$f_{\mathbf{w},b}(x) >= 0.5$이면 $y=1$를 예측하세요.
  
$f_{\mathbf{w},b}(x) < 0.5$이면 $y=0$를 예측하세요.
  
  
* $g(z) >= 0.5$가 어디에 있는지 알아보기 위해 sigmoid 함수를 플로팅해 보겠습니다.

````python
# Plot sigmoid(z) over a range of values from -10 to 10
z = np.arange(-10,11)

fig,ax = plt.subplots(1,1,figsize=(5,3))
# Plot z vs sigmoid(z)
ax.plot(z, sigmoid(z), c="b")

ax.set_title("Sigmoid function")
ax.set_ylabel('sigmoid(z)')
ax.set_xlabel('z')
draw_vthresh(ax,0)
````

* 보시다시피 $z >=0$의 경우 $g(z) >= 0.5$입니다.

* logistic regression 모델의 경우 $z = \mathbf{w} \cdot \mathbf{x} + b$입니다. 그러므로,

$\mathbf{w} \cdot \mathbf{x} + b >= 0$인 경우 모델은 $y=1$를 예측합니다.
  
$\mathbf{w} \cdot \mathbf{x} + b < 0$인 경우 모델은 $y=0$를 예측합니다.
  
  
  
### 결정 경계 그리기

이제 logistic regression 모델이 예측값을 만드는 방법을 이해하기 위해 예제로 돌아가 보겠습니다.

* 당사의 logistic regression 모델은 다음과 같은 형식을 갖습니다.

  $f(\mathbf{x}) = g(-3 + x_0+x_1)$


* 위에서 배운 내용을 통해 이 모델이 $-3 + x_0+x_1 >= 0$인 경우 $y=1$를 예측한다는 것을 알 수 있습니다.

이것이 그래픽으로 어떻게 보이는지 살펴보겠습니다. $x_1 = 3 - x_0$와 동일한 $-3 + x_0+x_1 = 0$를 플로팅하는 것부터 시작하겠습니다.

### 참조 솔루션: 한 번에 한 줄씩

아래의 각 연습은 다음 코드 셀의 한 줄에 하나의 표현식으로 해결됩니다. 번호가 매겨진 참조는 해당 표현식을 코드 순서대로 설명합니다. 학생용 버전과 비교할 때 주변 루프와 반환 문을 유지하세요.

참고 표기: `m`는 예시 개수입니다. `i`는 예시를 선택합니다. `j`는 feature를 선택합니다. 예측값은 모델 출력을 의미합니다. 기울기는 매개변수가 변경될 때 비용이 어떻게 변경되는지를 의미합니다.

1. **결정 경계 높이 계산**

$x_1$에 대해 $x_0+x_1=3$를 해결합니다. `x0`에는 수평 좌표가 포함되어 있습니다.

**함수 참조:** [내적](https://numpy.org/doc/stable/reference/generated/numpy.dot.html), [열 평균 및 축](https://numpy.org/doc/stable/reference/generated/numpy.mean.html), [지수](https://numpy.org/doc/stable/reference/generated/numpy.exp.html). 아래 원본 실습에는 전체 수학적 맥락이 포함되어 있습니다.

출처: [C1_W3_Lab03_Decision_Boundary_Soln.ipynb](https://github.com/greyhatguy007/Machine-Learning-Specialization-Coursera/blob/d5c95ffb5b1524367e1a9147707ec4725298d715/C1%20-%20Supervised%20Machine%20Learning%20-%20Regression%20and%20Classification/week3/Optional%20Labs/C1_W3_Lab03_Decision_Boundary_Soln.ipynb)

````python
# Choose values between 0 and 6
x0 = np.arange(0,6)

x1 = -x0 + 3
fig,ax = plt.subplots(1,1,figsize=(5,4))
# Plot the decision boundary
ax.plot(x0,x1, c="b")
ax.axis([0, 4, 0, 3.5])

# Fill the region below the line
ax.fill_between(x0,x1, alpha=0.2)

# Plot the original data
plot_data(X,y,ax)
ax.set_ylabel(r'$x_1$')
ax.set_xlabel(r'$x_0$')
plt.show()
````

* 위 플롯에서 파란색 선은 $x_0 + x_1 - 3 = 0$ 선을 나타내며 x1 축과 3($x_1$ = 3, $x_0$ = 0으로 설정한 경우) 및 3($x_1$ = 0, $x_0$ = 3으로 설정한 경우)에서 x0 축과 교차해야 합니다.


* 음영처리된 부분은 $-3 + x_0+x_1 < 0$를 나타냅니다. 선 위의 영역은 $-3 + x_0+x_1 > 0$입니다.


* 음영처리된 영역(선 아래)의 모든 지점은 $y=0$로 분류됩니다.  선 위나 위의 모든 점은 $y=1$로 분류됩니다. 이 선을 "결정 경계"라고 합니다.

강의에서 본 것처럼 고차 다항식 항(예: $f(x) = g( x_0^2 + x_1 -1)$)을 사용하면 더 복잡한 비선형 경계를 생각해 낼 수 있습니다.

## logistic regression에 대한 제곱 오류?
<img align="left" src="https://raw.githubusercontent.com/greyhatguy007/Machine-Learning-Specialization-Coursera/d5c95ffb5b1524367e1a9147707ec4725298d715/C1%20-%20Supervised%20Machine%20Learning%20-%20Regression%20and%20Classification/week3/Optional%20Labs/images/C1_W3_SqErrorVsLogistic.png"     style=" width:400px; padding: 10px; " > **선형** 회귀 분석에서 **제곱 오류 cost function**를 사용했습니다.
하나의 변수를 사용한 제곱 오류 비용의 방정식은 다음과 같습니다.
  $$J(w,b) = \frac{1}{2m} \sum\limits_{i = 0}^{m-1} (f_{w,b}(x^{(i)}) - y^{(i)})^2 \tag{1}$$

어디서
  $$f_{w,b}(x^{(i)}) = wx^{(i)} + b \tag{2}$$

제곱 오차 비용은 비용의 미분을 따라가면 최소값이 되는 좋은 속성을 갖고 있다는 점을 기억하세요.

````python
soup_bowl()
````

이 Cost Function은 linear regression에 잘 작동했으며 logistic regression에도 고려하는 것이 당연합니다. 그러나 위의 슬라이드에서 알 수 있듯이 $f_{wb}(x)$에는 이제 비선형 구성 요소인 Sigmoid 함수인 $f_{w,b}(x^{(i)}) = sigmoid(wx^{(i)} + b )$가 있습니다.   이제 sigmoid를 포함하여 이전 실습의 예에서 오차 비용 제곱을 시도해 보겠습니다.

학습 데이터는 다음과 같습니다.

````python
x_train = np.array([0., 1, 2, 3, 4, 5],dtype=np.longdouble)
y_train = np.array([0,  0, 0, 1, 1, 1],dtype=np.longdouble)
plt_simple_example(x_train, y_train)
````

이제 *제곱 오류 비용*을 사용하여 비용의 표면 플롯을 구해 보겠습니다.
  $$J(w,b) = \frac{1}{2m} \sum\limits_{i = 0}^{m-1} (f_{w,b}(x^{(i)}) - y^{(i)})^2 $$

어디서
  $$f_{w,b}(x^{(i)}) = sigmoid(wx^{(i)} + b )$$

````python
plt.close('all')
plt_logistic_squared_error(x_train,y_train)
plt.show()
````

이것은 매우 흥미로운 플롯을 생성하지만 위의 표면은 linear regression의 '수프 그릇'만큼 매끄럽지 않습니다!

Logistic regression에는 비선형 특성에 더 적합한 Cost Function이 필요합니다. 이것은 Loss 함수으로 시작됩니다. 이에 대해서는 아래에 설명되어 있습니다.

## Logistic Loss 함수
<img align="left" src="https://raw.githubusercontent.com/greyhatguy007/Machine-Learning-Specialization-Coursera/d5c95ffb5b1524367e1a9147707ec4725298d715/C1%20-%20Supervised%20Machine%20Learning%20-%20Regression%20and%20Classification/week3/Optional%20Labs/images/C1_W3_LogisticLoss_a.png"     style=" width:250px; padding: 2px; " >
<img align="left" src="https://raw.githubusercontent.com/greyhatguy007/Machine-Learning-Specialization-Coursera/d5c95ffb5b1524367e1a9147707ec4725298d715/C1%20-%20Supervised%20Machine%20Learning%20-%20Regression%20and%20Classification/week3/Optional%20Labs/images/C1_W3_LogisticLoss_b.png"     style=" width:250px; padding: 2px; " >
<img align="left" src="https://raw.githubusercontent.com/greyhatguy007/Machine-Learning-Specialization-Coursera/d5c95ffb5b1524367e1a9147707ec4725298d715/C1%20-%20Supervised%20Machine%20Learning%20-%20Regression%20and%20Classification/week3/Optional%20Labs/images/C1_W3_LogisticLoss_c.png"     style=" width:250px; padding: 2px; " >

Logistic Regression는 target가 숫자가 아닌 0 또는 1인 분류 작업에 더 적합한 손실 함수를 사용합니다.

>**정의 참고:** 이 과정에서는 다음 정의가 사용됩니다. **손실**은 target 값에 대한 단일 예제의 차이를 측정한 반면 **비용**은 학습 세트에 대한 손실을 측정한 것입니다.


이는 다음과 같이 정의됩니다.
* $loss(f_{\mathbf{w},b}(\mathbf{x}^{(i)}), y^{(i)})$는 단일 데이터 포인트에 대한 비용입니다.

\begin{방정식} 손실(f_{\mathbf{w},b}(\mathbf{x}^{(i)}), y^{(i)}) = \begin{cases}
    - \log\left(f_{\mathbf{w},b}\left( \mathbf{x}^{(i)} \right) \right) & \text{$y^{(i)}=1$인 경우}\\
    - \log \left( 1 - f_{\mathbf{w},b}\left( \mathbf{x}^{(i)} \right) \right) & \text{if $y^{(i)}=0$} \end{cases} \end{방정식}


* $f_{\mathbf{w},b}(\mathbf{x}^{(i)})$는 모델의 예측값이고, $y^{(i)}$는 target 값입니다.

* $f_{\mathbf{w},b}(\mathbf{x}^{(i)}) = g(\mathbf{w} \cdot\mathbf{x}^{(i)}+b)$ 여기서 $g$ 함수는 sigmoid 함수입니다.

이 손실 함수의 feature를 정의하는 것은 두 개의 별도 곡선을 사용한다는 사실입니다. 하나는 target가 0이거나($y=0$)인 경우이고, 다른 하나는 target가 1인 경우($y=1$)입니다. 결합된 이러한 곡선은 손실 함수에 유용한 동작, 즉 예측값가 target과 일치할 때 0이 되고 예측값가 target과 다르면 값이 급격히 증가하는 동작을 제공합니다. 아래 곡선을 고려하십시오.

````python
plt_two_logistic_loss_curves()
````

결합된 곡선은 제곱 오류 손실의 2차 곡선과 유사합니다. 참고로 x축은 sigmoid의 출력인 $f_{\mathbf{w},b}$입니다. sigmoid 출력은 엄격하게 0과 1 사이입니다.

위의 손실 함수는 구현하기 쉽도록 다시 작성할 수 있습니다.
    $$loss(f_{\mathbf{w},b}(\mathbf{x}^{(i)}), y^{(i)}) = (-y^{(i)} \log\left(f_{\mathbf{w},b}\left( \mathbf{x}^{(i)} \right) \right) - \left( 1 - y^{(i)}\right) \log \left( 1 - f_{\mathbf{w},b}\left( \mathbf{x}^{(i)} \right) \right)$$
  
이것은 다소 무시무시해 보이는 방정식이다. $y^{(i)}$가 0과 1의 두 가지 값만 가질 수 있다고 생각하면 덜 어렵지 않습니다. 그런 다음 방정식을 두 부분으로 나누어 고려할 수 있습니다. $ y^{(i)} = 0$에서는 왼쪽 항이 제거됩니다.
$$
\begin{align}
loss(f_{\mathbf{w},b}(\mathbf{x}^{(i)}), 0) &= (-(0) \log\left(f_{\mathbf{w},b}\left( \mathbf{x}^{(i)} \right) \right) - \left( 1 - 0\right) \log \left( 1 - f_{\mathbf{w},b}\left( \mathbf{x}^{(i)} \right) \right) \\
&= -\log \left( 1 - f_{\mathbf{w},b}\left( \mathbf{x}^{(i)} \right) \right)
\end{align}
$$
$ y^{(i)} = 1$일 때 오른쪽 항은 제거됩니다.
$$
\begin{align}
  loss(f_{\mathbf{w},b}(\mathbf{x}^{(i)}), 1) &=  (-(1) \log\left(f_{\mathbf{w},b}\left( \mathbf{x}^{(i)} \right) \right) - \left( 1 - 1\right) \log \left( 1 - f_{\mathbf{w},b}\left( \mathbf{x}^{(i)} \right) \right)\\
  &=  -\log\left(f_{\mathbf{w},b}\left( \mathbf{x}^{(i)} \right) \right)
\end{align}
$$

좋습니다. 이 새로운 로지스틱 손실 함수를 사용하면 모든 예제의 손실을 통합하는 Cost Function을 생성할 수 있습니다. 이것이 다음 실습의 주제가 될 것입니다. 지금은 위에서 고려한 간단한 예에 대한 비용 대 매개변수 곡선을 살펴보겠습니다.

````python
plt.close('all')
cst = plt_logistic_cost(x_train,y_train)
````

이 곡선은 gradient descent에 매우 적합합니다! 정체기, 국소 최소치 또는 불연속성이 없습니다. 주의할 점은 제곱오차처럼 그릇이 아니라는 점입니다. 비용이 작을 때 곡선이 기울기를 가지며 계속해서 감소한다는 사실을 밝히기 위해 비용과 비용 로그를 모두 표시합니다. 알림: 마우스를 사용하여 위의 플롯을 회전할 수 있습니다.

## Cost function

이전 실습에서는 *로지스틱 손실* 함수를 개발했습니다. 기억하세요, 손실은 한 가지 예에 적용되도록 정의됩니다. 여기에서는 손실을 결합하여 모든 예가 포함된 **비용**을 형성합니다.


logistic regression의 경우 cost function의 형식은 다음과 같습니다.

$$ J(\mathbf{w},b) = \frac{1}{m} \sum_{i=0}^{m-1} \left[ loss(f_{\mathbf{w},b}(\mathbf{x}^{(i)}), y^{(i)}) \right] \tag{1}$$

어디서
* $loss(f_{\mathbf{w},b}(\mathbf{x}^{(i)}), y^{(i)})$는 단일 데이터 포인트에 대한 비용입니다.

    $$loss(f_{\mathbf{w},b}(\mathbf{x}^{(i)}), y^{(i)}) = -y^{(i)} \log\left(f_{\mathbf{w},b}\left( \mathbf{x}^{(i)} \right) \right) - \left( 1 - y^{(i)}\right) \log \left( 1 - f_{\mathbf{w},b}\left( \mathbf{x}^{(i)} \right) \right) \tag{2}$$
    
* 여기서 m은 데이터 세트의 학습 예제 수이며 다음과 같습니다.
$$
\begin{align}
  f_{\mathbf{w},b}(\mathbf{x^{(i)}}) &= g(z^{(i)})\tag{3} \\
  z^{(i)} &= \mathbf{w} \cdot \mathbf{x}^{(i)}+ b\tag{4} \\
  g(z^{(i)}) &= \frac{1}{1+e^{-z^{(i)}}}\tag{5}
\end{align}
$$

<a name='ex-02'></a>
#### 코드 설명

`compute_cost_logistic`의 알고리즘은 모든 예제를 반복하여 각 예제의 손실을 계산하고 총계를 누적합니다.

변수 X 및 y는 스칼라 값이 아니라 각각 ($m, n$) 및 ($𝑚$) 모양의 행렬입니다. 여기서 $𝑛$는 feature의 수이고 $𝑚$는 학습 예제의 수입니다.

### 참조 솔루션: 한 번에 한 줄씩

아래의 각 연습은 다음 코드 셀의 한 줄에 하나의 표현식으로 해결됩니다. 번호가 매겨진 참조는 해당 표현식을 코드 순서대로 설명합니다. 학생용 버전과 비교할 때 주변 루프와 반환 문을 유지하세요.

참고 표기: `m`는 예시 개수입니다. `i`는 예시를 선택합니다. `j`는 feature를 선택합니다. 예측값은 모델 출력을 의미합니다. 기울기는 매개변수가 변경될 때 비용이 어떻게 변경되는지를 의미합니다.

1. **sigmoid 이전 점수를 계산합니다**

$z_i=\sum_j X_{ij}w_j+b$. `np.dot`, `X[i]`, `w` 및 `b`를 사용합니다. sigmoid는 이미 다음 라인에 적용되어 있습니다.

2. **이 예의 Logistic Loss를 계산합니다**

$\ell_i=-y_i\log(p_i)-(1-y_i)\log(1-p_i)$. 여기서 `p_i`는 `f_wb_i`이고 `y_i`는 `y[i]`입니다. `np.log`를 사용하세요. 기존 `+=`는 이미 이 손실을 총계에 추가했습니다.

3. **평균 Logistic Loss 계산**

$J=\frac{1}{m}\sum_i\ell_i$. 누적된 `cost`를 `m`로 나눕니다.

**함수 참조:** [내적](https://numpy.org/doc/stable/reference/generated/numpy.dot.html), [열 평균 및 축](https://numpy.org/doc/stable/reference/generated/numpy.mean.html), [지수](https://numpy.org/doc/stable/reference/generated/numpy.exp.html). 아래 원본 실습에는 전체 수학적 맥락이 포함되어 있습니다.

출처: [C1_W3_Lab05_Cost_Function_Soln.ipynb](https://github.com/greyhatguy007/Machine-Learning-Specialization-Coursera/blob/d5c95ffb5b1524367e1a9147707ec4725298d715/C1%20-%20Supervised%20Machine%20Learning%20-%20Regression%20and%20Classification/week3/Optional%20Labs/C1_W3_Lab05_Cost_Function_Soln.ipynb)

````python
def compute_cost_logistic(X, y, w, b):
    """
    Computes cost

    Args:
      X (ndarray (m,n)): Data, m examples with n features
      y (ndarray (m,)) : target values
      w (ndarray (n,)) : model parameters
      b (scalar)       : model parameter

    Returns:
      cost (scalar): cost
    """

    m = X.shape[0]
    cost = 0.0
    for i in range(m):
        z_i = float(X[i] @ w) + b
        f_wb_i = sigmoid(z_i)
        cost +=  -(y[i] * np.log(f_wb_i) + (1.0 - y[i]) * np.log1p(-f_wb_i))

    cost = cost / float(m)
    return cost
````

````python
# Verify the reference implementation
np.testing.assert_allclose(compute_cost_logistic(np.array([[1., 2.], [3., 4.]]), np.array([0., 1.]), np.zeros(2), 0.), 0.6931471805599453)
print("Check passed.")
````

아래 셀을 사용하여 cost function의 구현을 확인하세요.

````python
w_tmp = np.array([1,1])
b_tmp = -3
print(compute_cost_logistic(X_train, y_train, w_tmp, b_tmp))
````

**예상 출력**: 0.3668667864055175

## 예
이제 $w$의 다른 값에 대한 cost function 출력이 무엇인지 살펴보겠습니다.

* 이전 실습에서는 $b = -3, w_0 = 1, w_1 = 1$에 대한 결정 경계를 그렸습니다. 즉, `b = -3, w = np.array([1,1])`가 있었습니다.

* $b = -4, w_0 = 1, w_1 = 1$ 또는 `b = -4, w = np.array([1,1])`가 더 나은 모델을 제공하는지 확인하고 싶다고 가정해 보겠습니다.

먼저 이 두 개의 서로 다른 $b$ 값에 대한 결정 경계를 그려 어느 값이 데이터에 더 잘 맞는지 살펴보겠습니다.

* $b = -3, w_0 = 1, w_1 = 1$의 경우 $-3 + x_0+x_1 = 0$를 플롯합니다(파란색으로 표시).
* $b = -4, w_0 = 1, w_1 = 1$의 경우 $-4 + x_0+x_1 = 0$를 플롯합니다(자홍색으로 표시됨)

### 참조 솔루션: 한 번에 한 줄씩

아래의 각 연습은 다음 코드 셀의 한 줄에 하나의 표현식으로 해결됩니다. 번호가 매겨진 참조는 해당 표현식을 코드 순서대로 설명합니다. 학생용 버전과 비교할 때 주변 루프와 반환 문을 유지하세요.

참고 표기: `m`는 예시 개수입니다. `i`는 예시를 선택합니다. `j`는 feature를 선택합니다. 예측값은 모델 출력을 의미합니다. 기울기는 매개변수가 변경될 때 비용이 어떻게 변경되는지를 의미합니다.

1. **첫 번째 결정 경계 계산**

$x_1$에 대해 $x_0+x_1=3$를 해결합니다. 첫 번째 좌표로 `x0`를 사용합니다.

2. **두 번째 결정 경계 계산**

$x_1$에 대해 $x_0+x_1=4$를 해결합니다. 첫 번째 좌표로 `x0`를 사용합니다.

**함수 참조:** [내적](https://numpy.org/doc/stable/reference/generated/numpy.dot.html), [열 평균 및 축](https://numpy.org/doc/stable/reference/generated/numpy.mean.html), [지수](https://numpy.org/doc/stable/reference/generated/numpy.exp.html). 아래 원본 실습에는 전체 수학적 맥락이 포함되어 있습니다.

출처: [C1_W3_Lab05_Cost_Function_Soln.ipynb](https://github.com/greyhatguy007/Machine-Learning-Specialization-Coursera/blob/d5c95ffb5b1524367e1a9147707ec4725298d715/C1%20-%20Supervised%20Machine%20Learning%20-%20Regression%20and%20Classification/week3/Optional%20Labs/C1_W3_Lab05_Cost_Function_Soln.ipynb)

이 플롯에서 `b = -4, w = np.array([1,1])`가 학습 데이터에 대해 더 나쁜 모델임을 알 수 있습니다. cost function 구현이 이를 반영하는지 살펴보겠습니다.

````python
w_array1 = np.array([1,1])
b_1 = -3
w_array2 = np.array([1,1])
b_2 = -4

print("Cost for b = -3 : ", compute_cost_logistic(X_train, y_train, w_array1, b_1))
print("Cost for b = -4 : ", compute_cost_logistic(X_train, y_train, w_array2, b_2))
````

## Logistic Gradient Descent
<img align="right" src="https://raw.githubusercontent.com/greyhatguy007/Machine-Learning-Specialization-Coursera/d5c95ffb5b1524367e1a9147707ec4725298d715/C1%20-%20Supervised%20Machine%20Learning%20-%20Regression%20and%20Classification/week3/Optional%20Labs/images/C1_W3_Logistic_gradient_descent.png"     style=" width:400px; padding: 10px; " >

gradient descent 알고리즘은 기울기 계산을 활용합니다.
$$\begin{align*}
&\text{repeat until convergence:} \; \lbrace \\
&  \; \; \;w_j = w_j -  \alpha \frac{\partial J(\mathbf{w},b)}{\partial w_j} \tag{1}  \; & \text{for j := 0..n-1} \\
&  \; \; \;  \; \;b = b -  \alpha \frac{\partial J(\mathbf{w},b)}{\partial b} \\
&\rbrace
\end{align*}$$

각 반복은 모든 $j$에 대해 $w_j$에서 동시 업데이트를 수행합니다.
$$\begin{align*}
\frac{\partial J(\mathbf{w},b)}{\partial w_j}  &= \frac{1}{m} \sum\limits_{i = 0}^{m-1} (f_{\mathbf{w},b}(\mathbf{x}^{(i)}) - y^{(i)})x_{j}^{(i)} \tag{2} \\
\frac{\partial J(\mathbf{w},b)}{\partial b}  &= \frac{1}{m} \sum\limits_{i = 0}^{m-1} (f_{\mathbf{w},b}(\mathbf{x}^{(i)}) - y^{(i)}) \tag{3}
\end{align*}$$

* m은 데이터 세트의 학습 예제 수입니다.      
* $f_{\mathbf{w},b}(x^{(i)})$는 해당 모델의 예측값이고, $y^{(i)}$는 target입니다.
* logistic regression 모델의 경우  
    $z = \mathbf{w} \cdot \mathbf{x} + b$  
    $f_{\mathbf{w},b}(x) = g(z)$  
    여기서 $g(z)$는 sigmoid 함수입니다.  
    $g(z) = \frac{1}{1+e^{-z}}$
    


### Gradient Descent 구현
gradient descent 알고리즘 구현에는 두 가지 구성 요소가 있습니다.
- 위의 방정식 (1)을 구현하는 루프. 이는 아래의 `gradient_descent`이며 일반적으로 Optional Lab과 연습 과제에서 제공됩니다.
- 위의 방정식(2,3)에 따라 현재 기울기를 계산합니다. 아래 `compute_gradient_logistic` 입니다. 이번 주 연습 과제에서 이 함수를 구현합니다.

#### 기울기 계산, 코드 설명
모든 $w_j$ 및 $b$에 대해 위의 방정식 (2), (3)을 구현합니다. 이를 구현하는 방법에는 여러 가지가 있습니다. 아래에 정리한 내용은 다음과 같습니다.
- `dj_dw` 및 `dj_db`를 누적하도록 변수를 초기화합니다.
- 각 예에 대해
    - 해당 예 $g(\mathbf{w} \cdot \mathbf{x}^{(i)} + b) - \mathbf{y}^{(i)}$에 대한 오류를 계산합니다.
    - 이 예의 각 입력 값 $x_{j}^{(i)}$에 대해  
        - 오류에 입력 $x_{j}^{(i)}$를 곱하고 `dj_dw`의 해당 요소에 추가합니다. (위의 방정식 2)
    - `dj_db`에 오류를 추가합니다(위의 방정식 3).

- `dj_db` 및 `dj_dw`를 총 예제 수(m)로 나눕니다.
- numpy `X[i,:]` 또는 `X[i]` 및 $x_{j}^{(i)}$의 $\mathbf{x}^{(i)}$는 `X[i,j]`입니다.

### 참조 솔루션: 한 번에 한 줄씩

아래의 각 연습은 다음 코드 셀의 한 줄에 하나의 표현식으로 해결됩니다. 번호가 매겨진 참조는 해당 표현식을 코드 순서대로 설명합니다. 학생용 버전과 비교할 때 주변 루프와 반환 문을 유지하세요.

참고 표기: `m`는 예시 개수입니다. `i`는 예시를 선택합니다. `j`는 feature를 선택합니다. 예측값은 모델 출력을 의미합니다. 기울기는 매개변수가 변경될 때 비용이 어떻게 변경되는지를 의미합니다.

1. **실제 라벨을 뺀 확률 계산**

$e_i=p_i-y_i$. `f_wb_i` 및 `y[i]`를 사용하십시오.

2. **이 가중치 기울기를 누적합니다**

새로운 $g_j$ = 기존 $g_j + e_i X_{ij}$. 오류 이름은 이전 줄에 `err` 또는 `err_i`로 표시됩니다. 누적 합계는 `dj_dw[j]`입니다.

3. **편향 기울기 누적**

새로운 $g_b$ = 기존 $g_b+e_i$. `dj_db`에 `err` 또는 `err_i`를 추가합니다.

4. **가중치 기울기 평균을 구합니다**

누적된 `dj_dw`를 예시 개수 `m`로 나눕니다. 배열의 나눗셈은 모든 가중치에 대해 동시에 작동합니다.

5. **편향 기울기 평균화**

누적된 `dj_db`를 예제 `m`의 개수로 나눕니다.

**함수 참조:** [내적](https://numpy.org/doc/stable/reference/generated/numpy.dot.html), [열 평균 및 축](https://numpy.org/doc/stable/reference/generated/numpy.mean.html), [지수](https://numpy.org/doc/stable/reference/generated/numpy.exp.html). 아래 원본 실습에는 전체 수학적 맥락이 포함되어 있습니다.

출처: [C1_W3_Lab06_Gradient_Descent_Soln.ipynb](https://github.com/greyhatguy007/Machine-Learning-Specialization-Coursera/blob/d5c95ffb5b1524367e1a9147707ec4725298d715/C1%20-%20Supervised%20Machine%20Learning%20-%20Regression%20and%20Classification/week3/Optional%20Labs/C1_W3_Lab06_Gradient_Descent_Soln.ipynb)

````python
def compute_gradient_logistic(X, y, w, b):
    """
    Computes the gradient for logistic regression

    Args:
      X (ndarray (m,n): Data, m examples with n features
      y (ndarray (m,)): target values
      w (ndarray (n,)): model parameters
      b (scalar)      : model parameter
    Returns
      dj_dw (ndarray (n,)): The gradient of the cost w.r.t. the parameters w.
      dj_db (scalar)      : The gradient of the cost w.r.t. the parameter b.
    """
    m,n = X.shape
    dj_dw = np.zeros((n,))                           #(n,)
    dj_db = 0.

    for i in range(m):
        f_wb_i = sigmoid(np.dot(X[i],w) + b)          #(n,)(n,)=scalar
        err_i  = f_wb_i - y[i]                       #scalar
        for j in range(n):
            dj_dw[j] = dj_dw[j] + X[i, j] * err_i      #scalar
        dj_db = dj_db + err_i
    dj_dw = dj_dw / float(m)                                   #(n,)
    dj_db = dj_db / float(m)                                   #scalar

    return dj_db, dj_dw
````

````python
# Verify the reference implementation
test_db, test_dw = compute_gradient_logistic(np.array([[1., 2.], [3., 4.]]), np.array([0., 1.]), np.zeros(2), 0.)
np.testing.assert_allclose(test_db, 0.)
np.testing.assert_allclose(test_dw, [-0.5, -0.5])
print("Check passed.")
````

아래 셀을 이용하여 기울기 함수의 구현을 확인해보세요.

````python
X_tmp = np.array([[0.5, 1.5], [1,1], [1.5, 0.5], [3, 0.5], [2, 2], [1, 2.5]])
y_tmp = np.array([0, 0, 0, 1, 1, 1])
w_tmp = np.array([2.,3.])
b_tmp = 1.
dj_db_tmp, dj_dw_tmp = compute_gradient_logistic(X_tmp, y_tmp, w_tmp, b_tmp)
print(f"dj_db: {dj_db_tmp}" )
print(f"dj_dw: {dj_dw_tmp.tolist()}" )
````

#### Gradient Descent 코드
위의 식 (1)을 구현한 코드는 아래와 같이 구현됩니다. 잠시 시간을 내어 루틴의 함수를 찾아 위의 방정식과 비교해 보세요.

### 참조 솔루션: 한 번에 한 줄씩

아래의 각 연습은 다음 코드 셀의 한 줄에 하나의 표현식으로 해결됩니다. 번호가 매겨진 참조는 해당 표현식을 코드 순서대로 설명합니다. 학생용 버전과 비교할 때 주변 루프와 반환 문을 유지하세요.

참고 표기: `m`는 예시 개수입니다. `i`는 예시를 선택합니다. `j`는 feature를 선택합니다. 예측값은 모델 출력을 의미합니다. 기울기는 매개변수가 변경될 때 비용이 어떻게 변경되는지를 의미합니다.

1. **가중치 업데이트**

$w_{new}=w-\alpha\nabla_w J$. `w`, `alpha` 및 `dj_dw`를 사용하십시오. 배열 산술은 모든 가중치를 업데이트합니다.

2. **편향 업데이트**

$b_{new}=b-\alpha\,\partial J/\partial b$. `b`, `alpha` 및 `dj_db`를 사용합니다.

**함수 참조:** [내적](https://numpy.org/doc/stable/reference/generated/numpy.dot.html), [열 평균 및 축](https://numpy.org/doc/stable/reference/generated/numpy.mean.html), [지수](https://numpy.org/doc/stable/reference/generated/numpy.exp.html). 아래 원본 실습에는 전체 수학적 맥락이 포함되어 있습니다.

출처: [C1_W3_Lab06_Gradient_Descent_Soln.ipynb](https://github.com/greyhatguy007/Machine-Learning-Specialization-Coursera/blob/d5c95ffb5b1524367e1a9147707ec4725298d715/C1%20-%20Supervised%20Machine%20Learning%20-%20Regression%20and%20Classification/week3/Optional%20Labs/C1_W3_Lab06_Gradient_Descent_Soln.ipynb)

````python
def gradient_descent(X, y, w_in, b_in, alpha, num_iters):
    """
    Performs batch gradient descent

    Args:
      X (ndarray (m,n)   : Data, m examples with n features
      y (ndarray (m,))   : target values
      w_in (ndarray (n,)): Initial values of model parameters
      b_in (scalar)      : Initial values of model parameter
      alpha (float)      : Learning rate
      num_iters (scalar) : number of iterations to run gradient descent

    Returns:
      w (ndarray (n,))   : Updated values of parameters
      b (scalar)         : Updated value of parameter
    """
    # An array to store cost J and w's at each iteration primarily for graphing later
    J_history = []
    w = copy.deepcopy(w_in)  #avoid modifying global w within function
    b = b_in

    for i in range(num_iters):
        # Calculate the gradient and update the parameters
        dj_db, dj_dw = compute_gradient_logistic(X, y, w, b)

        # Update Parameters using w, b, alpha and gradient
        w = w - dj_dw * alpha
        b = b - dj_db * alpha

        # Save cost J at each iteration
        if i<100000:      # prevent resource exhaustion
            J_history.append( compute_cost_logistic(X, y, w, b) )

        # Print cost every at intervals 10 times or as many iterations if < 10
        if i% math.ceil(num_iters / 10) == 0:
            print(f"Iteration {i:4d}: Cost {J_history[-1]}   ")

    return w, b, J_history         #return final w,b and J history for graphing
````

````python
# Verify the reference implementation
test_w, test_b, *_ = gradient_descent(np.array([[1., 2.], [3., 4.]]), np.array([0., 1.]), np.zeros(2), 0., 0.1, 1)
np.testing.assert_allclose(test_w, [0.05, 0.05])
np.testing.assert_allclose(test_b, 0.)
print("Check passed.")
````

데이터 세트에서 gradient descent를 실행해 보겠습니다.

````python
w_tmp  = np.zeros_like(X_train[0])
b_tmp  = 0.
alph = 0.1
iters = 10000

w_out, b_out, _ = gradient_descent(X_train, y_train, w_tmp, b_tmp, alph, iters)
print(f"\nupdated parameters: w:{w_out}, b:{b_out}")
````

### 참조 솔루션: 한 번에 한 줄씩

아래의 각 연습은 다음 코드 셀의 한 줄에 하나의 표현식으로 해결됩니다. 번호가 매겨진 참조는 해당 표현식을 코드 순서대로 설명합니다. 학생용 버전과 비교할 때 주변 루프와 반환 문을 유지하세요.

참고 표기: `m`는 예시 개수입니다. `i`는 예시를 선택합니다. `j`는 feature를 선택합니다. 예측값은 모델 출력을 의미합니다. 기울기는 매개변수가 변경될 때 비용이 어떻게 변경되는지를 의미합니다.

1. **경계가 첫 번째 축과 교차하는 위치 찾기**

$x_0=0$에서 $w_1x_1+b=0$를 푼다. `w_out[1]` 및 `b_out`를 사용하세요.

**함수 참조:** [내적](https://numpy.org/doc/stable/reference/generated/numpy.dot.html), [열 평균 및 축](https://numpy.org/doc/stable/reference/generated/numpy.mean.html), [지수](https://numpy.org/doc/stable/reference/generated/numpy.exp.html). 아래 원본 실습에는 전체 수학적 맥락이 포함되어 있습니다.

위 슬라이드는 선형 및 logistic regression 모두에 대한 비용 및 기울기 함수를 보여줍니다. 참고:
- 비용
    - 비용 함수는 선형과 logistic regression 간에 크게 다르지만 방정식에 Regularization을 추가하는 것은 동일합니다.
- 기울기
    - 선형 및 logistic regression의 기울기 함수는 매우 유사합니다. $f_{wb}$ 구현에서만 다릅니다.

## Regularization이 적용된 비용 함수
Regularization된 linear regression에 대한 ### Cost function

Regularization된 cost function linear regression의 방정식은 다음과 같습니다.
$$J(\mathbf{w},b) = \frac{1}{2m} \sum\limits_{i = 0}^{m-1} (f_{\mathbf{w},b}(\mathbf{x}^{(i)}) - y^{(i)})^2  + \frac{\lambda}{2m}  \sum_{j=0}^{n-1} w_j^2 \tag{1}$$
여기서:
$$ f_{\mathbf{w},b}(\mathbf{x}^{(i)}) = \mathbf{w} \cdot \mathbf{x}^{(i)} + b  \tag{2} $$


이를 Regularization이 없는 cost function(이전 실습에서 구현한)와 비교해 보세요. 형식은 다음과 같습니다.

$$J(\mathbf{w},b) = \frac{1}{2m} \sum\limits_{i = 0}^{m-1} (f_{\mathbf{w},b}(\mathbf{x}^{(i)}) - y^{(i)})^2 $$

차이점은 Regularization 용어인 <span style="color:blue">입니다.
    $\frac{\lambda}{2m}  \sum_{j=0}^{n-1} w_j^2$ </span>
    
이 용어를 포함하면 gradient descent가 매개변수의 크기를 최소화하도록 권장됩니다. 이 예에서는 매개변수 $b$가 Regularization되지 않았습니다. 이는 표준 관행입니다.

아래는 방정식 (1)과 (2)를 구현한 것입니다. 이는 모든 `m` 예제에 대해 *이 과정의 표준 패턴*인 `for loop`를 사용한다는 점에 유의하세요.

### 참조 솔루션: 한 번에 한 줄씩

아래의 각 연습은 다음 코드 셀의 한 줄에 하나의 표현식으로 해결됩니다. 번호가 매겨진 참조는 해당 표현식을 코드 순서대로 설명합니다. 학생용 버전과 비교할 때 주변 루프와 반환 문을 유지하세요.

참고 표기: `m`는 예시 개수입니다. `i`는 예시를 선택합니다. `j`는 feature를 선택합니다. 예측값은 모델 출력을 의미합니다. 기울기는 매개변수가 변경될 때 비용이 어떻게 변경되는지를 의미합니다.

1. **예측값을 계산합니다**

하나의 feature에 대해 $\hat y = wx+b$; 여러 feature의 경우 $\hat y = \sum_j x_j w_j+b$. `i`로 인덱싱된 현재 예제를 사용하세요.

2. **합계에 제곱오차를 더합니다**

새로운 합계 = 기존 `cost` + $(\hat y_i-y_i)^2$. 이전 줄에서 예측값 변수를 사용합니다.

3. **평균 비용 계산**

$J = \frac{1}{2m}\sum_i(\hat y_i-y_i)^2$. 누적된 오류를 `m`의 두 배로 나눕니다.

4. **이 제곱 가중치 페널티를 계산합니다**

$w_j^2$. `w[j]`를 사용하세요. 기존 `+=`는 `reg_cost`에 추가되었습니다. 이전 합계를 다시 추가하지 마세요.

5. **총 가중치 페널티 규모 조정**

$R=\frac{\lambda}{2m}\sum_j w_j^2$. 합계는 이미 `reg_cost`에 있습니다. `lambda_` 및 `m`를 사용하여 크기를 조정하세요.

6. **예측값 비용과 가중치 패널티 결합**

$J_{total}=J+R$. `cost`와 `reg_cost`를 결합합니다. 편향에는 이 페널티를 적용하지 않습니다.

**함수 참조:** [내적](https://numpy.org/doc/stable/reference/generated/numpy.dot.html), [열 평균 및 축](https://numpy.org/doc/stable/reference/generated/numpy.mean.html), [지수](https://numpy.org/doc/stable/reference/generated/numpy.exp.html). 아래 원본 실습에는 전체 수학적 맥락이 포함되어 있습니다.

출처: [C1_W3_Lab09_Regularization_Soln.ipynb](https://github.com/greyhatguy007/Machine-Learning-Specialization-Coursera/blob/d5c95ffb5b1524367e1a9147707ec4725298d715/C1%20-%20Supervised%20Machine%20Learning%20-%20Regression%20and%20Classification/week3/Optional%20Labs/C1_W3_Lab09_Regularization_Soln.ipynb)

````python
def compute_cost_linear_reg(X, y, w, b, lambda_ = 1):
    """
    Computes the cost over all examples
    Args:
      X (ndarray (m,n): Data, m examples with n features
      y (ndarray (m,)): target values
      w (ndarray (n,)): model parameters
      b (scalar)      : model parameter
      lambda_ (scalar): Controls amount of regularization
    Returns:
      total_cost (scalar):  cost
    """

    m  = X.shape[0]
    n  = len(w)
    cost = 0.
    for i in range(m):
        f_wb_i = float(X[i] @ w) + b                                   #(n,)(n,)=scalar, see np.dot
        cost = cost + np.square(y[i] - f_wb_i)                               #scalar
    cost = cost / (2.0 * m)                                              #scalar

    reg_cost = 0
    for j in range(n):
        reg_cost += (float(w[j]) ** 2)                                          #scalar
    reg_cost = reg_cost * lambda_ / (2.0 * m)                              #scalar

    total_cost = reg_cost + cost                                       #scalar
    return total_cost                                                  #scalar
````

````python
# Verify the reference implementation
np.testing.assert_allclose(compute_cost_linear_reg(np.array([[1., 2.], [3., 4.]]), np.array([0., 1.]), np.array([0.2, -0.3]), 0.4, 2.), 0.4249999999999998, atol=1e-10)
print("Check passed.")
````

아래 셀을 실행하여 실제로 작동하는 모습을 확인하세요.

````python
np.random.seed(1)
X_tmp = np.random.rand(5,6)
y_tmp = np.array([0,1,0,1,0])
w_tmp = np.random.rand(X_tmp.shape[1]).reshape(-1,)-0.5
b_tmp = 0.5
lambda_tmp = 0.7
cost_tmp = compute_cost_linear_reg(X_tmp, y_tmp, w_tmp, b_tmp, lambda_tmp)

print("Regularized cost:", cost_tmp)
````

Regularization된 logistic regression에 대한 ### Cost function
Regularization된 **로지스틱** 회귀의 경우 Cost Function은 다음과 같은 형식입니다.
$$J(\mathbf{w},b) = \frac{1}{m}  \sum_{i=0}^{m-1} \left[ -y^{(i)} \log\left(f_{\mathbf{w},b}\left( \mathbf{x}^{(i)} \right) \right) - \left( 1 - y^{(i)}\right) \log \left( 1 - f_{\mathbf{w},b}\left( \mathbf{x}^{(i)} \right) \right) \right] + \frac{\lambda}{2m}  \sum_{j=0}^{n-1} w_j^2 \tag{3}$$
여기서:
$$ f_{\mathbf{w},b}(\mathbf{x}^{(i)}) = sigmoid(\mathbf{w} \cdot \mathbf{x}^{(i)} + b)  \tag{4} $$

이를 Regularization이 없는 cost function(이전 실습에서 구현한)와 비교해 보세요.

$$ J(\mathbf{w},b) = \frac{1}{m}\sum_{i=0}^{m-1} \left[ (-y^{(i)} \log\left(f_{\mathbf{w},b}\left( \mathbf{x}^{(i)} \right) \right) - \left( 1 - y^{(i)}\right) \log \left( 1 - f_{\mathbf{w},b}\left( \mathbf{x}^{(i)} \right) \right)\right] $$

위의 linear regression의 경우와 마찬가지로 Regularization 항인 <span style="color:blue">가 차이점입니다.
    $\frac{\lambda}{2m}  \sum_{j=0}^{n-1} w_j^2$ </span>

이 용어를 포함하면 gradient descent가 매개변수의 크기를 최소화하도록 권장됩니다. 이 예에서는 매개변수 $b$가 Regularization되지 않았습니다. 이는 표준 관행입니다.

### 참조 솔루션: 한 번에 한 줄씩

아래의 각 연습은 다음 코드 셀의 한 줄에 하나의 표현식으로 해결됩니다. 번호가 매겨진 참조는 해당 표현식을 코드 순서대로 설명합니다. 학생용 버전과 비교할 때 주변 루프와 반환 문을 유지하세요.

참고 표기: `m`는 예시 개수입니다. `i`는 예시를 선택합니다. `j`는 feature를 선택합니다. 예측값은 모델 출력을 의미합니다. 기울기는 매개변수가 변경될 때 비용이 어떻게 변경되는지를 의미합니다.

1. **sigmoid 이전 점수를 계산합니다**

$z_i=\sum_j X_{ij}w_j+b$. `np.dot`, `X[i]`, `w` 및 `b`를 사용합니다. sigmoid는 이미 다음 라인에 적용되어 있습니다.

2. **이 예의 Logistic Loss를 계산합니다**

$\ell_i=-y_i\log(p_i)-(1-y_i)\log(1-p_i)$. 여기서 `p_i`는 `f_wb_i`이고 `y_i`는 `y[i]`입니다. `np.log`를 사용하세요. 기존 `+=`는 이미 이 손실을 총계에 추가했습니다.

3. **평균 Logistic Loss 계산**

$J=\frac{1}{m}\sum_i\ell_i$. 누적된 `cost`를 `m`로 나눕니다.

4. **이 제곱 가중치 페널티를 계산합니다**

$w_j^2$. `w[j]`를 사용하세요. 기존 `+=`는 `reg_cost`에 추가되었습니다. 이전 합계를 다시 추가하지 마세요.

5. **총 가중치 페널티 규모 조정**

$R=\frac{\lambda}{2m}\sum_j w_j^2$. 합계는 이미 `reg_cost`에 있습니다. `lambda_` 및 `m`를 사용하여 크기를 조정하세요.

6. **예측값 비용과 가중치 패널티 결합**

$J_{total}=J+R$. `cost`와 `reg_cost`를 결합합니다. 편향에는 이 페널티를 적용하지 않습니다.

**함수 참조:** [내적](https://numpy.org/doc/stable/reference/generated/numpy.dot.html), [열 평균 및 축](https://numpy.org/doc/stable/reference/generated/numpy.mean.html), [지수](https://numpy.org/doc/stable/reference/generated/numpy.exp.html). 아래 원본 실습에는 전체 수학적 맥락이 포함되어 있습니다.

출처: [C1_W3_Lab09_Regularization_Soln.ipynb](https://github.com/greyhatguy007/Machine-Learning-Specialization-Coursera/blob/d5c95ffb5b1524367e1a9147707ec4725298d715/C1%20-%20Supervised%20Machine%20Learning%20-%20Regression%20and%20Classification/week3/Optional%20Labs/C1_W3_Lab09_Regularization_Soln.ipynb)

````python
def compute_cost_logistic_reg(X, y, w, b, lambda_ = 1):
    """
    Computes the cost over all examples
    Args:
    Args:
      X (ndarray (m,n): Data, m examples with n features
      y (ndarray (m,)): target values
      w (ndarray (n,)): model parameters
      b (scalar)      : model parameter
      lambda_ (scalar): Controls amount of regularization
    Returns:
      total_cost (scalar):  cost
    """

    m,n  = X.shape
    cost = 0.
    for i in range(m):
        z_i = float(X[i] @ w) + b                                      #(n,)(n,)=scalar, see np.dot
        f_wb_i = sigmoid(z_i)                                          #scalar
        cost +=  -(y[i] * np.log(f_wb_i) + (1.0 - y[i]) * np.log1p(-f_wb_i))      #scalar

    cost = cost / float(m)                                                      #scalar

    reg_cost = 0
    for j in range(n):
        reg_cost += (float(w[j]) ** 2)                                          #scalar
    reg_cost = reg_cost * lambda_ / (2.0 * m)                              #scalar

    total_cost = reg_cost + cost                                       #scalar
    return total_cost                                                  #scalar
````

````python
# Verify the reference implementation
np.testing.assert_allclose(compute_cost_logistic_reg(np.array([[1., 2.], [3., 4.]]), np.array([0., 1.]), np.array([0.2, -0.3]), 0.4, 2.), 0.8106430249707686, atol=1e-10)
print("Check passed.")
````

아래 셀을 실행하여 실제로 작동하는 모습을 확인하세요.

````python
np.random.seed(1)
X_tmp = np.random.rand(5,6)
y_tmp = np.array([0,1,0,1,0])
w_tmp = np.random.rand(X_tmp.shape[1]).reshape(-1,)-0.5
b_tmp = 0.5
lambda_tmp = 0.7
cost_tmp = compute_cost_logistic_reg(X_tmp, y_tmp, w_tmp, b_tmp, lambda_tmp)

print("Regularized cost:", cost_tmp)
````

## Regularization을 사용한 Gradient descent
gradient descent를 실행하기 위한 기본 알고리즘은 Regularization에 따라 변경되지 않습니다.
$$\begin{align*}
&\text{repeat until convergence:} \; \lbrace \\
&  \; \; \;w_j = w_j -  \alpha \frac{\partial J(\mathbf{w},b)}{\partial w_j} \tag{1}  \; & \text{for j := 0..n-1} \\
&  \; \; \;  \; \;b = b -  \alpha \frac{\partial J(\mathbf{w},b)}{\partial b} \\
&\rbrace
\end{align*}$$ 여기서 각 반복은 모든 $j$에 대해 $w_j$에서 동시 업데이트를 수행합니다.

Regularization으로 인해 변경되는 사항은 기울기를 계산하는 것입니다.

### Regularization으로 기울기 계산(선형/로지스틱 모두)
선형 및 logistic regression의 기울기 계산은 거의 동일하며 $f_{\mathbf{w}b}$의 계산만 다릅니다.
$$\begin{align*}
\frac{\partial J(\mathbf{w},b)}{\partial w_j}  &= \frac{1}{m} \sum\limits_{i = 0}^{m-1} (f_{\mathbf{w},b}(\mathbf{x}^{(i)}) - y^{(i)})x_{j}^{(i)}  +  \frac{\lambda}{m} w_j \tag{2} \\
\frac{\partial J(\mathbf{w},b)}{\partial b}  &= \frac{1}{m} \sum\limits_{i = 0}^{m-1} (f_{\mathbf{w},b}(\mathbf{x}^{(i)}) - y^{(i)}) \tag{3}
\end{align*}$$

* m은 데이터 세트의 학습 예제 수입니다.      
* $f_{\mathbf{w},b}(x^{(i)})$는 해당 모델의 예측값이고, $y^{(i)}$는 target입니다.

      
* <span style="color:blue"> **선형** </span> 회귀 모델의 경우  
    $f_{\mathbf{w},b}(x) = \mathbf{w} \cdot \mathbf{x} + b$  
* <span style="color:blue"> **로지스틱** </span> 회귀 모델의 경우  
    $z = \mathbf{w} \cdot \mathbf{x} + b$  
    $f_{\mathbf{w},b}(x) = g(z)$  
    여기서 $g(z)$는 sigmoid 함수입니다.  
    $g(z) = \frac{1}{1+e^{-z}}$
    
Regularization을 추가하는 용어는 <span style="color:blue">$\frac{\lambda}{m} w_j $</span>입니다.

### Regularization된 linear regression에 대한 기울기 함수

### 참조 솔루션: 한 번에 한 줄씩

아래의 각 연습은 다음 코드 셀의 한 줄에 하나의 표현식으로 해결됩니다. 번호가 매겨진 참조는 해당 표현식을 코드 순서대로 설명합니다. 학생용 버전과 비교할 때 주변 루프와 반환 문을 유지하세요.

참고 표기: `m`는 예시 개수입니다. `i`는 예시를 선택합니다. `j`는 feature를 선택합니다. 예측값은 모델 출력을 의미합니다. 기울기는 매개변수가 변경될 때 비용이 어떻게 변경되는지를 의미합니다.

1. **예측값에서 실제 값을 뺀 값을 계산합니다**

$e_i=(\sum_j X_{ij}w_j+b)-y_i$. 현재 행 `X[i]`, 가중치 `w`, 편향 `b` 및 레이블 `y[i]`를 사용합니다.

2. **이 가중치 기울기를 누적합니다**

새로운 $g_j$ = 기존 $g_j + e_i X_{ij}$. 오류 이름은 이전 줄에 `err` 또는 `err_i`로 표시됩니다. 누적 합계는 `dj_dw[j]`입니다.

3. **편향 기울기 누적**

새로운 $g_b$ = 기존 $g_b+e_i$. `dj_db`에 `err` 또는 `err_i`를 추가합니다.

4. **가중치 기울기 평균을 구합니다**

누적된 `dj_dw`를 예시 개수 `m`로 나눕니다. 배열의 나눗셈은 모든 가중치에 대해 동시에 작동합니다.

5. **편향 기울기 평균화**

누적된 `dj_db`를 예제 `m`의 개수로 나눕니다.

6. **가중치 페널티 기울기 추가**

$g_j^{regularized}=g_j+\frac{\lambda}{m}w_j$. `dj_dw[j]`, `lambda_`, `m` 및 `w[j]`를 사용합니다.

**함수 참조:** [내적](https://numpy.org/doc/stable/reference/generated/numpy.dot.html), [열 평균 및 축](https://numpy.org/doc/stable/reference/generated/numpy.mean.html), [지수](https://numpy.org/doc/stable/reference/generated/numpy.exp.html). 아래 원본 실습에는 전체 수학적 맥락이 포함되어 있습니다.

출처: [C1_W3_Lab09_Regularization_Soln.ipynb](https://github.com/greyhatguy007/Machine-Learning-Specialization-Coursera/blob/d5c95ffb5b1524367e1a9147707ec4725298d715/C1%20-%20Supervised%20Machine%20Learning%20-%20Regression%20and%20Classification/week3/Optional%20Labs/C1_W3_Lab09_Regularization_Soln.ipynb)

````python
def compute_gradient_linear_reg(X, y, w, b, lambda_):
    """
    Computes the gradient for linear regression
    Args:
      X (ndarray (m,n): Data, m examples with n features
      y (ndarray (m,)): target values
      w (ndarray (n,)): model parameters
      b (scalar)      : model parameter
      lambda_ (scalar): Controls amount of regularization

    Returns:
      dj_dw (ndarray (n,)): The gradient of the cost w.r.t. the parameters w.
      dj_db (scalar):       The gradient of the cost w.r.t. the parameter b.
    """
    m,n = X.shape           #(number of examples, number of features)
    dj_dw = np.zeros((n,))
    dj_db = 0.

    for i in range(m):
        err = float(X[i] @ w) + b - y[i]
        for j in range(n):
            dj_dw[j] = dj_dw[j] + X[i, j] * err
        dj_db = dj_db + err
    dj_dw = dj_dw / float(m)
    dj_db = dj_db / float(m)

    for j in range(n):
        dj_dw[j] = dj_dw[j] + lambda_ * w[j] / float(m)

    return dj_db, dj_dw
````

````python
# Verify the reference implementation
test_db, test_dw = compute_gradient_linear_reg(np.array([[1., 2.], [3., 4.]]), np.array([0., 1.]), np.array([0.2, -0.3]), 0.4, 2.)
np.testing.assert_allclose(test_db, -0.5999999999999999, atol=1e-10)
np.testing.assert_allclose(test_dw, [-1.5999999999999996, -2.6999999999999993], atol=1e-10)
print("Check passed.")
````

아래 셀을 실행하여 실제로 작동하는 모습을 확인하세요.

````python
np.random.seed(1)
X_tmp = np.random.rand(5,3)
y_tmp = np.array([0,1,0,1,0])
w_tmp = np.random.rand(X_tmp.shape[1])
b_tmp = 0.5
lambda_tmp = 0.7
dj_db_tmp, dj_dw_tmp =  compute_gradient_linear_reg(X_tmp, y_tmp, w_tmp, b_tmp, lambda_tmp)

print(f"dj_db: {dj_db_tmp}", )
print(f"Regularized dj_dw:\n {dj_dw_tmp.tolist()}", )
````

### Regularization된 logistic regression에 대한 기울기 함수

### 참조 솔루션: 한 번에 한 줄씩

아래의 각 연습은 다음 코드 셀의 한 줄에 하나의 표현식으로 해결됩니다. 번호가 매겨진 참조는 해당 표현식을 코드 순서대로 설명합니다. 학생용 버전과 비교할 때 주변 루프와 반환 문을 유지하세요.

참고 표기: `m`는 예시 개수입니다. `i`는 예시를 선택합니다. `j`는 feature를 선택합니다. 예측값은 모델 출력을 의미합니다. 기울기는 매개변수가 변경될 때 비용이 어떻게 변경되는지를 의미합니다.

1. **실제 라벨을 뺀 확률 계산**

$e_i=p_i-y_i$. `f_wb_i` 및 `y[i]`를 사용하십시오.

2. **이 가중치 기울기를 누적합니다**

새로운 $g_j$ = 기존 $g_j + e_i X_{ij}$. 오류 이름은 이전 줄에 `err` 또는 `err_i`로 표시됩니다. 누적 합계는 `dj_dw[j]`입니다.

3. **편향 기울기 누적**

새로운 $g_b$ = 기존 $g_b+e_i$. `dj_db`에 `err` 또는 `err_i`를 추가합니다.

4. **가중치 기울기 평균을 구합니다**

누적된 `dj_dw`를 예시 개수 `m`로 나눕니다. 배열의 나눗셈은 모든 가중치에 대해 동시에 작동합니다.

5. **편향 기울기 평균화**

누적된 `dj_db`를 예제 `m`의 개수로 나눕니다.

6. **가중치 페널티 기울기 추가**

$g_j^{regularized}=g_j+\frac{\lambda}{m}w_j$. `dj_dw[j]`, `lambda_`, `m` 및 `w[j]`를 사용합니다.

**함수 참조:** [내적](https://numpy.org/doc/stable/reference/generated/numpy.dot.html), [열 평균 및 축](https://numpy.org/doc/stable/reference/generated/numpy.mean.html), [지수](https://numpy.org/doc/stable/reference/generated/numpy.exp.html). 아래 원본 실습에는 전체 수학적 맥락이 포함되어 있습니다.

출처: [C1_W3_Lab09_Regularization_Soln.ipynb](https://github.com/greyhatguy007/Machine-Learning-Specialization-Coursera/blob/d5c95ffb5b1524367e1a9147707ec4725298d715/C1%20-%20Supervised%20Machine%20Learning%20-%20Regression%20and%20Classification/week3/Optional%20Labs/C1_W3_Lab09_Regularization_Soln.ipynb)

````python
def compute_gradient_logistic_reg(X, y, w, b, lambda_):
    """
    Computes the gradient for linear regression

    Args:
      X (ndarray (m,n): Data, m examples with n features
      y (ndarray (m,)): target values
      w (ndarray (n,)): model parameters
      b (scalar)      : model parameter
      lambda_ (scalar): Controls amount of regularization
    Returns
      dj_dw (ndarray Shape (n,)): The gradient of the cost w.r.t. the parameters w.
      dj_db (scalar)            : The gradient of the cost w.r.t. the parameter b.
    """
    m,n = X.shape
    dj_dw = np.zeros((n,))                            #(n,)
    dj_db = 0.0                                       #scalar

    for i in range(m):
        f_wb_i = sigmoid(np.dot(X[i],w) + b)          #(n,)(n,)=scalar
        err_i  = f_wb_i - y[i]                       #scalar
        for j in range(n):
            dj_dw[j] = dj_dw[j] + X[i, j] * err_i      #scalar
        dj_db = dj_db + err_i
    dj_dw = dj_dw / float(m)                                   #(n,)
    dj_db = dj_db / float(m)                                   #scalar

    for j in range(n):
        dj_dw[j] = dj_dw[j] + lambda_ * w[j] / float(m)

    return dj_db, dj_dw
````

````python
# Verify the reference implementation
test_db, test_dw = compute_gradient_logistic_reg(np.array([[1., 2.], [3., 4.]]), np.array([0., 1.]), np.array([0.2, -0.3]), 0.4, 2.)
np.testing.assert_allclose(test_db, -0.02491699865623892, atol=1e-10)
np.testing.assert_allclose(test_dw, [-0.3747509959687167, -0.8996679946249557], atol=1e-10)
print("Check passed.")
````

아래 셀을 실행하여 실제로 작동하는 모습을 확인하세요.

````python
np.random.seed(1)
X_tmp = np.random.rand(5,3)
y_tmp = np.array([0,1,0,1,0])
w_tmp = np.random.rand(X_tmp.shape[1])
b_tmp = 0.5
lambda_tmp = 0.7
dj_db_tmp, dj_dw_tmp =  compute_gradient_logistic_reg(X_tmp, y_tmp, w_tmp, b_tmp, lambda_tmp)

print(f"dj_db: {dj_db_tmp}", )
print(f"Regularized dj_dw:\n {dj_dw_tmp.tolist()}", )
````
